# Teacher Model — Notebook Overview

This notebook trains and evaluates the **teacher model** of the teacher-student knowledge
distillation pipeline for the 10-participant multimodal VR emotion study.

---

## Main Purpose

The teacher is a **signal-rich** model: it consumes the full 70-channel facial Action Unit (AU)
stream available from the Meta Quest Pro face tracker. Its role is twofold:

1. Establish the upper bound of emotion-recognition performance achievable from facial AUs alone.
2. Produce **soft labels** (class-probability vectors) that are exported and consumed by the
   student notebook for RQ2 knowledge distillation.

The student model, by contrast, is signal-sparse (eye gaze, head pose, upper-face AUs only) and is
trained in a separate notebook.

---

## Model

A **Mixture-of-Experts Temporal Convolutional Network (MoE-TCN)**:

- 3 TCN experts (natural / acted / shared), each a stack of dilated causal residual blocks
- A Top-2 gate with temperature annealing, load-balancing loss, and entropy regularisation
- Input shape `(SEQ_LENGTH=5, 70)` — 0.5 s of AU signal at 10 Hz
- Output: 7-class Ekman emotion softmax

---

## Notebook Structure

### Part 1: Setup and Environment
Google Drive mount, dependency installation, imports, and global random seeds.

### Part 2: Self-Report Validity Audit
Compares each participant's free-text self-report against the emotion that was actually elicited,
and produces the exclusion lists used later to filter unreliable recordings.

### Part 3: Labels, Emotion Classes, and the AU Feature Set
Emotion enum, data-source enum, label encoder, and the canonical 70-AU channel list.

### Part 4: Paths and Data-Loading Helpers
Output directories plus the windowing, thresholding, and participant-exclusive splitting helpers.

### Part 5: Dataset Construction
Loads the three data sources (our natural recordings, our acted recordings, EmojiHeroVR), applies the
self-report filter, builds participant-exclusive 70/10/20 splits, and stacks everything into tensors.

### Part 6: Training and Evaluation Helpers
Scoring and training wrappers shared by all teacher variants.

### Part 7: Teacher Model — MoE-TCN
Architecture definition, hyperparameters, training run, and result export.

### Part 8: Fine-Tuning Variant
Loads the original teacher and fine-tunes it on the new participants only, as an alternative to
joint retraining.

### Part 9: Teacher Variant Comparisons
Three-way and five-way comparisons across all teacher variants and all test sets.

### Part 10: Teacher Soft Labels for Distillation
Exports the soft labels and hard labels consumed by the student notebook (RQ2).

### Part 11: Original vs. +10-Participant Teacher (head-to-head)
Full metric suite, confusion matrices, per-class F1 plots, and reproducible raw predictions.

# Part 2: Self-Report Validity Audit

Participants reported, in a Google Form, which emotion they felt after each video. This section
compares those free-text self-reports against the emotion that the stimulus was designed to elicit.

Recordings where the intended emotion never appears in the self-report are treated as invalid
supervision and are excluded downstream (see Section 5.5). This is a data-validity step, not a
modelling step.

## 2.1 Self-Report vs. Elicited Emotion

Maps each participant folder to their Google Form submission, reads the per-video emotion order from
`emotion_flow.txt`, and parses the free-text self-report, keeping only the seven Ekman labels.

The cell prints a participant x emotion match grid and writes two files:

- `all_pairs.csv` — every (participant, video, intended emotion, reported emotion) pair
- `records_to_remove.csv` — the pairs where the intended emotion was never reported

The exclusion lists in Section 5.5 are transcribed from this output.

````python


NEW_DATA_ROOT = Path("/content/drive/MyDrive/Data Collection (New)")
#NEW_DATA_ROOT = Path("/content/drive/MyDrive/Multi-modal Detection for Meta Quest 3/Data Collection (New)")
FORMS_PATH    = "/content/drive/MyDrive/Forms/Forms-15_participants.csv"
#FORMS_PATH = "/content/drive/MyDrive/Multi-modal Detection for Meta Quest 3/Forms/Forms-15_participants.csv"

PARTICIPANT_TO_FORM_TIMESTAMP = {
    'P35': '6/3/2026 10:00:43',
    'P36': '6/3/2026 12:26:47',
    'P37': '6/4/2026 11:44:02',
    'P38': '6/5/2026 11:25:10',
    'P39': '6/5/2026 12:45:23',
    'P40': '6/10/2026 12:53:50',
    'P41': '6/10/2026 13:15:20',
    'P42': '6/11/2026 11:02:32',
    'P43': '6/11/2026 13:11:16',
    'P44': '6/16/2026 13:49:57',
    'P45': '7/01/2026 14:11:49',
    'P46': '7/02/2026 12:06:41',
    'P47': '7/15/2026 10:45:26',
    'P48': '7/17/2026 15:02:24',
    'P49': '7/24/2026 15:36:10',
}

TOLERANCE_SEC = 5  # CSV timestamps drift a second or two from the mapping

forms = pd.read_csv(FORMS_PATH)

# CSV side: strip the "AST" tz suffix, then parse
csv_raw = forms['Timestamp'].astype(str).str.replace(r'\s*[A-Z]{3}\s*$', '', regex=True)
csv_dt  = pd.to_datetime(csv_raw, errors='coerce')  # handles "2026/06/03 10:00:42 AM"

vid_root = NEW_DATA_ROOT / "VideoRecordings"

header = f"{'Participant':<18} {'folder':<8} {'flow.txt':<10} {'form_ts':<9} {'matched CSV time'}"
print(header)
print("-" * len(header))

missing = []
for p, form_ts in PARTICIPANT_TO_FORM_TIMESTAMP.items():
    folder_ok = (vid_root / p).is_dir()
    flow_ok   = (vid_root / p / "emotion_flow.txt").is_file()

    target = pd.to_datetime(form_ts, errors='coerce')  # mapping side, dayfirst not needed (M/D/Y)
    if pd.isna(target):
        ts_ok, matched_str = False, "(bad mapping ts)"
    else:
        deltas = (csv_dt - target).abs()
        nearest = deltas.min()
        ts_ok = pd.notna(nearest) and nearest.total_seconds() <= TOLERANCE_SEC
        matched_str = str(csv_dt[deltas.idxmin()]) if ts_ok else f"(nearest off by {nearest.total_seconds():.0f}s)"

    print(f"{p:<18} {'✓' if folder_ok else '✗':<8} "
          f"{'✓' if flow_ok else '✗':<10} {'✓' if ts_ok else '✗':<9} {matched_str}")

    if not (folder_ok and flow_ok and ts_ok):
        missing.append(p)

print("-" * len(header))
print(f"{len(PARTICIPANT_TO_FORM_TIMESTAMP)} checked, "
      f"{len(PARTICIPANT_TO_FORM_TIMESTAMP) - len(missing)} fully present, {len(missing)} with issues")
if not missing:
    print(f"\nAll {len(PARTICIPANT_TO_FORM_TIMESTAMP)} participants verified — safe to proceed.")
````

````python
import pandas as pd
import re
from pathlib import Path

# ── Participant → Form timestamp mapping (user-verified) ─────────────────────
PARTICIPANT_TO_FORM_TIMESTAMP = {
    'P35': '6/3/2026 10:00:43',
    'P36': '6/3/2026 12:26:47',
    'P37': '6/4/2026 11:44:02',
    'P38': '6/5/2026 11:25:10',
    'P39': '6/5/2026 12:45:23',
    'P40': '6/10/2026 12:53:50',
    'P41': '6/10/2026 13:15:20',
    'P42': '6/11/2026 11:02:32',
    'P43': '6/11/2026 13:11:16',
    'P44': '6/16/2026 13:49:57',
    'P45': '7/01/2026 14:11:49',
    'P46': '7/02/2026 12:06:41',
    'P47': '7/15/2026 10:45:26',
    'P48': '7/17/2026 15:02:24',
    'P49': '7/24/2026 15:36:10',
}

# ── Only 7 Ekman emotions — no aliases, no mapping of non-Ekman labels ────────
EKMAN = {e.lower(): e for e in ['Anger', 'Disgust', 'Fear', 'Happiness', 'Neutral', 'Sadness', 'Surprise']}

def parse_reported(raw):
    """Extract ALL Ekman emotions mentioned in free-text self-report."""
    if pd.isna(raw) or str(raw).strip().lower() in ('n/a', 'n/a video failed', ''):
        return []
    text = str(raw).lower()
    return list({v for k, v in EKMAN.items() if k in text})

# ── Load emotion_flow files ───────────────────────────────────────────────────
NEW_DATA_ROOT = Path("/content/drive/MyDrive/Data Collection (New)")
FORMS_PATH    = "/content/drive/MyDrive/Forms/Forms-15_participants.csv"

flow_data = {}
for flow_file in sorted((NEW_DATA_ROOT / "VideoRecordings").glob("*/emotion_flow.txt")):
    participant = flow_file.parent.name
    lines = flow_file.read_text(encoding='utf-8-sig').strip().splitlines()
    video_map = {}
    for line in lines[1:]:
        m = re.match(r'(\d+)\. Video - (\w+) -', line.strip())
        if m:
            video_map[int(m.group(1))] = m.group(2)
    flow_data[participant] = video_map

# ── Load Forms.csv ────────────────────────────────────────────────────────────
forms    = pd.read_csv(FORMS_PATH)
emo_cols = [c for c in forms.columns if c.startswith('What emotion')]

# CSV timestamps look like '2026/06/03 10:00:42 AM AST' — strip tz, parse to datetime
_csv_dt = pd.to_datetime(
    forms['Timestamp'].astype(str).str.replace(r'\s*[A-Z]{3}\s*$', '', regex=True),
    errors='coerce',
)
TOLERANCE_SEC = 5  # mapping timestamps drift a second or two from the CSV

def find_form_idx(form_ts):
    target = pd.to_datetime(form_ts, errors='coerce')
    if pd.isna(target):
        return None
    deltas = (_csv_dt - target).abs()
    if pd.isna(deltas.min()) or deltas.min().total_seconds() > TOLERANCE_SEC:
        return None
    return int(deltas.idxmin())

# ── Build results ─────────────────────────────────────────────────────────────
results = []
for participant, form_ts in PARTICIPANT_TO_FORM_TIMESTAMP.items():
    form_idx = find_form_idx(form_ts)
    if form_idx is None:
        print(f"WARNING: form timestamp not found for {participant}: {form_ts}")
        continue
    form_row = forms.iloc[form_idx]
    video_map = flow_data.get(participant, {})
    for video_idx in range(1, 8):
        actual = video_map.get(video_idx)
        if actual is None:
            continue
        reported_raw = form_row[emo_cols[video_idx - 1]]
        reported     = parse_reported(reported_raw)
        match        = (actual in reported) if reported else False
        results.append({
            'participant'   : participant,
            'video_idx'     : video_idx,
            'actual_emotion': actual,
            'reported_raw'  : str(reported_raw),
            'reported_parsed': reported,
            'match'         : match,
        })

res_df = pd.DataFrame(results)

# ── Grid table ────────────────────────────────────────────────────────────────
EMOTIONS = ['Anger', 'Disgust', 'Fear', 'Happiness', 'Neutral', 'Sadness', 'Surprise']
col_w    = 11
header   = f"{'Participant':<20} " + "".join(f"{e:<{col_w}}" for e in EMOTIONS) + "Match"
sep      = "-" * len(header)

print("=" * len(header))
print("SELF-REPORT vs ACTUAL EMOTION")
print("✓ = match   ✗ = mismatch   · = not shown to participant")
print("=" * len(header))
print(header)
print(sep)

mismatch_records = []

for participant in sorted(PARTICIPANT_TO_FORM_TIMESTAMP.keys()):
    sub = res_df[res_df['participant'] == participant]
    emotion_marks = {}
    for _, r in sub.iterrows():
        emotion_marks[r['actual_emotion']] = '✓' if r['match'] else '✗'
        if not r['match']:
            mismatch_records.append({
                'participant'   : participant,
                'actual_emotion': r['actual_emotion'],
                'reported_raw'  : r['reported_raw'],
                'reported_parsed': r['reported_parsed'],
            })

    row_str  = f"{participant:<20} "
    row_str += "".join(f"{emotion_marks.get(e, '·'):<{col_w}}" for e in EMOTIONS)
    n_match  = sub['match'].sum()
    row_str += f"{n_match}/{len(sub)}"
    print(row_str)

print(sep)
total   = len(res_df)
matched = int(res_df['match'].sum())
print(f"Total: {total}  |  Match: {matched} ({100*matched/total:.1f}%)  |  Mismatch: {total-matched} ({100*(total-matched)/total:.1f}%)")

# ── Mismatch detail table ─────────────────────────────────────────────────────
mismatch_df = pd.DataFrame(mismatch_records)

print()
print("=" * 80)
print("RECORDS TO REMOVE (actual emotion NOT in self-report)")
print("=" * 80)
print(f"{'Participant':<20} {'Remove Emotion':<16} {'Self-Report (raw)'}")
print("-" * 80)
for _, r in mismatch_df.iterrows():
    print(f"{r['participant']:<20} {r['actual_emotion']:<16} {r['reported_raw']}")

print()
print(f"Total records to remove: {len(mismatch_df)}")
print()
print("By emotion:")
print(mismatch_df.groupby('actual_emotion').size().sort_values(ascending=False).to_string())
print()
print("By participant:")
print(mismatch_df.groupby('participant').size().sort_values(ascending=False).to_string())

# ── Export ────────────────────────────────────────────────────────────────────
out_dir = NEW_DATA_ROOT / "self_report_filter"
out_dir.mkdir(parents=True, exist_ok=True)

res_df.to_csv(out_dir / "all_pairs.csv", index=False)
mismatch_df[['participant', 'actual_emotion', 'reported_raw']].to_csv(
    out_dir / "records_to_remove.csv", index=False
)
print(f"\nSaved to: {out_dir}")
print("  - all_pairs.csv")
print("  - records_to_remove.csv")
````

## 4.4 Core Data-Loading Helpers

Defines the loading and windowing pipeline used for our own recordings:

- `data_files` / `all_files_list` — glob the per-participant CSVs for a given data source
  (`ACT_<emotion>.csv` for acted, `weights_<emotion>.csv` for natural)
- `subject_from_path` — extract the participant ID from the folder name; this is the grouping key
  that guarantees participant-exclusive splits
- `data_frame` — drop the timestamp column and reorder the AU columns to match `AU_NAMES` exactly
- `sep_process_one` — the **windowing rule**: a frame is marked active if any AU crosses the
  threshold (for Neutral, if *all* AUs stay below it); a 5-frame window is emitted only when it
  contains 3 consecutive active frames. Windows are non-overlapping (`step = SEQ_LENGTH`).
- `stack_sequences` / `to_numpy_labels` — convert window lists into `(N, 5, 70)` tensors
- `train_val_test_data_by_subject` — participant-exclusive 70/10/20 split via `StratifiedGroupKFold`
  (10 folds: 0-6 train, 7 validation, 8-9 test), with assertions that no participant appears in more
  than one split
- `mean_centered_max_scale` — normalize each AU independently within the current recording. The recording-level mean is subtracted from each feature, and the centered values are divided by the largest absolute centered value. This produces approximately zero-centered values within `[-1, 1]`.

Mean Centered Max Scale Normalization applied here:

```
Dataset            Function                            Normalization unit
-----------------  ----------------------------------  ---------------------------------------------------------
Original natural   sep_process_one()                   Applied independently to each CSV recording
Original acted     sep_process_one()                   Applied independently to each CSV recording
New participants   make_windows_from_new_recording()   Applied to each CSV recording after 60->10 Hz downsampling
EmojiHero          average_every_3_rows()              Applied independently to each parent_seq after 30->10 Hz downsampling
```

````python
def mean_centered_max_scale(df):
    """
    Apply mean-centered absolute-max scaling independently to each feature.

    For every feature column:
    1. Subtract the recording-level mean.
    2. Divide by the largest absolute centered value.

    The resulting values are approximately within [-1, 1].
    """
    centered = df - df.mean()

    absolute_max = centered.abs().max()

    scaled = centered / (absolute_max + 1e-8)

    return scaled
````

````python
def all_files_list(data_type: DataType):
    # Materialize generator -> list so we can split by files
    return list(data_files(data_type))

def forever(x):
    while True:
        yield x

def data_files_emo(emo: Emotion, data_type:DataType):
  if data_type == DataType.OUR_ACTED:
    yield from zip(
        glob(f"{data_folder_path}/*/ACT_{emo.value}.csv"),
        forever(emo),
    )
  elif data_type == DataType.OUR_NATURAL:
    yield from zip(
        glob(f"{data_folder_path}/*/weights_{emo.value}.csv"),
        forever(emo),
    )

def data_files(data_type:DataType):
    for emo in Emotion:
        yield from data_files_emo(emo, data_type)

# NEW: extract participant id from the folder between .../vr_filtered_clean/*/...
def subject_from_path(file_path: str) -> str:
    return os.path.basename(os.path.dirname(file_path))

def data_frame(file: str):
    df = pd.read_csv(file)

    itimes = df.iloc[:, 0].astype(int)
    df = df.iloc[:, 1:]

    missing = [au for au in AU_NAMES if au not in df.columns]

    if missing:
        raise ValueError(f"Missing AU columns: {missing}")

    df = df[AU_NAMES]

    return df, itimes


# Chunk the data from the file and give them in (chunk, label) pairs
def sep_process_one(file: str, emo: Emotion, thresh: Callable[[float], bool]):

    df, itimes = data_frame(file)                 # raw, for labeling
    #df_scaled = mean_centered_max_scale(df)       # per-recording, model input only

    labels = [0] * df.shape[0]
    try:
      emo_label = le.transform([emo.value])[0]
    except Exception as e:
      print(f"Error occured for file {file}")
      print(f"Emotion: {emo.value}")
      raise e

    for i in range(df.shape[0]):
      if emo == Emotion.NEUTRAL:
          au_values = df.iloc[i].tolist()
          if all(thresh(value) for value in au_values):
            labels[i]=1
      else:
        for au in AU_NAMES:
        # for au in EMO_TO_AU[emo]:
            if thresh(df[au][i]):
                labels[i] = 1
                break

    step = SEQ_LENGTH
    i, n = 0, df.shape[0] - SEQ_LENGTH

    pct = 0.2
    consecutive = 3

    def seq_sum(thresh):
        def fn(x, y):
            if x >= thresh:
                return thresh
            else:
                return 0 if y == 0 else x + y
        return fn

    while i <= n:
        count = reduce(seq_sum(consecutive), labels[i:i + SEQ_LENGTH])
        # if count >= SEQ_LENGTH * pct:
        if count == consecutive:
            yield (df[i:i + SEQ_LENGTH], emo_label)
            i += step
        else:
            i += 1

def sep_process(files: Generator[tuple[str, Emotion], None, None], thresh: dict[Emotion, Callable[[float], bool]]):
    for file, emo in files:
        yield from sep_process_one(file, emo, thresh[emo])


def stack_sequences(seq_list):
    """
    seq_list: list of pandas DataFrames of shape (SEQ_LENGTH, n_features)
    returns: np.ndarray of shape (N, SEQ_LENGTH, n_features)
    """
    return np.stack([df.values.astype(np.float32) for df in seq_list], axis=0)

def to_numpy_labels(labels_list):
    return np.asarray(labels_list, dtype=np.int64)

# NEW: subject-exclusive 70/10/20 split using StratifiedGroupKFold
def train_val_test_data_by_subject(
    thresh: dict[Emotion, Callable[[float], bool]],
    data_type: DataType,
    random_state: int = 42,
    n_splits: int = 10
):
    """
    Strict subject-exclusive split (participant in exactly one split),
    â‰ˆ70/10/20 by combining 10 StratifiedGroupKFold folds:
      - train: folds 0..6
      - val:   fold 7
      - test:  folds 8..9
    Stratifies by emotion at the file level while grouping by participant.
    Returns: X_train, X_val, X_test, y_train, y_val, y_test (lists)
    """
    all_files = all_files_list(data_type)
    if not all_files:
        raise RuntimeError("No files found. Check your glob path.")

    files, emos = zip(*all_files)

    # y = emotion labels for stratification (use your LabelEncoder if available)
    try:
        y = np.array([le.transform([e.value])[0] for e in emos], dtype=int)
    except Exception:
        # fallback to raw string labels if le not fitted here
        y = np.array([str(e.value) for e in emos])

    # groups = subject ids extracted from folder name
    groups = np.array([subject_from_path(p) for p in files])

    # build 10 folds stratified by emotion and grouped by subject
    sgkf = StratifiedGroupKFold(n_splits=n_splits, shuffle=True, random_state=random_state)
    folds = []
    dummy_X = np.zeros((len(y), 1))
    for _, test_idx in sgkf.split(dummy_X, y, groups=groups):
        folds.append(test_idx)
    if len(folds) != n_splits:
        raise RuntimeError("Unexpected number of folds produced.")

    # combine folds: 7 train, 1 val, 2 test
    train_idx = np.concatenate(folds[:7])
    val_idx   = folds[7]
    test_idx  = np.concatenate(folds[8:])

    # sanity: ensure subject exclusivity
    train_subjects = set(groups[train_idx])
    val_subjects   = set(groups[val_idx])
    test_subjects  = set(groups[test_idx])
    assert train_subjects.isdisjoint(val_subjects)
    assert train_subjects.isdisjoint(test_subjects)
    assert val_subjects.isdisjoint(test_subjects)

    # build split-specific file lists (tuples of (file, emo))
    train_files = [all_files[i] for i in train_idx]
    val_files   = [all_files[i] for i in val_idx]
    test_files  = [all_files[i] for i in test_idx]

    # chunk per split (no leakage)
    train_samples = list(sep_process(iter(train_files), thresh))
    val_samples   = list(sep_process(iter(val_files),   thresh))
    test_samples  = list(sep_process(iter(test_files),  thresh))

    if not train_samples:
        raise RuntimeError("No training sequences created. Adjust thresholds/windowing or verify data.")
    if not val_samples:
        raise RuntimeError("No validation sequences created. Adjust thresholds/windowing or verify data.")
    if not test_samples:
        raise RuntimeError("No test sequences created. Adjust thresholds/windowing or verify data.")

    X_train, y_train = zip(*train_samples)
    X_val,   y_val   = zip(*val_samples)
    X_test,  y_test  = zip(*test_samples)

    return list(X_train), list(X_val), list(X_test), list(y_train), list(y_val), list(y_test)
````

## 4.5 Helpers for the New Participant Recordings

The new participant recordings are stored in a different folder layout and were captured at 60 Hz.
These helpers bring them into the same format the teacher already expects: keep only the 70 AU
columns in `AU_NAMES`, downsample 60 Hz to 10 Hz, and emit `(SEQ_LENGTH, 70)` windows that can be
concatenated directly onto the existing data.

The teacher architecture and input shape are unchanged by this integration.

### Helper definitions

- `downsample_60hz_to_10hz_face_au` — coerces the AU columns to numeric, fills gaps, and averages
  every 6 frames (60 Hz -> 10 Hz)
- `new_collection_records` — walks `ActingRecordings/` and `VideoRecordings/` and returns one row per
  recording with participant, emotion, and domain (1 = acted, 0 = natural)
- `make_windows_from_new_recording` — applies the same threshold and 3-consecutive-frame rule as the
  original pipeline, so window selection is identical across data sources
- `load_new_collection_split` — expands a set of records into windows, emotion labels, and domain
  labels

These helpers keep the teacher model unchanged. They read the new participant CSVs, keep only the same 60 face AU columns in AU_NAMES, downsample 60 Hz to 10 Hz, create (SEQ_LENGTH, 60) windows, and return lists that can be appended to the previous data.

````python
from pathlib import Path

DOWNSAMPLE_FACTOR = 6  # 60 Hz -> 10 Hz


def downsample_60hz_to_10hz_face_au(df, au_names=AU_NAMES):
    """Keep only teacher face-AU features and convert new recordings from 60 Hz to 10 Hz."""
    missing = [c for c in au_names if c not in df.columns]
    if missing:
        raise ValueError(f"Missing face AU columns: {missing[:10]} ... total={len(missing)}")

    df = df[au_names].apply(pd.to_numeric, errors="coerce")
    df = df.replace([np.inf, -np.inf], np.nan)
    df = df.ffill().bfill().fillna(0.0)

    usable_rows = (len(df) // DOWNSAMPLE_FACTOR) * DOWNSAMPLE_FACTOR
    df = df.iloc[:usable_rows]

    return (
        df.groupby(np.arange(len(df)) // DOWNSAMPLE_FACTOR)
          .mean()
          .reset_index(drop=True)
    )


def new_collection_records(root=NEW_DATA_ROOT):
    """Return one row per new recording file with participant, emotion, and domain metadata."""
    rows = []
    valid_emotions = {e.value for e in Emotion}

    for recording_type in ["ActingRecordings", "VideoRecordings"]:
        base = root / recording_type
        if not base.exists():
            print(f"Warning: folder not found: {base}")
            continue

        for csv_path in base.glob("*/*/*.csv"):
            participant = csv_path.parents[1].name
            emotion_name = csv_path.parent.name

            if emotion_name not in valid_emotions:
                continue

            rows.append({
                "path": str(csv_path),
                "participant": participant,
                "emotion": Emotion(emotion_name),
                "recording_type": recording_type,
                "domain": 1 if recording_type == "ActingRecordings" else 0,
            })

    records = pd.DataFrame(rows)
    if records.empty:
        raise RuntimeError(
            f"No new participant CSV files found under {root}. "
            "Check NEW_DATA_ROOT and folder structure."
        )

    return records


def make_windows_from_new_recording(csv_path, emotion, thresh):
    """Create teacher-compatible windows from one new participant recording."""
    raw = pd.read_csv(csv_path)

    df = downsample_60hz_to_10hz_face_au(raw, AU_NAMES)   # raw downsampled
    #df_scaled = mean_centered_max_scale(df)               # model input only

    labels = [0] * len(df)

    for i in range(len(df)):
        if emotion == Emotion.NEUTRAL:
            if all(thresh[emotion](value) for value in df.iloc[i].tolist()):
                labels[i] = 1
        else:
            if any(thresh[emotion](df[au].iloc[i]) for au in AU_NAMES):
                labels[i] = 1

    windows, y = [], []
    emo_label = le.transform([emotion.value])[0]

    i = 0
    n = len(df) - SEQ_LENGTH
    consecutive = 3

    while i < n:
        has_consecutive = any(
            sum(labels[i + j:i + j + consecutive]) == consecutive
            for j in range(SEQ_LENGTH - consecutive + 1)
        )

        if has_consecutive:
            windows.append(df.iloc[i:i + SEQ_LENGTH][AU_NAMES])   #
            y.append(emo_label)
            i += SEQ_LENGTH
        else:
            i += 1

    return windows, y


def load_new_collection_split(record_df, thresh):
    """Load a split of new participant recordings into X, emotion labels, and domain labels."""
    X, y, domains = [], [], []

    for _, row in record_df.iterrows():
        windows, labels = make_windows_from_new_recording(
            row["path"],
            row["emotion"],
            thresh,
        )

        X.extend(windows)
        y.extend(labels)
        domains.extend([row["domain"]] * len(labels))

    return X, y, domains
````

````python
emojihero_splits = {
    "train": train_em_hero_df,
    "validation": valid_em_hero_df,
    "test": test_em_hero_df,
}

for split_name, df in emojihero_splits.items():
    # Preserve the complete sequence identifier.
    df["sequence_id"] = df["sequence_id"].astype(str)
    df["parent_seq"] = df["sequence_id"]

    # Extract the emotion ID from the final part of sequence_id.
    df["emotion"] = (
        df["sequence_id"]
        .str.rsplit("-", n=1)
        .str[-1]
        .astype(int)
    )

    # Verify that the extracted emotion matches the provided label.
    if "Label" in df.columns:
        label_values = pd.to_numeric(
            df["Label"],
            errors="raise",
        ).astype(int)

        if not (df["emotion"] == label_values).all():
            raise ValueError(
                f"{split_name}: sequence_id emotion IDs "
                "do not match the Label column."
            )

    # Sort frames chronologically within each sequence.
    df.sort_values(
        ["parent_seq", "timestamp"],
        kind="stable",
        inplace=True,
    )

    df.reset_index(
        drop=True,
        inplace=True,
    )

    # Verify that every sequence contains exactly 30 frames.
    frame_counts = df.groupby(
        "parent_seq",
        sort=False,
    ).size()

    invalid_frame_counts = frame_counts[
        frame_counts != 30
    ]

    if not invalid_frame_counts.empty:
        raise ValueError(
            f"{split_name}: some sequences do not contain "
            f"exactly 30 frames:\n"
            f"{invalid_frame_counts.head()}"
        )

    # Verify that every sequence contains one emotion only.
    emotion_counts = df.groupby(
        "parent_seq",
        sort=False,
    )["emotion"].nunique(dropna=False)

    mixed_sequences = emotion_counts[
        emotion_counts != 1
    ]

    if not mixed_sequences.empty:
        raise ValueError(
            f"{split_name}: some sequences contain multiple "
            f"emotion labels:\n"
            f"{mixed_sequences.head()}"
        )

    print(
        f"{split_name:10s}: "
        f"{len(df):6d} frames, "
        f"{df['parent_seq'].nunique():4d} sequences"
    )
````

## 5.3 Match the EmojiHeroVR Frame Rate to 10 Hz

### Average every 3 frames

EmojiHeroVR is recorded at 30 Hz. Averaging every 3 consecutive frames within a `parent_seq` group
brings it to the same 10 Hz sampling rate as our own downsampled recordings, so that a 5-frame window
spans 0.5 s in every data source.

````python
import pandas as pd

import pandas as pd
import numpy as np

def average_every_3_rows(df: pd.DataFrame, AU_NAMES: list) -> pd.DataFrame:
    """
    Average every 3 consecutive rows within each parent_seq group.
    - Only averages the numeric columns listed in AU_NAMES.
    - Keeps 'parent_seq' and 'emotion' from any of the 3 rows (they're identical).
    - Expects groups of 30 rows per parent_seq (produces 10 rows per group).

    Parameters
    ----------
    df : pd.DataFrame
        Input dataframe with string columns 'parent_seq' and 'emotion', and numeric AU columns.
    AU_NAMES : list
        List of column names that should be averaged (numeric columns).

    Returns
    -------
    pd.DataFrame
        Aggregated dataframe with columns: ['parent_seq', 'emotion'] + AU_NAMES
    """
    d = df.copy()

    # Ensure we only try to average columns that actually exist
    num_cols = [c for c in AU_NAMES if c in d.columns]

    # Create chunk index for each group of 3 within each parent_seq (0..9 for 30-row groups)
    d["_chunk"] = d.groupby("parent_seq").cumcount() // 3

    # Build aggregation spec: mean for AU columns, first for emotion
    agg_spec = {c: "mean" for c in num_cols}
    agg_spec["emotion"] = "first"

    out = (
        d.groupby(["parent_seq", "_chunk"], as_index=False)
         .agg(agg_spec)
         .drop(columns=["_chunk"])
    )

    # Mean-centered max scaling per recording (parent_seq)
    #out[num_cols] = (
     #   out.groupby("parent_seq", group_keys=False)[num_cols]
      #     .apply(mean_centered_max_scale)
    #)

    # Reorder columns
    ordered_cols = ["parent_seq", "emotion"] + num_cols
    return out[ordered_cols].reset_index(drop=True)

train_em_hero_df = average_every_3_rows(train_em_hero_df, AU_NAMES)
valid_em_hero_df = average_every_3_rows(valid_em_hero_df, AU_NAMES)
test_em_hero_df = average_every_3_rows(test_em_hero_df, AU_NAMES)

print("EmojiHero train columns:", train_em_hero_df.columns.tolist())
print("Number of EmojiHero AU features:", len([c for c in train_em_hero_df.columns if c in AU_NAMES]))

for eye_au in EYE_AU_NAMES:
    assert eye_au in train_em_hero_df.columns, f"{eye_au} missing from EmojiHero train!"
    assert eye_au in valid_em_hero_df.columns, f"{eye_au} missing from EmojiHero valid!"
    assert eye_au in test_em_hero_df.columns, f"{eye_au} missing from EmojiHero test!"

assert len([c for c in train_em_hero_df.columns if c in AU_NAMES]) == 70
````

## 5.4 Build 5-Frame EmojiHeroVR Windows

### Sliding 5-frame windows

Builds 5-frame windows with stride 1 and shuffles them. The window label is taken from the emotion
column of the frames it contains.

**Note:** unlike our own recordings, these windows overlap and are not constrained to stay inside a
single `parent_seq`. Because train/validation/test come from EmojiHeroVR's own predefined splits,
this does not leak across splits, but it does mean EmojiHeroVR contributes more (and more correlated)
windows than a non-overlapping scheme would.

````python
def create_emojihero_windows(
    df,
    feature_cols,
    window_size=5,
    stride=1,
):
    """
    Create sliding windows within each EmojiHero sequence.

    A window is never allowed to cross a sequence or emotion
    boundary.
    """
    X = []
    y = []

    for sequence_id, sequence_df in df.groupby(
        "parent_seq",
        sort=False,
    ):
        sequence_df = sequence_df.reset_index(drop=True)

        labels = sequence_df["emotion"].unique()

        if len(labels) != 1:
            raise ValueError(
                f"Sequence {sequence_id} contains multiple "
                f"emotion labels: {labels.tolist()}"
            )

        if len(sequence_df) != 10:
            raise ValueError(
                f"Sequence {sequence_id} contains "
                f"{len(sequence_df)} downsampled frames; "
                "expected 10."
            )

        label = int(labels[0])

        for start in range(
            0,
            len(sequence_df) - window_size + 1,
            stride,
        ):
            end = start + window_size

            window = (
                sequence_df
                .iloc[start:end][feature_cols]
                .copy()
            )

            if window.shape != (
                window_size,
                len(feature_cols),
            ):
                raise ValueError(
                    f"Invalid window shape for sequence "
                    f"{sequence_id}: {window.shape}"
                )

            X.append(window)
            y.append(label)

    return X, y


X_train_em_hero, Y_train_em_hero = (
    create_emojihero_windows(
        train_em_hero_df,
        AU_NAMES,
        window_size=SEQ_LENGTH,
        stride=1,
    )
)

X_val_em_hero, Y_val_em_hero = (
    create_emojihero_windows(
        valid_em_hero_df,
        AU_NAMES,
        window_size=SEQ_LENGTH,
        stride=1,
    )
)

X_test_em_hero, Y_test_em_hero = (
    create_emojihero_windows(
        test_em_hero_df,
        AU_NAMES,
        window_size=SEQ_LENGTH,
        stride=1,
    )
)


# Shuffle windows and labels together to preserve alignment.
X_train_em_hero, Y_train_em_hero = sk_shuffle(
    X_train_em_hero,
    Y_train_em_hero,
    random_state=42,
)

X_val_em_hero, Y_val_em_hero = sk_shuffle(
    X_val_em_hero,
    Y_val_em_hero,
    random_state=42,
)

X_test_em_hero, Y_test_em_hero = sk_shuffle(
    X_test_em_hero,
    Y_test_em_hero,
    random_state=42,
)


print("EmojiHero window counts:")
print("Train:", len(X_train_em_hero))
print("Validation:", len(X_val_em_hero))
print("Test:", len(X_test_em_hero))


# Final consistency checks.
assert len(X_train_em_hero) == len(Y_train_em_hero)
assert len(X_val_em_hero) == len(Y_val_em_hero)
assert len(X_test_em_hero) == len(Y_test_em_hero)

assert all(
    window.shape == (SEQ_LENGTH, len(AU_NAMES))
    for window in X_train_em_hero
)

assert all(
    window.shape == (SEQ_LENGTH, len(AU_NAMES))
    for window in X_val_em_hero
)

assert all(
    window.shape == (SEQ_LENGTH, len(AU_NAMES))
    for window in X_test_em_hero
)
````

## 5.5 Build the Final Training Corpus (with the new participants)



### 5.5.1 Full dataset build

This single cell performs the complete corpus construction:

1. **Thresholds** — an AU is active above 0.25; a Neutral frame requires all AUs at or below 0.25.
2. **Our data** — natural and acted recordings are split participant-exclusively (70/10/20).
3. **Domain labels** — 0 = natural (video-elicited), 1 = acted. EmojiHeroVR is merged into the acted
   domain.
4. **Self-report filter** — two participants are dropped entirely (0/7 and 1/7 self-report matches),
   plus eight specific (participant, emotion) pairs where the intended emotion was never reported.
   These lists come from Part 2.
5. **New participant split** — `StratifiedGroupKFold` with one fold per remaining participant, then
   allocated to train/validation/test by a 70/10/20 participant ratio, with disjointness assertions.
6. **Combination and shuffling** — natural + acted/EmojiHeroVR + new participants, shuffled with a
   fixed seed.
7. **Tensors** — every split is stacked into `(N, 5, 70)` float arrays.
8. **Class weights** — balanced weights, renormalised to average 1.0, to counter the class imbalance
   across the seven emotions.
9. **Assertions** — every tensor must still have shape `(SEQ_LENGTH, 70)`.

#### Corpus without EmojiHero in training

````python
# ============================================================
# 5.5.1  ACTED-ONLY CORPUS, EmojiHero NOT IN TRAINING (volume augmentation runs)
# Steps 1-4 are identical to the 'With Self-Filtering-Only-Acted' cell.
# ============================================================

thresh = {emo: lambda x: x > 0.25 for emo in Emotion}
thresh[Emotion.NEUTRAL] = lambda x: x <= 0.25


# ============================================================
# 1. OLD VR ACTED DATA
# Participant-exclusive split is performed inside this function.
# ============================================================

(
    X_train_act,
    X_val_act,
    X_test_act,
    Y_train_act,
    Y_val_act,
    Y_test_act,
) = train_val_test_data_by_subject(
    thresh,
    DataType.OUR_ACTED,
)


# ============================================================
# 2. NEW PARTICIPANTS — ACTED RECORDINGS ONLY
# ============================================================

records_new_all = new_collection_records()

records_new = (
    records_new_all[
        records_new_all["recording_type"] == "ActingRecordings"
    ]
    .copy()
    .reset_index(drop=True)
)

assert not records_new.empty
assert set(records_new["recording_type"]) == {"ActingRecordings"}
assert set(records_new["domain"]) == {1}

print("New acted participants:",
      records_new["participant"].nunique())

print("New acted recording count:",
      len(records_new))

print(
    pd.crosstab(
        records_new["participant"],
        records_new["emotion"].apply(lambda e: e.value),
    )
)

records_new["Yemo"] = records_new["emotion"].apply(
    lambda e: le.transform([e.value])[0]
)
records_new["fold"] = -1


# ============================================================
# 3. PARTICIPANT-EXCLUSIVE SPLIT FOR NEW ACTED DATA
# With 15 participants this normally becomes:
# 10 train / 2 validation / 3 test
# ============================================================

n_participants = records_new["participant"].nunique()

assert n_participants >= 3

sgkf_new = StratifiedGroupKFold(
    n_splits=n_participants,
    shuffle=True,
    random_state=42,
)

for fold_id, (_, heldout_idx) in enumerate(
    sgkf_new.split(
        records_new["path"],
        records_new["Yemo"],
        groups=records_new["participant"],
    )
):
    records_new.loc[heldout_idx, "fold"] = fold_id


VAL_RATIO = 0.10
TEST_RATIO = 0.20

n_test = max(1, round(TEST_RATIO * n_participants))
n_val = max(1, round(VAL_RATIO * n_participants))
n_train = n_participants - n_val - n_test

assert n_train > 0
assert n_val > 0
assert n_test > 0

train_folds = list(range(0, n_train))
val_folds = list(range(n_train, n_train + n_val))
test_folds = list(range(n_train + n_val, n_participants))

new_train_records = records_new[
    records_new["fold"].isin(train_folds)
].reset_index(drop=True)

new_val_records = records_new[
    records_new["fold"].isin(val_folds)
].reset_index(drop=True)

new_test_records = records_new[
    records_new["fold"].isin(test_folds)
].reset_index(drop=True)


train_participants = set(new_train_records["participant"])
val_participants = set(new_val_records["participant"])
test_participants = set(new_test_records["participant"])

assert train_participants.isdisjoint(val_participants)
assert train_participants.isdisjoint(test_participants)
assert val_participants.isdisjoint(test_participants)

assert set(new_train_records["recording_type"]) == {"ActingRecordings"}
assert set(new_val_records["recording_type"]) == {"ActingRecordings"}
assert set(new_test_records["recording_type"]) == {"ActingRecordings"}


# ============================================================
# 4. CREATE WINDOWS FOR NEW ACTED DATA
# ============================================================

X_train_new, Y_train_new, Y_train_domain_new = (
    load_new_collection_split(new_train_records, thresh)
)

X_val_new, Y_val_new, Y_val_domain_new = (
    load_new_collection_split(new_val_records, thresh)
)

X_test_new, Y_test_new, Y_test_domain_new = (
    load_new_collection_split(new_test_records, thresh)
)

assert set(Y_train_domain_new) == {1}
assert set(Y_val_domain_new) == {1}
assert set(Y_test_domain_new) == {1}


# ============================================================
# 5. FINAL TRAINING CORPUS — EmojiHero is NOT trained on
#
# Training:
#   old VR acted + new VR acted
#
# Validation used by early stopping:
#   old VR acted validation + new VR acted validation
#
# EmojiHero validation and test are kept for reporting only.
# ============================================================

SOURCE_OLD_ACTED = 0
SOURCE_NEW_ACTED = 1
NUM_SOURCES = 2

X_train_comb = X_train_act + X_train_new
Y_train_comb = Y_train_act + Y_train_new

train_sources = (
    [SOURCE_OLD_ACTED] * len(X_train_act)
    + [SOURCE_NEW_ACTED] * len(X_train_new)
)

assert len(X_train_comb) == len(Y_train_comb) == len(train_sources)

# Primary deployment validation: real VR acted only.
X_val_vr = X_val_act + X_val_new
Y_val_vr = Y_val_act + Y_val_new

# Primary test: real VR acted only.
X_test_vr = X_test_act + X_test_new
Y_test_vr = Y_test_act + Y_test_new

# All acted sources, for secondary reporting.
X_test_all_acted = X_test_vr + X_test_em_hero
Y_test_all_acted = Y_test_vr + Y_test_em_hero


# ============================================================
# 6. DETERMINISTIC SHUFFLE
# ============================================================

X_train_comb, Y_train_comb, train_sources = sk_shuffle(
    X_train_comb,
    Y_train_comb,
    train_sources,
    random_state=42,
)

X_val_vr, Y_val_vr = sk_shuffle(X_val_vr, Y_val_vr, random_state=42)
X_test_vr, Y_test_vr = sk_shuffle(X_test_vr, Y_test_vr, random_state=42)
X_test_all_acted, Y_test_all_acted = sk_shuffle(
    X_test_all_acted,
    Y_test_all_acted,
    random_state=42,
)


# ============================================================
# 7. NUMPY TENSORS
# ============================================================

X_train_np = stack_sequences(X_train_comb)
Y_train_np = to_numpy_labels(Y_train_comb)
train_sources_np = np.asarray(train_sources, dtype=np.int64)

# train_then_test uses X_val_np, so point it to VR acted validation.
X_val_np = stack_sequences(X_val_vr)
Y_val_np = to_numpy_labels(Y_val_vr)

X_val_emojihero_np = stack_sequences(X_val_em_hero)
Y_val_emojihero_np = to_numpy_labels(Y_val_em_hero)

X_test_vr_np = stack_sequences(X_test_vr)
Y_test_vr_np = to_numpy_labels(Y_test_vr)

X_test_emojihero_np = stack_sequences(X_test_em_hero)
Y_test_emojihero_np = to_numpy_labels(Y_test_em_hero)

X_test_all_acted_np = stack_sequences(X_test_all_acted)
Y_test_all_acted_np = to_numpy_labels(Y_test_all_acted)

# Compatibility alias for older helper cells.
X_test_comb_np = X_test_all_acted_np
Y_test_comb_np = Y_test_all_acted_np

# Domain labels are all acted and are not used by the single-input model.
Y_train_domain_np = np.ones(len(Y_train_np), dtype=np.int64)
Y_val_domain_np = np.ones(len(Y_val_np), dtype=np.int64)
Y_test_domain_comb_np = np.ones(len(Y_test_comb_np), dtype=np.int64)


# ============================================================
# 8. CLASS WEIGHTS x SOURCE WEIGHTS = SAMPLE WEIGHTS
# Each source contributes half of the source-level loss.
# ============================================================

present_classes = np.unique(Y_train_np)
assert set(present_classes) == set(range(len(Emotion)))

class_weights = compute_class_weight(
    class_weight="balanced",
    classes=present_classes,
    y=Y_train_np,
)
class_weight_by_id = {
    int(class_id): float(weight)
    for class_id, weight in zip(present_classes, class_weights)
}

source_counts = np.bincount(train_sources_np, minlength=NUM_SOURCES)
assert np.all(source_counts > 0)
source_weights = len(train_sources_np) / (NUM_SOURCES * source_counts)

train_sample_weights = np.asarray(
    [
        source_weights[int(source_id)] * class_weight_by_id[int(class_id)]
        for source_id, class_id in zip(train_sources_np, Y_train_np)
    ],
    dtype=np.float32,
)
train_sample_weights /= train_sample_weights.mean()

assert np.isfinite(train_sample_weights).all()
assert np.all(train_sample_weights > 0)


# ============================================================
# 9. FINAL ASSERTIONS
# ============================================================

assert len(X_train_np) == len(Y_train_np) == len(train_sample_weights)
assert set(np.unique(train_sources_np)) == {SOURCE_OLD_ACTED, SOURCE_NEW_ACTED}

for name, X, y in [
    ("train", X_train_np, Y_train_np),
    ("VR validation", X_val_np, Y_val_np),
    ("EmojiHero validation", X_val_emojihero_np, Y_val_emojihero_np),
    ("VR test", X_test_vr_np, Y_test_vr_np),
    ("EmojiHero test", X_test_emojihero_np, Y_test_emojihero_np),
    ("all acted test", X_test_all_acted_np, Y_test_all_acted_np),
]:
    assert X.shape[1:] == (SEQ_LENGTH, len(AU_NAMES)), (name, X.shape)
    assert len(X) == len(y), name
    assert np.isfinite(X).all(), name

source_names = {SOURCE_OLD_ACTED: "old acted", SOURCE_NEW_ACTED: "new acted"}
print("Source counts:", source_counts)
print("Source weights:", source_weights)
for source_id, source_name in source_names.items():
    mask = train_sources_np == source_id
    print(
        f"{source_name:10s} | n={mask.sum():4d} | "
        f"mean weight={train_sample_weights[mask].mean():.4f} | "
        f"total={train_sample_weights[mask].sum():.2f}"
    )

print("\nActed-only corpus WITHOUT EmojiHero training: build complete.")
print("Train:", X_train_np.shape, "(expected about 1553 + 844 = 2397)")
print("VR validation:", X_val_np.shape)
print("EmojiHero validation (reporting only):", X_val_emojihero_np.shape)
print("VR test:", X_test_vr_np.shape)
print("EmojiHero test (reporting only):", X_test_emojihero_np.shape)
````

##### volume augmentation

````python
# ======================================================================
# PART 5.8  PER-WINDOW VOLUME AUGMENTATION (training windows only)
#
# Every epoch shows every original training window once, plus AUG_COPIES
# scaled copies of it. Copies are redrawn every epoch and never saved.
# Validation and test windows are never changed.
#
# Face size = mean over the 5 frames of the sum of the 63 non-tongue
# signals. The multiplier is applied to all 70 signals.
# ======================================================================

import math

AUG_COPIES = 1                 # k; 0 = model A (no augmentation), 1 = model B
AUG_SEED = 42

FACE_SIZE_CHANNELS = slice(0, 63)   # tongue (63-69) is left out of the size only
TARGET_SIZE_LOW = 0.8
TARGET_SIZE_HIGH = 5.5
GAIN_MIN = 1.0 / 3.0
GAIN_MAX = 3.0

NEUTRAL_ID = int(le.transform([Emotion.NEUTRAL.value])[0])
NEUTRAL_GAIN_LOW = 0.5
NEUTRAL_GAIN_HIGH = 1.5
NEUTRAL_MAX_SIGNAL = 0.25           # the notebook's own Neutral rule

assert AU_NAMES[62] == "UpperLipRaiserR" and AU_NAMES[63].startswith("Tongue")


def window_face_size(X):
    """(N, 5, 70) -> (N,)"""
    return X[..., FACE_SIZE_CHANNELS].sum(axis=-1).mean(axis=-1)


def gain_ranges(X, y):
    """Allowed multiplier range per window."""
    size = np.maximum(window_face_size(X), 1e-6)
    low = np.clip(TARGET_SIZE_LOW / size, GAIN_MIN, GAIN_MAX)
    high = np.clip(TARGET_SIZE_HIGH / size, GAIN_MIN, GAIN_MAX)

    neutral = y == NEUTRAL_ID
    low[neutral] = NEUTRAL_GAIN_LOW
    high[neutral] = NEUTRAL_GAIN_HIGH
    return low, high


def draw_gains(X, y, low, high, rng):
    """One log-uniform multiplier per window; Neutral is capped at 0.25."""
    gains = np.exp(rng.uniform(np.log(low), np.log(high)))

    neutral = y == NEUTRAL_ID
    peak = X[neutral].max(axis=(1, 2))
    gains[neutral] = np.minimum(
        gains[neutral],
        NEUTRAL_MAX_SIGNAL / np.maximum(peak, 1e-6),
    )
    return gains.astype(np.float32)


class VolumeAugmentedWindows(tf.keras.utils.PyDataset):
    """Originals + `copies` freshly scaled copies per window, reshuffled every epoch.

    Each copy keeps its original's label and sample weight, so the class and
    source balance of the sample weights is unchanged.
    """

    def __init__(self, X, y, sample_weight, copies, batch_size, seed, **kwargs):
        super().__init__(**kwargs)
        self.X = np.asarray(X, dtype=np.float32)
        self.y = np.asarray(y, dtype=np.int64)
        self.w = np.asarray(sample_weight, dtype=np.float32)
        self.copies = int(copies)
        self.batch_size = int(batch_size)
        self.rng = np.random.default_rng(seed)
        self.low, self.high = gain_ranges(self.X, self.y)
        self._new_epoch()

    def _new_epoch(self):
        n = len(self.X)
        self.source = np.tile(np.arange(n), 1 + self.copies)
        self.gain = np.ones(len(self.source), dtype=np.float32)
        self.is_copy = np.arange(len(self.source)) >= n
        if self.copies:
            self.gain[n:] = np.concatenate([
                draw_gains(self.X, self.y, self.low, self.high, self.rng)
                for _ in range(self.copies)
            ])
        order = self.rng.permutation(len(self.source))
        self.source = self.source[order]
        self.gain = self.gain[order]
        self.is_copy = self.is_copy[order]

    def __len__(self):
        return math.ceil(len(self.source) / self.batch_size)

    def __getitem__(self, index):
        batch = slice(index * self.batch_size, (index + 1) * self.batch_size)
        rows = self.source[batch]
        X_batch = np.clip(self.X[rows] * self.gain[batch, None, None], 0.0, 1.0)
        return X_batch, self.y[rows], self.w[rows]

    def on_epoch_end(self):
        self._new_epoch()


# ----------------------------------------------------------------------
# Sanity check on one epoch of copies (does not train anything)
# ----------------------------------------------------------------------

check = VolumeAugmentedWindows(
    X_train_np, Y_train_np, train_sample_weights,
    copies=max(AUG_COPIES, 1), batch_size=10, seed=AUG_SEED,
)
n_original = len(X_train_np)
copy_rows = check.source[check.is_copy]
copy_gain = check.gain[check.is_copy]
X_copies = np.clip(X_train_np[copy_rows] * copy_gain[:, None, None], 0.0, 1.0)
y_copies = Y_train_np[copy_rows]
expressive = y_copies != NEUTRAL_ID

print(f"Original training windows: {n_original}")
print(f"Windows per epoch with AUG_COPIES={AUG_COPIES}: {n_original * (1 + AUG_COPIES)}")

q = [0.05, 0.25, 0.5, 0.75, 0.95]
before = window_face_size(X_train_np[Y_train_np != NEUTRAL_ID])
after = window_face_size(X_copies[expressive])
print("\nExpressive face size, 5/25/50/75/95%:")
print("  originals:", np.round(np.quantile(before, q), 2))
print("  copies:   ", np.round(np.quantile(after, q), 2))

raw = X_train_np[copy_rows] * copy_gain[:, None, None]
print(f"\nCopy values clipped at 1.0: {np.mean(raw > 1.0):.3%} "
      f"(copies with any clipping: {np.mean((raw > 1.0).any(axis=(1, 2))):.1%})")
print(f"Largest Neutral signal in a copy: {X_copies[~expressive].max():.3f} (must be <= 0.25)")
assert X_copies[~expressive].max() <= NEUTRAL_MAX_SIGNAL + 1e-6
assert 0.0 <= X_copies.min() and X_copies.max() <= 1.0
del check, X_copies, raw
````

#### With Self-Filtering-Only-Acted

````python
thresh = {emo: lambda x: x > 0.25 for emo in Emotion}
thresh[Emotion.NEUTRAL] = lambda x: x <= 0.25


# ============================================================
# 1. OLD VR ACTED DATA
# Participant-exclusive split is performed inside this function.
# ============================================================

(
    X_train_act,
    X_val_act,
    X_test_act,
    Y_train_act,
    Y_val_act,
    Y_test_act,
) = train_val_test_data_by_subject(
    thresh,
    DataType.OUR_ACTED,
)


# ============================================================
# 2. NEW PARTICIPANTS — ACTED RECORDINGS ONLY
# ============================================================

records_new_all = new_collection_records()

records_new = (
    records_new_all[
        records_new_all["recording_type"] == "ActingRecordings"
    ]
    .copy()
    .reset_index(drop=True)
)

assert not records_new.empty
assert set(records_new["recording_type"]) == {"ActingRecordings"}
assert set(records_new["domain"]) == {1}

print("New acted participants:",
      records_new["participant"].nunique())

print("New acted recording count:",
      len(records_new))

print(
    pd.crosstab(
        records_new["participant"],
        records_new["emotion"].apply(lambda e: e.value),
    )
)

records_new["Yemo"] = records_new["emotion"].apply(
    lambda e: le.transform([e.value])[0]
)
records_new["fold"] = -1


# ============================================================
# 3. PARTICIPANT-EXCLUSIVE SPLIT FOR NEW ACTED DATA
# With 15 participants this normally becomes:
# 10 train / 2 validation / 3 test
# ============================================================

n_participants = records_new["participant"].nunique()

assert n_participants >= 3

sgkf_new = StratifiedGroupKFold(
    n_splits=n_participants,
    shuffle=True,
    random_state=42,
)

for fold_id, (_, heldout_idx) in enumerate(
    sgkf_new.split(
        records_new["path"],
        records_new["Yemo"],
        groups=records_new["participant"],
    )
):
    records_new.loc[heldout_idx, "fold"] = fold_id


VAL_RATIO = 0.10
TEST_RATIO = 0.20

n_test = max(1, round(TEST_RATIO * n_participants))
n_val = max(1, round(VAL_RATIO * n_participants))
n_train = n_participants - n_val - n_test

assert n_train > 0
assert n_val > 0
assert n_test > 0

train_folds = list(range(0, n_train))
val_folds = list(range(n_train, n_train + n_val))
test_folds = list(range(n_train + n_val, n_participants))

new_train_records = records_new[
    records_new["fold"].isin(train_folds)
].reset_index(drop=True)

new_val_records = records_new[
    records_new["fold"].isin(val_folds)
].reset_index(drop=True)

new_test_records = records_new[
    records_new["fold"].isin(test_folds)
].reset_index(drop=True)


train_participants = set(new_train_records["participant"])
val_participants = set(new_val_records["participant"])
test_participants = set(new_test_records["participant"])

assert train_participants.isdisjoint(val_participants)
assert train_participants.isdisjoint(test_participants)
assert val_participants.isdisjoint(test_participants)

assert set(new_train_records["recording_type"]) == {"ActingRecordings"}
assert set(new_val_records["recording_type"]) == {"ActingRecordings"}
assert set(new_test_records["recording_type"]) == {"ActingRecordings"}


# ============================================================
# 4. CREATE WINDOWS FOR NEW ACTED DATA
# ============================================================

X_train_new, Y_train_new, Y_train_domain_new = (
    load_new_collection_split(new_train_records, thresh)
)

X_val_new, Y_val_new, Y_val_domain_new = (
    load_new_collection_split(new_val_records, thresh)
)

X_test_new, Y_test_new, Y_test_domain_new = (
    load_new_collection_split(new_test_records, thresh)
)

assert set(Y_train_domain_new) == {1}
assert set(Y_val_domain_new) == {1}
assert set(Y_test_domain_new) == {1}


# ============================================================
# 5. FINAL TRAINING CORPUS
#
# Training:
#   old VR acted + EmojiHero training + new VR acted
#
# Validation used by early stopping:
#   old VR acted validation + new VR acted validation
#
# EmojiHero validation is kept separately as a guardrail so its
# overlapping windows do not dominate early stopping.
# ============================================================

X_train_comb = (
    X_train_act
    + X_train_em_hero
    + X_train_new
)

Y_train_comb = (
    Y_train_act
    + Y_train_em_hero
    + Y_train_new
)

# Primary deployment validation: real VR acted only.
X_val_vr = X_val_act + X_val_new
Y_val_vr = Y_val_act + Y_val_new

# Optional all-acted validation, for reporting only.
X_val_all_acted = X_val_vr + X_val_em_hero
Y_val_all_acted = Y_val_vr + Y_val_em_hero

# Primary test: real VR acted only.
X_test_vr = X_test_act + X_test_new
Y_test_vr = Y_test_act + Y_test_new

# All acted sources, for secondary reporting.
X_test_all_acted = X_test_vr + X_test_em_hero
Y_test_all_acted = Y_test_vr + Y_test_em_hero


# ============================================================
# 6. DETERMINISTIC SHUFFLE
# ============================================================

X_train_comb, Y_train_comb = sk_shuffle(
    X_train_comb,
    Y_train_comb,
    random_state=42,
)

X_val_vr, Y_val_vr = sk_shuffle(
    X_val_vr,
    Y_val_vr,
    random_state=42,
)

X_val_all_acted, Y_val_all_acted = sk_shuffle(
    X_val_all_acted,
    Y_val_all_acted,
    random_state=42,
)

X_test_vr, Y_test_vr = sk_shuffle(
    X_test_vr,
    Y_test_vr,
    random_state=42,
)

X_test_all_acted, Y_test_all_acted = sk_shuffle(
    X_test_all_acted,
    Y_test_all_acted,
    random_state=42,
)


# ============================================================
# 7. NUMPY TENSORS
# ============================================================

X_train_np = stack_sequences(X_train_comb)
Y_train_np = to_numpy_labels(Y_train_comb)

# train_then_test uses X_val_np, so point it to VR acted validation.
X_val_np = stack_sequences(X_val_vr)
Y_val_np = to_numpy_labels(Y_val_vr)

X_val_emojihero_np = stack_sequences(X_val_em_hero)
Y_val_emojihero_np = to_numpy_labels(Y_val_em_hero)

X_val_all_acted_np = stack_sequences(X_val_all_acted)
Y_val_all_acted_np = to_numpy_labels(Y_val_all_acted)

X_test_vr_np = stack_sequences(X_test_vr)
Y_test_vr_np = to_numpy_labels(Y_test_vr)

X_test_emojihero_np = stack_sequences(X_test_em_hero)
Y_test_emojihero_np = to_numpy_labels(Y_test_em_hero)

X_test_all_acted_np = stack_sequences(X_test_all_acted)
Y_test_all_acted_np = to_numpy_labels(Y_test_all_acted)

# Compatibility alias for older helper cells.
X_test_comb_np = X_test_all_acted_np
Y_test_comb_np = Y_test_all_acted_np

# Domain labels are all acted and are not used by the single-input model.
Y_train_domain_np = np.ones(len(Y_train_np), dtype=np.int64)
Y_val_domain_np = np.ones(len(Y_val_np), dtype=np.int64)
Y_test_domain_comb_np = np.ones(
    len(Y_test_comb_np),
    dtype=np.int64,
)


# ============================================================
# 8. CLASS WEIGHTS
# ============================================================

present_classes = np.unique(Y_train_np)

assert set(present_classes) == set(range(len(Emotion)))

class_weights = compute_class_weight(
    class_weight="balanced",
    classes=present_classes,
    y=Y_train_np,
)

class_weights = class_weights / class_weights.sum()
class_weights = class_weights * len(class_weights)


# ============================================================
# 9. FINAL ASSERTIONS
# ============================================================

for name, X, y in [
    ("train", X_train_np, Y_train_np),
    ("VR validation", X_val_np, Y_val_np),
    ("EmojiHero validation", X_val_emojihero_np, Y_val_emojihero_np),
    ("VR test", X_test_vr_np, Y_test_vr_np),
    ("EmojiHero test", X_test_emojihero_np, Y_test_emojihero_np),
    ("all acted test", X_test_all_acted_np, Y_test_all_acted_np),
]:
    assert X.shape[1:] == (SEQ_LENGTH, len(AU_NAMES)), (
        name,
        X.shape,
    )
    assert len(X) == len(y)
    assert np.isfinite(X).all()

print("Acted-only corpus build complete.")
print("Train:", X_train_np.shape)
print("VR validation:", X_val_np.shape)
print("EmojiHero validation:", X_val_emojihero_np.shape)
print("VR test:", X_test_vr_np.shape)
print("EmojiHero test:", X_test_emojihero_np.shape)
````

````python
from collections import Counter
import pandas as pd

class_names = [emotion.value for emotion in Emotion]

def source_summary(name, labels):
    counts = Counter(np.asarray(labels, dtype=int).tolist())

    row = {
        "source": name,
        "total": len(labels),
    }

    for class_id, class_name in enumerate(class_names):
        row[class_name] = counts.get(class_id, 0)

    return row


source_rows = [
    source_summary("old_acted_train", Y_train_act),
    source_summary("emojihero_train", Y_train_em_hero),
    source_summary("new_acted_train", Y_train_new),

    source_summary("old_acted_validation", Y_val_act),
    source_summary("emojihero_validation", Y_val_em_hero),
    source_summary("new_acted_validation", Y_val_new),

    source_summary("old_acted_test", Y_test_act),
    source_summary("emojihero_test", Y_test_em_hero),
    source_summary("new_acted_test", Y_test_new),
]

source_df = pd.DataFrame(source_rows)

display(source_df)

train_total = (
    len(Y_train_act)
    + len(Y_train_em_hero)
    + len(Y_train_new)
)

print("\nTraining source percentages:")

for name, labels in [
    ("old acted", Y_train_act),
    ("EmojiHero", Y_train_em_hero),
    ("new acted", Y_train_new),
]:
    percentage = 100 * len(labels) / train_total
    print(f"{name:15s}: {len(labels):5d} ({percentage:6.2f}%)")

assert train_total == len(Y_train_np)
````

````python
SOURCE_OLD_ACTED = 0
SOURCE_EMOJIHERO = 1
SOURCE_NEW_ACTED = 2

X_train_comb = (
    X_train_act
    + X_train_em_hero
    + X_train_new
)

Y_train_comb = (
    Y_train_act
    + Y_train_em_hero
    + Y_train_new
)

train_sources = (
    [SOURCE_OLD_ACTED] * len(X_train_act)
    + [SOURCE_EMOJIHERO] * len(X_train_em_hero)
    + [SOURCE_NEW_ACTED] * len(X_train_new)
)

assert (
    len(X_train_comb)
    == len(Y_train_comb)
    == len(train_sources)
)

X_train_comb, Y_train_comb, train_sources = sk_shuffle(
    X_train_comb,
    Y_train_comb,
    train_sources,
    random_state=42,
)
````

````python
# ============================================================
# CONVERT TO NUMPY + BUILD CLASS/SOURCE-BALANCED SAMPLE WEIGHTS
# ============================================================

# Training tensors
X_train_np = stack_sequences(X_train_comb)
Y_train_np = to_numpy_labels(Y_train_comb)

train_sources_np = np.asarray(
    train_sources,
    dtype=np.int64,
)

assert len(X_train_np) == len(Y_train_np)
assert len(Y_train_np) == len(train_sources_np)


# ------------------------------------------------------------
# Validation tensors
# Primary validation = real VR acted only
# ------------------------------------------------------------

X_val_np = stack_sequences(X_val_vr)
Y_val_np = to_numpy_labels(Y_val_vr)

X_val_emojihero_np = stack_sequences(X_val_em_hero)
Y_val_emojihero_np = to_numpy_labels(Y_val_em_hero)

X_val_all_acted_np = stack_sequences(X_val_all_acted)
Y_val_all_acted_np = to_numpy_labels(Y_val_all_acted)


# ------------------------------------------------------------
# Test tensors
# Do not run inference on these until the model is frozen.
# ------------------------------------------------------------

X_test_vr_np = stack_sequences(X_test_vr)
Y_test_vr_np = to_numpy_labels(Y_test_vr)

X_test_emojihero_np = stack_sequences(X_test_em_hero)
Y_test_emojihero_np = to_numpy_labels(Y_test_em_hero)

X_test_all_acted_np = stack_sequences(X_test_all_acted)
Y_test_all_acted_np = to_numpy_labels(Y_test_all_acted)

# Compatibility aliases for older notebook cells
X_test_comb_np = X_test_all_acted_np
Y_test_comb_np = Y_test_all_acted_np


# ------------------------------------------------------------
# Class weights
# ------------------------------------------------------------

present_classes = np.unique(Y_train_np)

assert set(present_classes) == set(range(len(Emotion)))

class_weights = compute_class_weight(
    class_weight="balanced",
    classes=present_classes,
    y=Y_train_np,
)

class_weight_by_id = {
    int(class_id): float(weight)
    for class_id, weight in zip(
        present_classes,
        class_weights,
    )
}


# ------------------------------------------------------------
# Source weights
# Each source contributes one third of the source-level loss.
# ------------------------------------------------------------

source_counts = np.bincount(
    train_sources_np,
    minlength=3,
)

assert np.all(source_counts > 0)

source_weights = (
    len(train_sources_np)
    / (3.0 * source_counts)
)

print("Source counts:", source_counts)
print("Source weights:", source_weights)


# ------------------------------------------------------------
# Combine class balance and source balance
# ------------------------------------------------------------

train_sample_weights = np.asarray(
    [
        source_weights[int(source_id)]
        * class_weight_by_id[int(class_id)]
        for source_id, class_id in zip(
            train_sources_np,
            Y_train_np,
        )
    ],
    dtype=np.float32,
)

# Keep average loss scale around 1
train_sample_weights /= train_sample_weights.mean()

assert np.isfinite(train_sample_weights).all()
assert np.all(train_sample_weights > 0)
assert np.isclose(
    train_sample_weights.mean(),
    1.0,
    atol=1e-6,
)


# ------------------------------------------------------------
# Shape and integrity checks
# ------------------------------------------------------------

for name, X, y in [
    ("train", X_train_np, Y_train_np),
    ("VR validation", X_val_np, Y_val_np),
    ("EmojiHero validation", X_val_emojihero_np, Y_val_emojihero_np),
    ("VR test", X_test_vr_np, Y_test_vr_np),
    ("EmojiHero test", X_test_emojihero_np, Y_test_emojihero_np),
    ("all acted test", X_test_all_acted_np, Y_test_all_acted_np),
]:
    assert len(X) == len(y), name
    assert X.shape[1:] == (SEQ_LENGTH, len(AU_NAMES)), (
        name,
        X.shape,
    )
    assert np.isfinite(X).all(), name


# ------------------------------------------------------------
# Report effective source contributions
# ------------------------------------------------------------

source_names = {
    SOURCE_OLD_ACTED: "old acted",
    SOURCE_EMOJIHERO: "EmojiHero",
    SOURCE_NEW_ACTED: "new acted",
}

print("\nEffective training weights:")

for source_id, source_name in source_names.items():
    mask = train_sources_np == source_id

    print(
        f"{source_name:12s} | "
        f"n={mask.sum():4d} | "
        f"mean={train_sample_weights[mask].mean():.4f} | "
        f"total={train_sample_weights[mask].sum():.2f}"
    )


print("\nTensor shapes:")
print("Train:", X_train_np.shape)
print("VR validation:", X_val_np.shape)
print("EmojiHero validation:", X_val_emojihero_np.shape)
print("VR test:", X_test_vr_np.shape)
print("EmojiHero test:", X_test_emojihero_np.shape)

print("\nSample weights are ready.")
````

````python
# ======================================================================
# PART 5.7  HOW EXAGGERATED IS EACH EMOTION, IN EACH DATA COLLECTION?
# Old acted + new acted (15) + EmojiHero on one ~10 Hz grid. Unit = one person.
# Needs Parts 1-5 to have run. Does NOT change the training data.
# ======================================================================
import re
from scipy import stats
from matplotlib.patches import Patch

EX_EMOTIONS = [e.value for e in Emotion]            # same order as le / EmojiHero ids
EX_COLLECTIONS = ["old acted", "new acted", "EmojiHero"]
EX_SHARED = list(range(63))       # 63 signals all collections have (EmojiHero has no tongue)
EX_MOVING = 0.05                  # a signal "moves" if its 99th percentile > 0.05
EX_GATE = 0.25                    # same activity rule as `thresh`
EX_PALETTE = {"old acted": "#4C72B0", "new acted": "#DD8452", "EmojiHero": "#55A868"}
EX_LEGEND = [Patch(color=EX_PALETTE[c], label=c) for c in EX_COLLECTIONS]
EX_REGION_PATTERNS = {
    "brow": r"BrowLowerer|InnerBrowRaiser|OuterBrowRaiser",
    "eye": r"EyesClosed|LidTightener|UpperLidRaiser",
    "gaze": r"EyesLook", "cheek": r"Cheek", "nose": r"NoseWrinkler",
    "mouth": r"Lip|Mouth|Dimpler", "jaw": r"Jaw", "chin": r"ChinRaiser", "tongue": r"Tongue",
}
EX_REGIONS = {r: [i for i, n in enumerate(AU_NAMES)
                  if next(k for k, p in EX_REGION_PATTERNS.items() if re.search(p, n)) == r]
              for r in EX_REGION_PATTERNS}
EX_OUT = Path(output_path) / "exaggeration_analysis"
EX_OUT.mkdir(parents=True, exist_ok=True)

def ex_save(name):
    plt.tight_layout(); plt.savefig(EX_OUT / f"{name}.png", dpi=150); plt.show()

# ---------------------------------------------------------------- 1. collect
ex_records = []

for emo in Emotion:                                           # old acted
    for f in sorted(glob(f"{data_folder_path}/*/ACT_{emo.value}.csv")):
        df, _ = data_frame(f)
        t = pd.to_numeric(pd.read_csv(f).iloc[:, 0], errors="coerce").to_numpy(float)
        dur = float(np.nanmax(t) - np.nanmin(t))
        X = df[AU_NAMES].to_numpy(float)
        ex_records.append(dict(collection="old acted", person=subject_from_path(f),
                               emotion=emo.value, frames=np.nan_to_num(X), raw_frames=len(X),
                               duration_s=dur, rate_hz=(len(X) - 1) / dur if dur > 0 else np.nan,
                               nonfinite=int((~np.isfinite(X)).sum()), recorded=70))

ex_new = new_collection_records()                             # new acted
ex_new = ex_new[ex_new["recording_type"] == "ActingRecordings"]
for _, row in ex_new.iterrows():
    raw = pd.read_csv(row["path"])
    df = downsample_60hz_to_10hz_face_au(raw, AU_NAMES)
    Xr = raw.reindex(columns=AU_NAMES).apply(pd.to_numeric, errors="coerce").to_numpy(float)
    ex_records.append(dict(collection="new acted", person=row["participant"],
                           emotion=row["emotion"].value, frames=df[AU_NAMES].to_numpy(float),
                           raw_frames=len(raw), duration_s=len(raw) / 60.0, rate_hz=60.0,
                           nonfinite=int((~np.isfinite(Xr)).sum()),
                           recorded=int(sum(a in raw.columns for a in AU_NAMES))))

# EmojiHero: already averaged to 10 frames per sequence (cell 43)
# parent_seq = <timestamp>-<set>-<participant>-<level>-<emoji>-<emotion_id>
for split, df in (("train", train_em_hero_df), ("validation", valid_em_hero_df),
                  ("test", test_em_hero_df)):
    for seq, g in df.groupby("parent_seq", sort=False):
        ex_records.append(dict(collection="EmojiHero", person="EH-" + str(seq).split("-")[2],
                               emotion=EX_EMOTIONS[int(g["emotion"].iloc[0])], split=split,
                               frames=g[AU_NAMES].to_numpy(float), raw_frames=30,
                               duration_s=np.nan, rate_hz=np.nan, nonfinite=0, recorded=63))

# ---------------------------------------------------------------- 2. inventory
ex_inv = pd.DataFrame([{k: v for k, v in r.items() if k != "frames"} for r in ex_records])

print("== 1. WHAT IS IN EACH COLLECTION ==")
display(ex_inv.groupby("collection").agg(
    people=("person", "nunique"), recordings=("person", "size"),
    median_duration_s=("duration_s", "median"), median_raw_frames=("raw_frames", "median"),
    median_rate_hz=("rate_hz", "median"), signals_recorded=("recorded", "min"),
    nonfinite_values=("nonfinite", "sum")).reindex(EX_COLLECTIONS).round(2))
print("recordings per emotion:")
display(pd.crosstab(ex_inv["collection"], ex_inv["emotion"]).reindex(EX_COLLECTIONS)[EX_EMOTIONS])
print("recordings per person per emotion (min / median / max):")
display(ex_inv.groupby(["collection", "person", "emotion"]).size()
        .groupby("collection").agg(["min", "median", "max"]).reindex(EX_COLLECTIONS))

print("signals that never move in a collection:")
for c in EX_COLLECTIONS:
    F = np.concatenate([r["frames"] for r in ex_records if r["collection"] == c])
    dead = [AU_NAMES[i] for i in range(70) if np.quantile(F[:, i], 0.99) <= EX_MOVING]
    print(f"  {c:10s} ({70 - len(dead)} of 70 move): {', '.join(dead) or '-'}")

ex_mv = []
for (c, p), idx in ex_inv.groupby(["collection", "person"]).groups.items():
    F = np.concatenate([ex_records[i]["frames"] for i in idx])
    ex_mv.append(dict(collection=c, person=p,
                      signals_moving=int((np.quantile(F, 0.99, axis=0) > EX_MOVING).sum())))
ex_mv = pd.DataFrame(ex_mv)
print("signals moving per person (of 70):")
display(ex_mv.groupby("collection")["signals_moving"].describe().reindex(EX_COLLECTIONS).round(1))

# ---------------------------------------------------------------- 3. measures
def ex_measures(F, emotion):
    s = F[:, EX_SHARED].sum(axis=1)                    # per-frame face size (63 signals)
    gate = (F <= EX_GATE).all(1) if emotion == "Neutral" else (F > EX_GATE).any(1)
    out = dict(face_size=s.mean(), peak=np.quantile(s, 0.95),
               signals_used=int((np.quantile(F[:, EX_SHARED], 0.99, axis=0) > EX_MOVING).sum()),
               saturation=float((F[:, EX_SHARED] >= 0.95).mean()), gate_pass=float(gate.mean()))
    for reg, ix in EX_REGIONS.items():
        if reg != "tongue":
            out["region_" + reg] = F[:, ix].sum(axis=1).mean()
    return out

ex_m = pd.DataFrame([dict(collection=r["collection"], person=r["person"], emotion=r["emotion"],
                          **ex_measures(r["frames"], r["emotion"])) for r in ex_records])
ex_M = ex_m.groupby(["collection", "person", "emotion"], as_index=False).mean(numeric_only=True)
ex_neutral = ex_M[ex_M.emotion == "Neutral"].set_index(["collection", "person"])["face_size"]
ex_M["range_from_rest"] = [r.face_size - ex_neutral.get((r.collection, r.person), np.nan)
                           for r in ex_M.itertuples()]
ex_M.to_csv(EX_OUT / "per_person_per_emotion.csv", index=False)

print("\n== 2. MEDIAN FACE SIZE PER EMOTION ==")
display(ex_M.groupby(["collection", "emotion"])["face_size"].median().unstack()
        .reindex(EX_COLLECTIONS)[EX_EMOTIONS].round(2))

# ---------------------------------------------------------------- 4. statistics
ex_rng = np.random.default_rng(42)

def ex_boot_ci(x, n=2000):
    x = np.asarray(x)
    return np.quantile([np.median(ex_rng.choice(x, len(x))) for _ in range(n)], [.025, .975])

def ex_cliffs(x, y):
    x, y = np.asarray(x)[:, None], np.asarray(y)[None, :]
    return float(((x > y).sum() - (x < y).sum()) / (x.size * y.size))

def ex_holm(p):
    order, adj, run = np.argsort(p), np.empty(len(p)), 0.0
    for rank, i in enumerate(order):
        run = max(run, (len(p) - rank) * p[i]); adj[i] = min(1.0, run)
    return adj

def ex_stats(measure, emotions):
    pairs = [("old acted", "new acted"), ("old acted", "EmojiHero"), ("new acted", "EmojiHero")]
    rows = []
    for emo in emotions:
        g = {c: ex_M[(ex_M.collection == c) & (ex_M.emotion == emo)][measure].dropna()
             for c in EX_COLLECTIONS}
        row = dict(emotion=emo, kruskal_p=round(stats.kruskal(*g.values()).pvalue, 4))
        for c in EX_COLLECTIONS:
            lo, hi = ex_boot_ci(g[c]); row[c] = f"{np.median(g[c]):.2f} [{lo:.2f}-{hi:.2f}]"
        adj = ex_holm(np.array([stats.mannwhitneyu(g[a], g[b]).pvalue for a, b in pairs]))
        for (a, b), p in zip(pairs, adj):
            row[f"{a} vs {b}"] = f"p={p:.3f}, d={ex_cliffs(g[a], g[b]):+.2f}"
        rows.append(row)
    return pd.DataFrame(rows).set_index("emotion")

print("\n== 3. STATISTICS ==")
print("FACE SIZE: median [95% CI]; pairwise Mann-Whitney (Holm); d = Cliff's delta (-1..+1)")
display(ex_stats("face_size", EX_EMOTIONS))
print("RANGE FROM REST (emotion minus the person's own Neutral)")
display(ex_stats("range_from_rest", [e for e in EX_EMOTIONS if e != "Neutral"]))
ex_pp = ex_M[ex_M.emotion != "Neutral"].groupby(["collection", "person"])["face_size"].mean()
print("how different people are inside each collection (CV of face size):")
display(ex_pp.groupby("collection").agg(lambda v: v.std() / v.mean()).reindex(EX_COLLECTIONS).round(2))

# ---------------------------------------------------------------- 5. plots
# 1. face size per emotion, one dot per person
g = sns.catplot(data=ex_M, x="collection", y="face_size", hue="collection",
                col="emotion", col_order=EX_EMOTIONS, order=EX_COLLECTIONS,
                hue_order=EX_COLLECTIONS, kind="box", palette=EX_PALETTE, legend=False,
                col_wrap=4, height=3, showfliers=False)
g.map_dataframe(sns.stripplot, x="collection", y="face_size", order=EX_COLLECTIONS,
                color="black", size=3, alpha=.5)
g.set_xticklabels(rotation=30); g.fig.suptitle("Face size per emotion (1 dot = 1 person)", y=1.02)
ex_save("1_face_size_per_emotion")

# 2. range from rest
g = sns.catplot(data=ex_M[ex_M.emotion != "Neutral"], x="emotion", y="range_from_rest",
                hue="collection", hue_order=EX_COLLECTIONS, kind="box", palette=EX_PALETTE,
                height=4, aspect=2.2, showfliers=False)
g.fig.suptitle("How far each emotion moves from the person's own Neutral", y=1.02)
ex_save("2_range_from_rest")

# 3. every person ranked  (legend fixed)
ex_rank = ex_pp.reset_index().sort_values("face_size")
plt.figure(figsize=(14, 4))
plt.bar(range(len(ex_rank)), ex_rank.face_size, color=ex_rank.collection.map(EX_PALETTE))
plt.legend(handles=EX_LEGEND)
plt.xticks([]); plt.xlabel("people, smallest to biggest face"); plt.ylabel("face size")
plt.title("Who exaggerates? Every person ranked"); ex_save("3_people_ranked")

# 4. per-frame face size distribution (expressive emotions)
plt.figure(figsize=(9, 4))
for c in EX_COLLECTIONS:
    s = np.concatenate([r["frames"][:, EX_SHARED].sum(1) for r in ex_records
                        if r["collection"] == c and r["emotion"] != "Neutral"])
    sns.ecdfplot(s, label=c, color=EX_PALETTE[c])
plt.xlabel("per-frame face size (63 signals)"); plt.ylabel("share of frames")
plt.legend(handles=EX_LEGEND)
plt.title("How big are expressive frames?"); ex_save("4_frame_size_distribution")

# 5. which regions move, per emotion
ex_reg = [c for c in ex_M.columns if c.startswith("region_")]
fig, axes = plt.subplots(1, 3, figsize=(16, 4), sharey=True)
for ax, c in zip(axes, EX_COLLECTIONS):
    H = ex_M[ex_M.collection == c].groupby("emotion")[ex_reg].median().reindex(EX_EMOTIONS)
    H.columns = [x.replace("region_", "") for x in H.columns]
    sns.heatmap(H, ax=ax, cmap="viridis", vmin=0, annot=True, fmt=".2f", cbar=False)
    ax.set_title(c)
fig.suptitle("Face regions per emotion (median over people)"); ex_save("5_region_strength")

# 6. each person's own signals (95th percentile)
ex_P = []
for (c, p), idx in ex_inv.groupby(["collection", "person"]).groups.items():
    F = np.concatenate([ex_records[i]["frames"] for i in idx])
    ex_P.append(pd.Series(np.quantile(F, 0.95, axis=0), index=AU_NAMES, name=f"{c} | {p}"))
ex_P = pd.DataFrame(ex_P)
ex_P = ex_P.loc[sorted(ex_P.index, key=lambda k: EX_COLLECTIONS.index(k.split(" | ")[0]))]
ex_P.to_csv(EX_OUT / "per_person_signal_p95.csv")
plt.figure(figsize=(18, 12))
sns.heatmap(ex_P, cmap="magma", vmin=0, vmax=1, cbar_kws={"label": "95th percentile"})
plt.title("Each person's own signals (rows grouped by collection)"); ex_save("6_person_by_signal")

# 7. share of frames passing the 0.25 rule
g = sns.catplot(data=ex_M, x="emotion", y="gate_pass", hue="collection", order=EX_EMOTIONS,
                hue_order=EX_COLLECTIONS, kind="bar", palette=EX_PALETTE, height=4, aspect=2.2)
g.fig.suptitle("Share of frames that pass the 0.25 training rule", y=1.02)
ex_save("7_gate_pass_rate")

# ---------------------------------------------------------------- 6. summary
ex_med = ex_M.groupby(["emotion", "collection"])["face_size"].median().unstack()[EX_COLLECTIONS]
ex_summary = ex_med.round(2).assign(
    most_to_least=ex_med.apply(lambda r: " > ".join(r.sort_values(ascending=False).index), axis=1)
).reindex(EX_EMOTIONS)
print("\n== 4. SUMMARY: most to least exaggerated, per emotion ==")
display(ex_summary); ex_summary.to_csv(EX_OUT / "summary_face_size.csv")

ex_aug = (ex_M[ex_M.emotion != "Neutral"].groupby(["collection", "person"])
          .agg(face_size=("face_size", "mean"), peak=("peak", "mean")))
ex_aug["neutral_size"] = ex_neutral
ex_aug = ex_aug.join(ex_mv.set_index(["collection", "person"]))
ex_aug.to_csv(EX_OUT / "per_person_for_augmentation.csv")
print("== 5. PER PERSON (input for per-person volume augmentation) ==")
display(ex_aug.round(2))
print("CSV files and plots saved to", EX_OUT)
````

## 7.2 Architecture

### MoE-TCN: 3 TCN experts + Top-2 gate

### 7.2.1 Architecture definition

**`_tcn_block`** — dilated causal 1-D convolution -> batch norm -> ReLU -> dropout, twice, with a
residual connection (1x1 projection when the channel count changes).

**`build_expert_tcn`** — three stacked TCN blocks with dilation rates 1, 2, 4 and 128 filters each,
followed by global average pooling. With kernel size 3, this gives a receptive field that covers the
full 5-frame window.

**`Top2Gate`** — a small MLP produces one logit per expert. Logits are divided by a temperature
`tau` and softmaxed; the top 2 experts are kept, their weights renormalised, and scattered back into
a sparse `(B, 3)` weight vector. Two auxiliary losses are added:

- a **load-balancing loss** (`lb_coef`) penalising deviation from uniform expert usage, so no expert
  is starved
- an **entropy regulariser** (`ent_coef`) encouraging exploration of all experts early in training

**`GateTemperatureScheduler`** — anneals `tau` from 2.0 to 0.7 over 20 epochs: soft, exploratory
routing at the start, sharper and more specialised routing later.

**`build_moe_single_input`** — assembles three experts (`exp_nat`, `exp_act`, `exp_shared`), gates
them on the pooled input sequence, takes the gate-weighted sum, and passes it through a 256-unit
dense classifier to a 7-way softmax. Optimised with SGD (lr = 1e-3) and sparse categorical
cross-entropy.

**Note:** the expert-stacking and fusion steps use `Lambda` layers. Keras cannot deserialise these
from a saved `.keras` file, which is why every reload in this notebook rebuilds the architecture and
calls `load_weights()` (Section 8.3) rather than `load_model()`.

````python
import tensorflow as tf
from tensorflow.keras import layers, models, callbacks, backend as K

# ---- TCN expert blocks ----
def _tcn_block(x, filters, kernel_size, dilation_rate, dropout=0.2, name=None):
    res = x
    y = layers.Conv1D(filters, kernel_size, padding="causal",
                      dilation_rate=dilation_rate, name=None if name is None else name+"_c1")(x)
    y = layers.BatchNormalization(name=None if name is None else name+"_bn1")(y)
    y = layers.ReLU(name=None if name is None else name+"_relu1")(y)
    y = layers.Dropout(dropout, name=None if name is None else name+"_do1")(y)
    y = layers.Conv1D(filters, kernel_size, padding="causal",
                      dilation_rate=dilation_rate, name=None if name is None else name+"_c2")(y)
    y = layers.BatchNormalization(name=None if name is None else name+"_bn2")(y)
    if res.shape[-1] != filters:
        res = layers.Conv1D(filters, 1, padding="same", name=None if name is None else name+"_proj")(res)
        res = layers.BatchNormalization(name=None if name is None else name+"_proj_bn")(res)
    y = layers.Add(name=None if name is None else name+"_add")([y, res])
    y = layers.ReLU(name=None if name is None else name+"_relu2")(y)
    return y

def build_expert_tcn(x, name_prefix="exp"):
    y = _tcn_block(x, 128, 3, 1, 0.2, name=name_prefix+"_tcn1")
    y = _tcn_block(y, 128, 3, 2, 0.2, name=name_prefix+"_tcn2")
    y = _tcn_block(y, 128, 3, 4, 0.2, name=name_prefix+"_tcn3")
    y = layers.GlobalAveragePooling1D(name=name_prefix+"_gap")(y)
    return y  # (B, F)

# ---- Top-2 gate as a Keras Layer ----
class Top2Gate(layers.Layer):
    def __init__(self, num_experts=3, gate_hidden=128, tau_init=2.0,
                 lb_coef=0.02, ent_coef=0.01, name="gate", **kwargs):
        super().__init__(name=name, **kwargs)
        self.num_experts = num_experts
        self.gate_hidden = gate_hidden
        self.tau = tf.Variable(tau_init, dtype=tf.float32, trainable=False, name=f"{name}_tau")
        self.lb_coef = lb_coef
        self.ent_coef = ent_coef
        self.d1 = layers.Dense(gate_hidden, activation="relu", name=f"{name}_dense1")
        self.d2 = layers.Dense(num_experts, activation=None, name=f"{name}_logits")

    def call(self, gate_in, training=None):
        logits = self.d2(self.d1(gate_in))           # (B, E)
        w_soft = tf.nn.softmax(logits / self.tau, -1)  # (B, E)

        top2 = tf.math.top_k(w_soft, k=2)
        idx = top2.indices            # (B, 2)
        vals = tf.nn.softmax(top2.values, -1)  # re-norm top-2

        B = tf.shape(w_soft)[0]; E = tf.shape(w_soft)[1]
        row_ids = tf.repeat(tf.range(B), repeats=2)
        col_ids = tf.reshape(idx, (-1,))
        scatter_idx = tf.stack([row_ids, col_ids], axis=1)   # (B*2, 2)
        scatter_vals = tf.reshape(vals, (-1,))
        w_sparse = tf.scatter_nd(scatter_idx, scatter_vals, (B, self.num_experts))  # (B,E)

        # Load-balance loss (uniform usage)
        mean_usage = tf.reduce_mean(w_sparse, axis=0)
        uniform = tf.fill([self.num_experts], 1.0 / tf.cast(self.num_experts, tf.float32))
        lb_loss = tf.reduce_sum(tf.square(mean_usage - uniform))
        self.add_loss(self.lb_coef * lb_loss)

        # Entropy reg on soft weights (encourage exploration early)
        ent = -tf.reduce_mean(tf.reduce_sum(w_soft * tf.math.log(tf.clip_by_value(w_soft, 1e-8, 1.0)), axis=-1))
        self.add_loss(self.ent_coef * (-ent))  # minimize negative entropy

        return w_sparse, w_soft  # sparse mixture, full mixture (for diagnostics)

# ---- Temperature scheduler ----
class GateTemperatureScheduler(callbacks.Callback):
    def __init__(self, gate_layer, start_tau=2.0, end_tau=0.7, anneal_epochs=20):
        super().__init__()
        self.gate = gate_layer
        self.start_tau = float(start_tau)
        self.end_tau = float(end_tau)
        self.anneal_epochs = max(1, int(anneal_epochs))
    def on_epoch_begin(self, epoch, logs=None):
        p = min(1.0, epoch / self.anneal_epochs)
        tau = self.start_tau + p * (self.end_tau - self.start_tau)
        self.gate.tau.assign(tau)
        K.set_value(self.gate.tau, tau)
        if epoch % 5 == 0:
            print(f"[GateTemperatureScheduler] epoch={epoch} tau={tau:.3f}")

# ---- Single-input MoE builder (no domain hint) ----
def build_moe_single_input(
    input_shape, num_classes,
    tau_init=2.0, tau_final=0.7,
    gate_hidden=128, classifier_hidden=256,
    lb_coef=0.02, ent_coef=0.01,
    name="MoE_SingleInput"
):
    seq_in = layers.Input(shape=input_shape, name="seq")

    # Experts
    exp_nat = build_expert_tcn(seq_in, "exp_nat")
    exp_act = build_expert_tcn(seq_in, "exp_act")
    exp_shared = build_expert_tcn(seq_in, "exp_shared")

    # Stack experts into (B, E=3, F) using Lambda + Concatenate (avoid tf.stack on KerasTensors)
    e1 = layers.Lambda(lambda z: tf.expand_dims(z, axis=1), name="exp_nat_expand")(exp_nat)       # (B,1,F)
    e2 = layers.Lambda(lambda z: tf.expand_dims(z, axis=1), name="exp_act_expand")(exp_act)       # (B,1,F)
    e3 = layers.Lambda(lambda z: tf.expand_dims(z, axis=1), name="exp_share_expand")(exp_shared)  # (B,1,F)
    E_stack = layers.Concatenate(axis=1, name="experts_stack")([e1, e2, e3])  # (B,3,F)

    # Gate input: pooled sequence only (single input mode)
    gate_in = layers.GlobalAveragePooling1D(name="gate_pool")(seq_in)

    gate_layer = Top2Gate(num_experts=3, gate_hidden=gate_hidden,
                          tau_init=tau_init, lb_coef=lb_coef, ent_coef=ent_coef, name="gate")
    w_sparse, w_soft = gate_layer(gate_in)   # (B,3), (B,3)

    # Weighted sum: (B,3,1) * (B,3,F) -> sum over experts -> (B,F)
    w_expanded = layers.Lambda(lambda z: tf.expand_dims(z, -1), name="w_expand")(w_sparse)  # (B,3,1)
    fused = layers.Lambda(lambda zs: tf.reduce_sum(zs[0] * zs[1], axis=1), name="fuse")([w_expanded, E_stack])  # (B,F)

    x = layers.Dense(classifier_hidden, activation="relu", name="clf_dense")(fused)
    x = layers.Dropout(0.3, name="clf_do")(x)
    y_out = layers.Dense(num_classes, activation="softmax", name="y")(x)

    model = models.Model(seq_in, y_out, name=name)
    model.compile(optimizer=tf.keras.optimizers.SGD(1e-3),
                  loss="sparse_categorical_crossentropy",
                  metrics=["accuracy"])

    # Provide a temperature scheduler that targets this gate instance
    anneal_cb = GateTemperatureScheduler(gate_layer, start_tau=tau_init, end_tau=tau_final, anneal_epochs=20)
    return model, anneal_cb
````

````python
import os
import random
import numpy as np
import tensorflow as tf

from sklearn.metrics import f1_score
from tensorflow.keras import callbacks
from tensorflow.keras.callbacks import CSVLogger


# ============================================================
# Configuration
# ============================================================

RANDOM_SEED = 42

# One name per run, so model A (k0) and model B (k1) never overwrite each other.
NEW_TEACHER_TAG = f"teacher-70AU-acted-only-oldacted-emojihero-new15acted-volaug-k{AUG_COPIES}"

NEW_MODEL_PATH = os.path.join(
    models_path,
    NEW_TEACHER_TAG + ".keras",
)

NEW_CHECKPOINT_PATH = os.path.join(
    models_path,
    NEW_TEACHER_TAG + "_best.keras",
)

NEW_TRAINING_LOG_PATH = os.path.join(
    models_path,
    NEW_TEACHER_TAG + "_training_log.csv",
)

os.makedirs(models_path, exist_ok=True)


# ============================================================
# Reproducibility
# ============================================================

os.environ["PYTHONHASHSEED"] = str(RANDOM_SEED)

random.seed(RANDOM_SEED)
np.random.seed(RANDOM_SEED)
tf.keras.utils.set_random_seed(RANDOM_SEED)

try:
    tf.config.experimental.enable_op_determinism()
    print("TensorFlow deterministic operations enabled.")
except Exception as exc:
    print("Could not enable full TensorFlow determinism:", exc)


# ============================================================
# Tensor conversion helpers
# ============================================================

def ensure_sequence_tensor(values):
    """Convert sequence collections to float32 tensors."""

    if isinstance(values, np.ndarray):
        tensor = values
    else:
        tensor = stack_sequences(values)

    tensor = np.asarray(tensor, dtype=np.float32)

    if tensor.ndim != 3:
        raise ValueError(
            f"Expected a rank-3 sequence tensor, received {tensor.shape}"
        )

    return tensor


def ensure_label_vector(values):
    """Convert labels to a flat int64 vector."""

    if isinstance(values, np.ndarray):
        labels = values
    else:
        labels = to_numpy_labels(values)

    return np.asarray(labels, dtype=np.int64).reshape(-1)


# ============================================================
# Build two independent validation domains (EmojiHero is not used for
# checkpoint selection; it is reported separately after training)
# ============================================================

# Previous VR acted validation data
X_val_old_acted_np = ensure_sequence_tensor(X_val_act)
Y_val_old_acted_np = ensure_label_vector(Y_val_act)

# New 15-participant acted validation data
X_val_new_acted_np = ensure_sequence_tensor(X_val_new)
Y_val_new_acted_np = ensure_label_vector(Y_val_new)

# EmojiHero validation data
X_val_emojihero_np = ensure_sequence_tensor(X_val_em_hero)
Y_val_emojihero_np = ensure_label_vector(Y_val_em_hero)


validation_domains = [
    (
        "old_acted",
        X_val_old_acted_np,
        Y_val_old_acted_np,
    ),
    (
        "new_acted",
        X_val_new_acted_np,
        Y_val_new_acted_np,
    ),
    (
        "emojihero",
        X_val_emojihero_np,
        Y_val_emojihero_np,
    ),
]



for domain_name, X_domain, y_domain in validation_domains:
    assert X_domain.shape[1:] == (
        SEQ_LENGTH,
        len(AU_NAMES),
    ), (
        domain_name,
        X_domain.shape,
    )

    assert len(X_domain) == len(y_domain), (
        domain_name,
        len(X_domain),
        len(y_domain),
    )

    assert set(np.unique(y_domain)).issubset(
        set(range(num_classes))
    )

    print(
        f"{domain_name:12s} | "
        f"X={X_domain.shape} | "
        f"y={y_domain.shape} | "
        f"classes={np.unique(y_domain).tolist()}"
    )


# ============================================================
# Create the pooled validation set
#
# This set is used only for standard Keras validation loss and
# accuracy logging. It does not select the best checkpoint.
# ============================================================

X_val_all_acted_np = np.concatenate(
    [
        X_val_old_acted_np,
        X_val_new_acted_np,
        X_val_emojihero_np,
    ],
    axis=0,
)

Y_val_all_acted_np = np.concatenate(
    [
        Y_val_old_acted_np,
        Y_val_new_acted_np,
        Y_val_emojihero_np,
    ],
    axis=0,
)


print("\nPooled validation:")
print("X:", X_val_all_acted_np.shape)
print("y:", Y_val_all_acted_np.shape)


# ============================================================
# Domain-balanced macro-F1 callback
#
# Each validation domain contributes exactly one half of the
# checkpoint-selection score, regardless of its window count.
# ============================================================

class DomainBalancedMacroF1(callbacks.Callback):
    def __init__(
        self,
        validation_domains,
        num_classes,
        prediction_batch_size=256,
    ):
        super().__init__()

        self.validation_domains = validation_domains
        self.num_classes = num_classes
        self.prediction_batch_size = prediction_batch_size

    def on_epoch_end(self, epoch, logs=None):
        logs = logs if logs is not None else {}

        domain_scores = {}

        for domain_name, X_domain, y_domain in self.validation_domains:
            probabilities = self.model.predict(
                X_domain,
                batch_size=self.prediction_batch_size,
                verbose=0,
            )

            predictions = np.argmax(probabilities, axis=1)

            macro_f1 = f1_score(
                y_domain,
                predictions,
                labels=np.arange(self.num_classes),
                average="macro",
                zero_division=0,
            )

            macro_f1 = float(macro_f1)

            domain_scores[domain_name] = macro_f1
            logs[f"val_{domain_name}_macro_f1"] = macro_f1

        domain_balanced_macro_f1 = float(
            np.mean(list(domain_scores.values()))
        )

        logs[
            "val_domain_balanced_macro_f1"
        ] = domain_balanced_macro_f1

        print(
            "\nValidation macro-F1 | "
            f"old acted: {domain_scores['old_acted']:.4f} | "
            f"new acted: {domain_scores['new_acted']:.4f} | "
            f"EmojiHero: {domain_scores['emojihero']:.4f} | "
            f"domain balanced: {domain_balanced_macro_f1:.4f}"
        )



# ============================================================
# Locate the previously prepared source-and-class-balanced
# training sample weights
# ============================================================

sample_weight_candidates = [
    "train_sample_weights",
    "sample_weight_train",
    "sample_weights_train",
    "training_sample_weights",
    "sample_weights",
]

selected_sample_weights = None
selected_sample_weight_name = None

for candidate_name in sample_weight_candidates:
    if candidate_name in globals():
        candidate_value = np.asarray(
            globals()[candidate_name],
            dtype=np.float32,
        ).reshape(-1)

        if len(candidate_value) == len(X_train_np):
            selected_sample_weights = candidate_value
            selected_sample_weight_name = candidate_name
            break

if selected_sample_weights is None:
    raise NameError(
        "The source-and-class-balanced training sample weights "
        "were not found. Run the cell that prints "
        "'Sample weights are ready' before this training cell."
    )


assert len(X_train_np) == len(Y_train_np)
assert len(selected_sample_weights) == len(Y_train_np)
assert np.all(np.isfinite(selected_sample_weights))
assert np.all(selected_sample_weights > 0)

print("\nTraining data:")
print("X:", X_train_np.shape)
print("y:", Y_train_np.shape)
print("Sample-weight variable:", selected_sample_weight_name)
print("Sample-weight minimum:", selected_sample_weights.min())
print("Sample-weight maximum:", selected_sample_weights.max())
print("Sample-weight mean:", selected_sample_weights.mean())


# ============================================================
# Build a new model from scratch
#
# The previously trained model is not reused or fine-tuned.
# ============================================================

tf.keras.backend.clear_session()

random.seed(RANDOM_SEED)
np.random.seed(RANDOM_SEED)
tf.keras.utils.set_random_seed(RANDOM_SEED)

model, anneal_cb = build_moe_single_input(
    input_shape=(SEQ_LENGTH, input_size),
    num_classes=num_classes,
    tau_init=2.0,
    tau_final=0.7,
    gate_hidden=128,
    classifier_hidden=256,
    lb_coef=0.02,
    ent_coef=0.01,
)


# ============================================================
# Configure model-selection callbacks
#
# DomainBalancedMacroF1 must run before EarlyStopping,
# ModelCheckpoint, and CSVLogger so the custom metric is
# available to those callbacks at the end of every epoch.
# ============================================================

domain_f1_cb = DomainBalancedMacroF1(
    validation_domains=validation_domains,
    num_classes=num_classes,
    prediction_batch_size=256,
)

early_stopping_cb = callbacks.EarlyStopping(
    monitor="val_domain_balanced_macro_f1",
    mode="max",
    patience=8,
    restore_best_weights=True,
    verbose=1,
)

checkpoint_cb = callbacks.ModelCheckpoint(
    filepath=NEW_CHECKPOINT_PATH,
    monitor="val_domain_balanced_macro_f1",
    mode="max",
    save_best_only=True,
    verbose=1,
)

csv_logger_cb = CSVLogger(
    NEW_TRAINING_LOG_PATH,
    append=False,
)


# ============================================================
# Train
#
# Do not also pass class_weight here. The selected sample
# weights already contain both emotion-class balancing and
# source balancing.
# ============================================================

# Originals + AUG_COPIES freshly scaled copies per window, every epoch.
# The dataset supplies the batches and the sample weights, so batch_size
# and sample_weight are NOT passed to fit.
train_data = VolumeAugmentedWindows(
    X_train_np,
    Y_train_np,
    selected_sample_weights,
    copies=AUG_COPIES,
    batch_size=batch_size,
    seed=AUG_SEED,
)

print("Augmentation copies per window:", AUG_COPIES)
print("Windows per epoch:", len(train_data.source))
print("Steps per epoch:", len(train_data))

history = model.fit(
    train_data,
    validation_data=(
        X_val_all_acted_np,
        Y_val_all_acted_np,
    ),
    epochs=epochs,
    callbacks=[
        domain_f1_cb,
        early_stopping_cb,
        checkpoint_cb,
        csv_logger_cb,
        anneal_cb,
    ],
    verbose=1,
)


# ============================================================
# Save the model containing the restored best weights
# ============================================================

trained_model = model
trained_model.save(NEW_MODEL_PATH)

best_epoch_index = int(
    np.argmax(
        history.history["val_domain_balanced_macro_f1"]
    )
)

best_epoch_number = best_epoch_index + 1
best_validation_score = history.history[
    "val_domain_balanced_macro_f1"
][best_epoch_index]

print("\nTraining complete.")
print("Best epoch:", best_epoch_number)
print(
    "Best domain-balanced validation macro-F1:",
    f"{best_validation_score:.6f}",
)
print("Final model:", NEW_MODEL_PATH)
print("Best checkpoint:", NEW_CHECKPOINT_PATH)
print("Training log:", NEW_TRAINING_LOG_PATH)
print("\nNo test dataset was evaluated in this cell.")
````

````python
# ======================================================================
# FULL TEST EVALUATION OF ONE SAVED RUN (default: model C)
#
# Everything we reported for D, A and B, in one cell:
#   1. overall metrics per test set, with D's numbers next to them
#   2. per-emotion precision / recall / F1 per test set
#   3. confusion matrices
#   4. false "Neutral" calls on the VR test windows
#   5. accuracy by face size (VR test, expressive windows)
#   6. all_acted_pooled and all_acted_domain_balanced (overall + per emotion)
#
# The gate temperature is NOT stored in .keras files, so it is set to 0.7,
# the value the deployed D was exported with. Nothing is retrained or tuned.
# Needs: the corpus cells, Cell B (window_face_size, NEUTRAL_ID) and cell 70.
# ======================================================================

import os
from pathlib import Path

import numpy as np
import pandas as pd
from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    confusion_matrix,
    f1_score,
    log_loss,
    precision_recall_fscore_support,
)

EVAL_TAG = "teacher-70AU-acted-only-oldacted-emojihero-new15acted-volaug-k1"   # model C
EVAL_NAME = "C"
EVAL_WEIGHTS = os.path.join(models_path, EVAL_TAG + ".keras")
GATE_TAU = 0.7
assert os.path.exists(EVAL_WEIGHTS), EVAL_WEIGHTS

eval_model, _ = build_moe_single_input(
    input_shape=(SEQ_LENGTH, input_size),
    num_classes=num_classes,
    tau_init=2.0,
    tau_final=0.7,
    gate_hidden=128,
    classifier_hidden=256,
    lb_coef=0.02,
    ent_coef=0.01,
)
eval_model.load_weights(EVAL_WEIGHTS)
eval_model.get_layer("gate").tau.assign(GATE_TAU)   # tau is not saved in .keras files
print("Run:", EVAL_TAG, "| gate tau:", float(eval_model.get_layer("gate").tau.numpy()))

CLASS_NAMES = [emotion.value for emotion in Emotion]
CLASS_LABELS = np.arange(len(CLASS_NAMES))

# Deployed D on the same test windows (its final test evaluation).
D_RESULTS = {
    "old VR acted":              {"accuracy": 0.5215, "macro-F1": 0.4937, "log_loss": 1.2827},
    "new VR acted":              {"accuracy": 0.4934, "macro-F1": 0.4137, "log_loss": 1.4230},
    "VR combined":               {"accuracy": 0.5074, "macro-F1": 0.4545, "log_loss": 1.3530},
    "EmojiHero":                 {"accuracy": 0.5899, "macro-F1": 0.5657, "log_loss": 1.3178},
    "all_acted_pooled":          {"accuracy": 0.5725, "macro-F1": 0.5460, "log_loss": 1.3252},
    "all_acted_domain_balanced": {"accuracy": 0.5349, "macro-F1": 0.4910, "log_loss": 1.3412},
}

test_sets = {
    "old VR acted": (stack_sequences(X_test_act), to_numpy_labels(Y_test_act)),
    "new VR acted": (stack_sequences(X_test_new), to_numpy_labels(Y_test_new)),
    "EmojiHero": (X_test_emojihero_np, Y_test_emojihero_np),
}


def all_metrics(y, prob):
    pred = prob.argmax(axis=1)
    p = np.clip(prob, 1e-12, 1.0)
    p = p / p.sum(axis=1, keepdims=True)
    return {
        "windows": len(y),
        "accuracy": accuracy_score(y, pred),
        "balanced accuracy": balanced_accuracy_score(y, pred),
        "macro-F1": f1_score(y, pred, labels=CLASS_LABELS, average="macro", zero_division=0),
        "weighted F1": f1_score(y, pred, labels=CLASS_LABELS, average="weighted", zero_division=0),
        "log_loss": log_loss(y, p, labels=CLASS_LABELS),
    }


# ------------------------------------------------------------
# Predict each test set once
# ------------------------------------------------------------

results = {}
for name, (X, y) in test_sets.items():
    X = np.asarray(X, dtype=np.float32)
    y = np.asarray(y, dtype=np.int64).reshape(-1)
    results[name] = (X, y, eval_model.predict(X, batch_size=256, verbose=0))

results["VR combined"] = tuple(
    np.concatenate([results["old VR acted"][i], results["new VR acted"][i]]) for i in range(3)
)
results["all_acted_pooled"] = tuple(
    np.concatenate([results[s][i] for s in ("old VR acted", "new VR acted", "EmojiHero")])
    for i in range(3)
)


# ------------------------------------------------------------
# 1. Overall
# ------------------------------------------------------------

overall_rows = []
per_set = {}
for name in ("old VR acted", "new VR acted", "VR combined", "EmojiHero", "all_acted_pooled"):
    X, y, prob = results[name]
    per_set[name] = all_metrics(y, prob)
balanced = {k: np.mean([per_set[s][k] for s in ("old VR acted", "new VR acted", "EmojiHero")])
            for k in per_set["EmojiHero"] if k != "windows"}
balanced["windows"] = np.nan
per_set["all_acted_domain_balanced"] = balanced

for name, m in per_set.items():
    overall_rows.append({
        "test set": name,
        **m,
        "D macro-F1": D_RESULTS[name]["macro-F1"],
        f"{EVAL_NAME} - D": m["macro-F1"] - D_RESULTS[name]["macro-F1"],
        "D log_loss": D_RESULTS[name]["log_loss"],
    })
overall_df = pd.DataFrame(overall_rows)
print("\n1. OVERALL")
display(overall_df.round(4))


# ------------------------------------------------------------
# 2. Per emotion
# ------------------------------------------------------------

emotion_rows = []
for name in ("old VR acted", "new VR acted", "VR combined", "EmojiHero", "all_acted_pooled"):
    X, y, prob = results[name]
    precision, recall, f1, support = precision_recall_fscore_support(
        y, prob.argmax(axis=1), labels=CLASS_LABELS, zero_division=0,
    )
    for i, emotion in enumerate(CLASS_NAMES):
        emotion_rows.append({"test set": name, "emotion": emotion, "windows": int(support[i]),
                             "precision": precision[i], "recall": recall[i], "f1": f1[i]})
emotion_df = pd.DataFrame(emotion_rows)

base = emotion_df[emotion_df["test set"].isin(["old VR acted", "new VR acted", "EmojiHero"])]
balanced_emotion = base.groupby("emotion", sort=False)[["precision", "recall", "f1"]].mean().reset_index()
balanced_emotion.insert(0, "test set", "all_acted_domain_balanced")
balanced_emotion.insert(2, "windows", np.nan)
emotion_df = pd.concat([emotion_df, balanced_emotion], ignore_index=True)

for metric in ("f1", "recall", "precision"):
    table = emotion_df.pivot(index="emotion", columns="test set", values=metric).reindex(CLASS_NAMES)
    table = table[["old VR acted", "new VR acted", "VR combined", "EmojiHero",
                   "all_acted_pooled", "all_acted_domain_balanced"]]
    print(f"\n2. PER-EMOTION {metric.upper()}")
    display(table.round(3))


# ------------------------------------------------------------
# 3. Confusion matrices
# ------------------------------------------------------------

for name in ("old VR acted", "new VR acted", "VR combined", "EmojiHero"):
    X, y, prob = results[name]
    cm = pd.DataFrame(
        confusion_matrix(y, prob.argmax(axis=1), labels=CLASS_LABELS),
        index=[f"true {c}" for c in CLASS_NAMES],
        columns=CLASS_NAMES,
    )
    print(f"\n3. CONFUSION MATRIX, {name} (rows = true, columns = predicted)")
    display(cm)


# ------------------------------------------------------------
# 4. False "Neutral" calls (VR combined)
# ------------------------------------------------------------

X, y, prob = results["VR combined"]
pred = prob.argmax(axis=1)
expressive = y != NEUTRAL_ID
neutral_called = int(np.sum(expressive & (pred == NEUTRAL_ID)))
neutral_precision = float(np.mean(y[pred == NEUTRAL_ID] == NEUTRAL_ID))
print("\n4. VR COMBINED: expressive windows called Neutral:",
      f"{neutral_called} of {int(expressive.sum())}",
      "(D 39, A 53, B 27) | Neutral precision:", f"{neutral_precision:.3f}",
      "(D 0.688, A 0.621, B 0.757)")
print("   correct windows:", int(np.sum(pred == y)), "of", len(y), "(D 308, A 301, B 308)")


# ------------------------------------------------------------
# 5. Accuracy by face size (VR combined, expressive windows)
# ------------------------------------------------------------

size = window_face_size(X)[expressive]
correct = (pred == y)[expressive]
size_bin = pd.cut(size, bins=[0, 1.5, 2.5, 4.0, np.inf],
                  labels=["quiet (<1.5)", "1.5-2.5", "2.5-4.0", "loud (>4.0)"], right=False)
size_df = (pd.DataFrame({"face size": size_bin, "correct": correct})
           .groupby("face size", observed=False)["correct"]
           .agg(windows="size", accuracy="mean"))
print("\n5. VR COMBINED, EXPRESSIVE WINDOWS, ACCURACY BY FACE SIZE (B: 0.18 / 0.18 / 0.37 / 0.64)")
display(size_df.round(3))


# ------------------------------------------------------------
# Save
# ------------------------------------------------------------

EVAL_DIR = Path(output_path) / "volume_augmentation_eval"
EVAL_DIR.mkdir(parents=True, exist_ok=True)
overall_df.to_csv(EVAL_DIR / f"{EVAL_TAG}_overall.csv", index=False)
emotion_df.to_csv(EVAL_DIR / f"{EVAL_TAG}_per_emotion.csv", index=False)
size_df.to_csv(EVAL_DIR / f"{EVAL_TAG}_by_face_size.csv")
for name in ("old VR acted", "new VR acted", "EmojiHero"):
    X, y, prob = results[name]
    np.savez_compressed(EVAL_DIR / f"{EVAL_TAG}_{name.replace(' ', '_')}_predictions.npz",
                        y_true=y, y_probability=prob, class_names=np.asarray(CLASS_NAMES))
print("\nSaved to:", EVAL_DIR)
````

````python
import os
import random
import numpy as np
import tensorflow as tf

from sklearn.metrics import f1_score
from tensorflow.keras import callbacks
from tensorflow.keras.callbacks import CSVLogger


# ============================================================
# Configuration
# ============================================================

RANDOM_SEED = 42

# One name per run, so model A (k0) and model B (k1) never overwrite each other.
NEW_TEACHER_TAG = f"teacher-70AU-oldacted-new15acted-volaug-k{AUG_COPIES}"

NEW_MODEL_PATH = os.path.join(
    models_path,
    NEW_TEACHER_TAG + ".keras",
)

NEW_CHECKPOINT_PATH = os.path.join(
    models_path,
    NEW_TEACHER_TAG + "_best.keras",
)

NEW_TRAINING_LOG_PATH = os.path.join(
    models_path,
    NEW_TEACHER_TAG + "_training_log.csv",
)

os.makedirs(models_path, exist_ok=True)


# ============================================================
# Reproducibility
# ============================================================

os.environ["PYTHONHASHSEED"] = str(RANDOM_SEED)

random.seed(RANDOM_SEED)
np.random.seed(RANDOM_SEED)
tf.keras.utils.set_random_seed(RANDOM_SEED)

try:
    tf.config.experimental.enable_op_determinism()
    print("TensorFlow deterministic operations enabled.")
except Exception as exc:
    print("Could not enable full TensorFlow determinism:", exc)


# ============================================================
# Tensor conversion helpers
# ============================================================

def ensure_sequence_tensor(values):
    """Convert sequence collections to float32 tensors."""

    if isinstance(values, np.ndarray):
        tensor = values
    else:
        tensor = stack_sequences(values)

    tensor = np.asarray(tensor, dtype=np.float32)

    if tensor.ndim != 3:
        raise ValueError(
            f"Expected a rank-3 sequence tensor, received {tensor.shape}"
        )

    return tensor


def ensure_label_vector(values):
    """Convert labels to a flat int64 vector."""

    if isinstance(values, np.ndarray):
        labels = values
    else:
        labels = to_numpy_labels(values)

    return np.asarray(labels, dtype=np.int64).reshape(-1)


# ============================================================
# Build two independent validation domains (EmojiHero is not used for
# checkpoint selection; it is reported separately after training)
# ============================================================

# Previous VR acted validation data
X_val_old_acted_np = ensure_sequence_tensor(X_val_act)
Y_val_old_acted_np = ensure_label_vector(Y_val_act)

# New 15-participant acted validation data
X_val_new_acted_np = ensure_sequence_tensor(X_val_new)
Y_val_new_acted_np = ensure_label_vector(Y_val_new)

# EmojiHero validation data
X_val_emojihero_np = ensure_sequence_tensor(X_val_em_hero)
Y_val_emojihero_np = ensure_label_vector(Y_val_em_hero)


validation_domains = [
    (
        "old_acted",
        X_val_old_acted_np,
        Y_val_old_acted_np,
    ),
    (
        "new_acted",
        X_val_new_acted_np,
        Y_val_new_acted_np,
    ),
]


for domain_name, X_domain, y_domain in validation_domains:
    assert X_domain.shape[1:] == (
        SEQ_LENGTH,
        len(AU_NAMES),
    ), (
        domain_name,
        X_domain.shape,
    )

    assert len(X_domain) == len(y_domain), (
        domain_name,
        len(X_domain),
        len(y_domain),
    )

    assert set(np.unique(y_domain)).issubset(
        set(range(num_classes))
    )

    print(
        f"{domain_name:12s} | "
        f"X={X_domain.shape} | "
        f"y={y_domain.shape} | "
        f"classes={np.unique(y_domain).tolist()}"
    )


# ============================================================
# Create the pooled validation set
#
# This set is used only for standard Keras validation loss and
# accuracy logging. It does not select the best checkpoint.
# ============================================================

X_val_all_acted_np = np.concatenate(
    [
        X_val_old_acted_np,
        X_val_new_acted_np,
    ],
    axis=0,
)

Y_val_all_acted_np = np.concatenate(
    [
        Y_val_old_acted_np,
        Y_val_new_acted_np,
    ],
    axis=0,
)

print("\nPooled validation:")
print("X:", X_val_all_acted_np.shape)
print("y:", Y_val_all_acted_np.shape)


# ============================================================
# Domain-balanced macro-F1 callback
#
# Each validation domain contributes exactly one half of the
# checkpoint-selection score, regardless of its window count.
# ============================================================

class DomainBalancedMacroF1(callbacks.Callback):
    def __init__(
        self,
        validation_domains,
        num_classes,
        prediction_batch_size=256,
    ):
        super().__init__()

        self.validation_domains = validation_domains
        self.num_classes = num_classes
        self.prediction_batch_size = prediction_batch_size

    def on_epoch_end(self, epoch, logs=None):
        logs = logs if logs is not None else {}

        domain_scores = {}

        for domain_name, X_domain, y_domain in self.validation_domains:
            probabilities = self.model.predict(
                X_domain,
                batch_size=self.prediction_batch_size,
                verbose=0,
            )

            predictions = np.argmax(probabilities, axis=1)

            macro_f1 = f1_score(
                y_domain,
                predictions,
                labels=np.arange(self.num_classes),
                average="macro",
                zero_division=0,
            )

            macro_f1 = float(macro_f1)

            domain_scores[domain_name] = macro_f1
            logs[f"val_{domain_name}_macro_f1"] = macro_f1

        domain_balanced_macro_f1 = float(
            np.mean(list(domain_scores.values()))
        )

        logs[
            "val_domain_balanced_macro_f1"
        ] = domain_balanced_macro_f1

        print(
            "\nValidation macro-F1 | "
            f"old acted: {domain_scores['old_acted']:.4f} | "
            f"new acted: {domain_scores['new_acted']:.4f} | "
            f"domain balanced: {domain_balanced_macro_f1:.4f}"
        )


# ============================================================
# Locate the previously prepared source-and-class-balanced
# training sample weights
# ============================================================

sample_weight_candidates = [
    "train_sample_weights",
    "sample_weight_train",
    "sample_weights_train",
    "training_sample_weights",
    "sample_weights",
]

selected_sample_weights = None
selected_sample_weight_name = None

for candidate_name in sample_weight_candidates:
    if candidate_name in globals():
        candidate_value = np.asarray(
            globals()[candidate_name],
            dtype=np.float32,
        ).reshape(-1)

        if len(candidate_value) == len(X_train_np):
            selected_sample_weights = candidate_value
            selected_sample_weight_name = candidate_name
            break

if selected_sample_weights is None:
    raise NameError(
        "The source-and-class-balanced training sample weights "
        "were not found. Run the cell that prints "
        "'Sample weights are ready' before this training cell."
    )


assert len(X_train_np) == len(Y_train_np)
assert len(selected_sample_weights) == len(Y_train_np)
assert np.all(np.isfinite(selected_sample_weights))
assert np.all(selected_sample_weights > 0)

print("\nTraining data:")
print("X:", X_train_np.shape)
print("y:", Y_train_np.shape)
print("Sample-weight variable:", selected_sample_weight_name)
print("Sample-weight minimum:", selected_sample_weights.min())
print("Sample-weight maximum:", selected_sample_weights.max())
print("Sample-weight mean:", selected_sample_weights.mean())


# ============================================================
# Build a new model from scratch
#
# The previously trained model is not reused or fine-tuned.
# ============================================================

tf.keras.backend.clear_session()

random.seed(RANDOM_SEED)
np.random.seed(RANDOM_SEED)
tf.keras.utils.set_random_seed(RANDOM_SEED)

model, anneal_cb = build_moe_single_input(
    input_shape=(SEQ_LENGTH, input_size),
    num_classes=num_classes,
    tau_init=2.0,
    tau_final=0.7,
    gate_hidden=128,
    classifier_hidden=256,
    lb_coef=0.02,
    ent_coef=0.01,
)


# ============================================================
# Configure model-selection callbacks
#
# DomainBalancedMacroF1 must run before EarlyStopping,
# ModelCheckpoint, and CSVLogger so the custom metric is
# available to those callbacks at the end of every epoch.
# ============================================================

domain_f1_cb = DomainBalancedMacroF1(
    validation_domains=validation_domains,
    num_classes=num_classes,
    prediction_batch_size=256,
)

early_stopping_cb = callbacks.EarlyStopping(
    monitor="val_domain_balanced_macro_f1",
    mode="max",
    patience=8,
    restore_best_weights=True,
    verbose=1,
)

checkpoint_cb = callbacks.ModelCheckpoint(
    filepath=NEW_CHECKPOINT_PATH,
    monitor="val_domain_balanced_macro_f1",
    mode="max",
    save_best_only=True,
    verbose=1,
)

csv_logger_cb = CSVLogger(
    NEW_TRAINING_LOG_PATH,
    append=False,
)


# ============================================================
# Train
#
# Do not also pass class_weight here. The selected sample
# weights already contain both emotion-class balancing and
# source balancing.
# ============================================================

# Originals + AUG_COPIES freshly scaled copies per window, every epoch.
# The dataset supplies the batches and the sample weights, so batch_size
# and sample_weight are NOT passed to fit.
train_data = VolumeAugmentedWindows(
    X_train_np,
    Y_train_np,
    selected_sample_weights,
    copies=AUG_COPIES,
    batch_size=batch_size,
    seed=AUG_SEED,
)

print("Augmentation copies per window:", AUG_COPIES)
print("Windows per epoch:", len(train_data.source))
print("Steps per epoch:", len(train_data))

history = model.fit(
    train_data,
    validation_data=(
        X_val_all_acted_np,
        Y_val_all_acted_np,
    ),
    epochs=epochs,
    callbacks=[
        domain_f1_cb,
        early_stopping_cb,
        checkpoint_cb,
        csv_logger_cb,
        anneal_cb,
    ],
    verbose=1,
)


# ============================================================
# Save the model containing the restored best weights
# ============================================================

trained_model = model
trained_model.save(NEW_MODEL_PATH)

best_epoch_index = int(
    np.argmax(
        history.history["val_domain_balanced_macro_f1"]
    )
)

best_epoch_number = best_epoch_index + 1
best_validation_score = history.history[
    "val_domain_balanced_macro_f1"
][best_epoch_index]

print("\nTraining complete.")
print("Best epoch:", best_epoch_number)
print(
    "Best domain-balanced validation macro-F1:",
    f"{best_validation_score:.6f}",
)
print("Final model:", NEW_MODEL_PATH)
print("Best checkpoint:", NEW_CHECKPOINT_PATH)
print("Training log:", NEW_TRAINING_LOG_PATH)
print("\nNo test dataset was evaluated in this cell.")
````

````python
# ======================================================================
# TEST EVALUATION OF ONE SAVED RUN (window level, Keras weights)
#
# Change EVAL_COPIES to pick the run: 1 = model B (augmented), 0 = model A.
# Loads the saved best weights, so it works even after the other model
# has been trained in the same session. Nothing is retrained or tuned here.
# ======================================================================

import os
from pathlib import Path

import numpy as np
import pandas as pd
from sklearn.metrics import accuracy_score, confusion_matrix, f1_score, recall_score

EVAL_COPIES = 1
EVAL_TAG = f"teacher-70AU-oldacted-new15acted-volaug-k{EVAL_COPIES}"
EVAL_WEIGHTS = os.path.join(models_path, EVAL_TAG + ".keras")
assert os.path.exists(EVAL_WEIGHTS), EVAL_WEIGHTS

# The architecture uses Lambda layers, so rebuild it and load the weights.
eval_model, _ = build_moe_single_input(
    input_shape=(SEQ_LENGTH, input_size),
    num_classes=num_classes,
    tau_init=2.0,
    tau_final=0.7,
    gate_hidden=128,
    classifier_hidden=256,
    lb_coef=0.02,
    ent_coef=0.01,
)
eval_model.load_weights(EVAL_WEIGHTS)

CLASS_NAMES = [emotion.value for emotion in Emotion]
CLASS_LABELS = np.arange(len(CLASS_NAMES))

# Deployed acted-only teacher (trained WITH EmojiHero), same test windows.
DEPLOYED_MACRO_F1 = {
    "old VR acted": 0.4937,
    "new VR acted": 0.4137,
    "VR combined": 0.4545,
    "EmojiHero": 0.5657,
}

test_sets = {
    "old VR acted": (stack_sequences(X_test_act), to_numpy_labels(Y_test_act)),
    "new VR acted": (stack_sequences(X_test_new), to_numpy_labels(Y_test_new)),
    "VR combined": (X_test_vr_np, Y_test_vr_np),
    "EmojiHero": (X_test_emojihero_np, Y_test_emojihero_np),
}

rows, recall_rows, predictions = [], [], {}
for name, (X, y) in test_sets.items():
    X = np.asarray(X, dtype=np.float32)
    y = np.asarray(y, dtype=np.int64).reshape(-1)
    pred = eval_model.predict(X, batch_size=256, verbose=0).argmax(axis=1)
    predictions[name] = (X, y, pred)

    f1 = f1_score(y, pred, labels=CLASS_LABELS, average="macro", zero_division=0)
    rows.append({
        "test set": name,
        "windows": len(y),
        "accuracy": accuracy_score(y, pred),
        "macro-F1": f1,
        "deployed macro-F1": DEPLOYED_MACRO_F1[name],
        "change": f1 - DEPLOYED_MACRO_F1[name],
    })
    recalls = recall_score(y, pred, labels=CLASS_LABELS, average=None, zero_division=0)
    recall_rows.append({"test set": name, **dict(zip(CLASS_NAMES, recalls))})

summary_df = pd.DataFrame(rows)
recall_df = pd.DataFrame(recall_rows)

print("Run:", EVAL_TAG)
display(summary_df.round(4))
print("Recall per emotion:")
display(recall_df.round(2))

for name in ("old VR acted", "new VR acted", "EmojiHero"):
    _, y, pred = predictions[name]
    cm = pd.DataFrame(
        confusion_matrix(y, pred, labels=CLASS_LABELS),
        index=[f"true {c}" for c in CLASS_NAMES],
        columns=CLASS_NAMES,
    )
    print(f"\nConfusion matrix, {name} (rows = true, columns = predicted):")
    display(cm)

# Accuracy by face size on the VR test windows (expressive emotions only):
# does the model do better on quiet faces?
X, y, pred = predictions["VR combined"]
expressive = y != NEUTRAL_ID
size = window_face_size(X)[expressive]
correct = (pred == y)[expressive]
bins = [0, 1.5, 2.5, 4.0, np.inf]
labels = ["quiet (<1.5)", "1.5-2.5", "2.5-4.0", "loud (>4.0)"]
size_bin = pd.cut(size, bins=bins, labels=labels, right=False)
size_df = (
    pd.DataFrame({"face size": size_bin, "correct": correct})
    .groupby("face size", observed=False)["correct"]
    .agg(windows="size", accuracy="mean")
)
print("\nVR test, expressive windows, accuracy by face size:")
display(size_df.round(3))

EVAL_DIR = Path(output_path) / "volume_augmentation_eval"
EVAL_DIR.mkdir(parents=True, exist_ok=True)
summary_df.to_csv(EVAL_DIR / f"{EVAL_TAG}_test_summary.csv", index=False)
recall_df.to_csv(EVAL_DIR / f"{EVAL_TAG}_test_recall.csv", index=False)
size_df.to_csv(EVAL_DIR / f"{EVAL_TAG}_test_by_face_size.csv")
print("\nSaved to:", EVAL_DIR)
````

````python
# ======================================================================
# PER-EMOTION TEST METRICS, MODEL A (k0) vs MODEL B (k1)
#
# Uses whichever of the two runs has been saved. With both, it adds a
# B - A column. Window level; nothing is retrained or tuned here.
# ======================================================================

import os
from pathlib import Path

import numpy as np
import pandas as pd
from sklearn.metrics import precision_recall_fscore_support

CLASS_NAMES = [emotion.value for emotion in Emotion]
CLASS_LABELS = np.arange(len(CLASS_NAMES))
RUNS = {"A (no aug)": 0, "B (aug)": 1}

per_emotion_test_sets = {
    "old VR acted": (stack_sequences(X_test_act), to_numpy_labels(Y_test_act)),
    "new VR acted": (stack_sequences(X_test_new), to_numpy_labels(Y_test_new)),
    "VR combined": (X_test_vr_np, Y_test_vr_np),
    "EmojiHero": (X_test_emojihero_np, Y_test_emojihero_np),
}


def load_run(copies):
    weights = os.path.join(
        models_path,
        f"teacher-70AU-oldacted-new15acted-volaug-k{copies}.keras",
    )
    if not os.path.exists(weights):
        return None
    run_model, _ = build_moe_single_input(
        input_shape=(SEQ_LENGTH, input_size),
        num_classes=num_classes,
        tau_init=2.0,
        tau_final=0.7,
        gate_hidden=128,
        classifier_hidden=256,
        lb_coef=0.02,
        ent_coef=0.01,
    )
    run_model.load_weights(weights)
    return run_model


per_emotion_rows = []
for run_name, copies in RUNS.items():
    run_model = load_run(copies)
    if run_model is None:
        print(f"{run_name}: not trained yet, skipped")
        continue
    for set_name, (X, y) in per_emotion_test_sets.items():
        X = np.asarray(X, dtype=np.float32)
        y = np.asarray(y, dtype=np.int64).reshape(-1)
        pred = run_model.predict(X, batch_size=256, verbose=0).argmax(axis=1)
        precision, recall, f1, support = precision_recall_fscore_support(
            y, pred, labels=CLASS_LABELS, zero_division=0,
        )
        for i, emotion in enumerate(CLASS_NAMES):
            per_emotion_rows.append({
                "run": run_name,
                "test set": set_name,
                "emotion": emotion,
                "windows": int(support[i]),
                "precision": precision[i],
                "recall": recall[i],
                "f1": f1[i],
            })

per_emotion_df = pd.DataFrame(per_emotion_rows)

for set_name in per_emotion_test_sets:
    part = per_emotion_df[per_emotion_df["test set"] == set_name]
    table = part.pivot(index="emotion", columns="run", values=["recall", "f1"])
    table = table.reindex(CLASS_NAMES)
    if {"A (no aug)", "B (aug)"} <= set(part["run"]):
        for metric in ("recall", "f1"):
            table[(metric, "B - A")] = (
                table[(metric, "B (aug)")] - table[(metric, "A (no aug)")]
            )
        table = table.sort_index(axis=1)
    windows = part.drop_duplicates("emotion").set_index("emotion")["windows"]
    table.insert(0, ("", "windows"), windows.reindex(CLASS_NAMES))
    print(f"\n{set_name}: per-emotion recall and F1")
    display(table.round(3))

EVAL_DIR = Path(output_path) / "volume_augmentation_eval"
EVAL_DIR.mkdir(parents=True, exist_ok=True)
per_emotion_df.to_csv(EVAL_DIR / "per_emotion_test_metrics.csv", index=False)
print("\nSaved to:", EVAL_DIR / "per_emotion_test_metrics.csv")
````

````python
# ======================================================================
# POOLED AND DOMAIN-BALANCED TEST METRICS, MODEL A (k0) vs MODEL B (k1)
#
# Same definitions as the original final test evaluation:
#   all_acted_pooled          = old VR + new VR + EmojiHero test windows, scored once
#   all_acted_domain_balanced = mean of the three test sets' scores
# Needs Cell E (load_run, per_emotion_test_sets) to have run.
# ======================================================================

from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    f1_score,
    log_loss,
    precision_score,
    recall_score,
)


def all_metrics(y, prob):
    pred = prob.argmax(axis=1)
    prob = np.clip(prob, 1e-12, 1.0)
    prob = prob / prob.sum(axis=1, keepdims=True)
    return {
        "n_windows": len(y),
        "accuracy": accuracy_score(y, pred),
        "balanced_accuracy": balanced_accuracy_score(y, pred),
        "precision_macro": precision_score(y, pred, labels=CLASS_LABELS, average="macro", zero_division=0),
        "recall_macro": recall_score(y, pred, labels=CLASS_LABELS, average="macro", zero_division=0),
        "f1_macro": f1_score(y, pred, labels=CLASS_LABELS, average="macro", zero_division=0),
        "f1_weighted": f1_score(y, pred, labels=CLASS_LABELS, average="weighted", zero_division=0),
        "log_loss": log_loss(y, prob, labels=CLASS_LABELS),
    }


pooled_rows = []
for run_name, copies in RUNS.items():
    run_model = load_run(copies)
    if run_model is None:
        print(f"{run_name}: not trained yet, skipped")
        continue

    ys, probs, per_set = [], [], []
    for set_name in ("old VR acted", "new VR acted", "EmojiHero"):
        X, y = per_emotion_test_sets[set_name]
        X = np.asarray(X, dtype=np.float32)
        y = np.asarray(y, dtype=np.int64).reshape(-1)
        prob = run_model.predict(X, batch_size=256, verbose=0)
        ys.append(y)
        probs.append(prob)
        per_set.append(all_metrics(y, prob))

    balanced = {k: np.mean([m[k] for m in per_set]) for k in per_set[0] if k != "n_windows"}
    pooled_rows.append({"run": run_name, "dataset": "all_acted_domain_balanced",
                        "n_windows": np.nan, **balanced})
    pooled_rows.append({"run": run_name, "dataset": "all_acted_pooled",
                        **all_metrics(np.concatenate(ys), np.concatenate(probs))})

pooled_df = pd.DataFrame(pooled_rows)
display(pooled_df.round(4))

EVAL_DIR = Path(output_path) / "volume_augmentation_eval"
EVAL_DIR.mkdir(parents=True, exist_ok=True)
pooled_df.to_csv(EVAL_DIR / "pooled_test_metrics.csv", index=False)
print("Saved to:", EVAL_DIR / "pooled_test_metrics.csv")
````

````python
# ======================================================================
# PER-EMOTION POOLED AND DOMAIN-BALANCED METRICS, MODEL A (k0) vs MODEL B (k1)
#
#   pooled           = old VR + new VR + EmojiHero windows together
#   domain balanced  = mean of the three test sets' per-emotion scores
# Needs Cell E (load_run, per_emotion_test_sets) to have run.
# ======================================================================

from sklearn.metrics import precision_recall_fscore_support

POOL_SETS = ("old VR acted", "new VR acted", "EmojiHero")

pooled_emotion_rows = []
for run_name, copies in RUNS.items():
    run_model = load_run(copies)
    if run_model is None:
        print(f"{run_name}: not trained yet, skipped")
        continue

    ys, preds, per_set = [], [], []
    for set_name in POOL_SETS:
        X, y = per_emotion_test_sets[set_name]
        X = np.asarray(X, dtype=np.float32)
        y = np.asarray(y, dtype=np.int64).reshape(-1)
        pred = run_model.predict(X, batch_size=256, verbose=0).argmax(axis=1)
        ys.append(y)
        preds.append(pred)
        per_set.append(precision_recall_fscore_support(
            y, pred, labels=CLASS_LABELS, zero_division=0)[:3])

    pooled = precision_recall_fscore_support(
        np.concatenate(ys), np.concatenate(preds),
        labels=CLASS_LABELS, zero_division=0,
    )
    balanced = np.mean(np.asarray(per_set), axis=0)   # (3 metrics, 7 classes)

    for i, emotion in enumerate(CLASS_NAMES):
        pooled_emotion_rows.append({
            "run": run_name, "dataset": "all_acted_pooled", "emotion": emotion,
            "windows": int(pooled[3][i]),
            "precision": pooled[0][i], "recall": pooled[1][i], "f1": pooled[2][i],
        })
        pooled_emotion_rows.append({
            "run": run_name, "dataset": "all_acted_domain_balanced", "emotion": emotion,
            "windows": np.nan,
            "precision": balanced[0][i], "recall": balanced[1][i], "f1": balanced[2][i],
        })

pooled_emotion_df = pd.DataFrame(pooled_emotion_rows)

for dataset in ("all_acted_pooled", "all_acted_domain_balanced"):
    part = pooled_emotion_df[pooled_emotion_df["dataset"] == dataset]
    table = part.pivot(index="emotion", columns="run", values=["recall", "f1"]).reindex(CLASS_NAMES)
    if {"A (no aug)", "B (aug)"} <= set(part["run"]):
        for metric in ("recall", "f1"):
            table[(metric, "B - A")] = table[(metric, "B (aug)")] - table[(metric, "A (no aug)")]
        table = table.sort_index(axis=1)
    print(f"\n{dataset}: per-emotion recall and F1")
    display(table.round(3))

EVAL_DIR = Path(output_path) / "volume_augmentation_eval"
EVAL_DIR.mkdir(parents=True, exist_ok=True)
pooled_emotion_df.to_csv(EVAL_DIR / "pooled_per_emotion_test_metrics.csv", index=False)
print("\nSaved to:", EVAL_DIR / "pooled_per_emotion_test_metrics.csv")
````

````python
# ======================================================================
# CELL H (STANDALONE): LOAD MODEL B FROM DRIVE AND EXPORT IT TO ONNX
#
# Works in a fresh runtime: run only Cell H0 (the installs) before it.
# No data is loaded. Keras <-> ONNX parity is checked on synthetic
# windows (values in [0, 1], like real AU windows), so no test data is used.
#
# Same graph interface as the deployed sparseVR_teacher_model_acted_only.onnx:
# input "seq" (batch, 5, 70), output "probabilities" (batch, 7), opset 17.
# ======================================================================

from google.colab import drive
drive.mount("/content/drive")

import hashlib
import json
from pathlib import Path

import numpy as np
import onnx
import onnxruntime as ort
import tensorflow as tf
import tf2onnx

MODELS_DIR = Path(
    "/content/drive/MyDrive/outputs_15_participants_70AU-NOT_normalize_teacher_data-"
    "self_filtered_NATURAL_ONLY-EmojiHero_Fixed-Only_Acted_data_training-Volume_Augmentation_with_EmojiHero/models"
)
EXPORT_TAG = "teacher-70AU-acted-only-oldacted-emojihero-new15acted-volaug-k1"     # model C     # model B; "-k0" = model A
EXPORT_WEIGHTS = MODELS_DIR / f"{EXPORT_TAG}.keras"
ONNX_EXPORT_PATH = MODELS_DIR / f"{EXPORT_TAG}.onnx"
ONNX_EXPORT_METADATA_PATH = MODELS_DIR / f"{EXPORT_TAG}.onnx.metadata.json"
OPSET_VERSION = 17

assert EXPORT_WEIGHTS.exists(), f"Not found: {EXPORT_WEIGHTS}"

SEQ_LENGTH = 5
CLASS_ORDER = ["Anger", "Disgust", "Fear", "Happiness", "Neutral", "Sadness", "Surprise"]
num_classes = len(CLASS_ORDER)

AU_NAMES = [
    "BrowLowererL",
    "BrowLowererR",
    "CheekPuffL",
    "CheekPuffR",
    "CheekRaiserL",
    "CheekRaiserR",
    "CheekSuckL",
    "CheekSuckR",
    "ChinRaiserB",
    "ChinRaiserT",
    "DimplerL",
    "DimplerR",
    "EyesClosedL",
    "EyesClosedR",
    "EyesLookDownL",
    "EyesLookDownR",
    "EyesLookLeftL",
    "EyesLookLeftR",
    "EyesLookRightL",
    "EyesLookRightR",
    "EyesLookUpL",
    "EyesLookUpR",
    "InnerBrowRaiserL",
    "InnerBrowRaiserR",
    "JawDrop",
    "JawSidewaysLeft",
    "JawSidewaysRight",
    "JawThrust",
    "LidTightenerL",
    "LidTightenerR",
    "LipCornerDepressorL",
    "LipCornerDepressorR",
    "LipCornerPullerL",
    "LipCornerPullerR",
    "LipFunnelerLB",
    "LipFunnelerLT",
    "LipFunnelerRB",
    "LipFunnelerRT",
    "LipPressorL",
    "LipPressorR",
    "LipPuckerL",
    "LipPuckerR",
    "LipStretcherL",
    "LipStretcherR",
    "LipSuckLB",
    "LipSuckLT",
    "LipSuckRB",
    "LipSuckRT",
    "LipTightenerL",
    "LipTightenerR",
    "LipsToward",
    "LowerLipDepressorL",
    "LowerLipDepressorR",
    "MouthLeft",
    "MouthRight",
    "NoseWrinklerL",
    "NoseWrinklerR",
    "OuterBrowRaiserL",
    "OuterBrowRaiserR",
    "UpperLidRaiserL",
    "UpperLidRaiserR",
    "UpperLipRaiserL",
    "UpperLipRaiserR",
    "TongueTipInterdental",
    "TongueTipAlveolar",
    "TongueFrontDorsalPalate",
    "TongueMidDorsalPalate",
    "TongueBackDorsalVelar",
    "TongueOut",
    "TongueRetreat",
]

input_size = len(AU_NAMES)
assert input_size == 70


# ----------------------------------------------------------------------
# Architecture: copied unchanged from Part 7.2 (cell 70)
# ----------------------------------------------------------------------

import tensorflow as tf
from tensorflow.keras import layers, models, callbacks, backend as K

# ---- TCN expert blocks ----
def _tcn_block(x, filters, kernel_size, dilation_rate, dropout=0.2, name=None):
    res = x
    y = layers.Conv1D(filters, kernel_size, padding="causal",
                      dilation_rate=dilation_rate, name=None if name is None else name+"_c1")(x)
    y = layers.BatchNormalization(name=None if name is None else name+"_bn1")(y)
    y = layers.ReLU(name=None if name is None else name+"_relu1")(y)
    y = layers.Dropout(dropout, name=None if name is None else name+"_do1")(y)
    y = layers.Conv1D(filters, kernel_size, padding="causal",
                      dilation_rate=dilation_rate, name=None if name is None else name+"_c2")(y)
    y = layers.BatchNormalization(name=None if name is None else name+"_bn2")(y)
    if res.shape[-1] != filters:
        res = layers.Conv1D(filters, 1, padding="same", name=None if name is None else name+"_proj")(res)
        res = layers.BatchNormalization(name=None if name is None else name+"_proj_bn")(res)
    y = layers.Add(name=None if name is None else name+"_add")([y, res])
    y = layers.ReLU(name=None if name is None else name+"_relu2")(y)
    return y

def build_expert_tcn(x, name_prefix="exp"):
    y = _tcn_block(x, 128, 3, 1, 0.2, name=name_prefix+"_tcn1")
    y = _tcn_block(y, 128, 3, 2, 0.2, name=name_prefix+"_tcn2")
    y = _tcn_block(y, 128, 3, 4, 0.2, name=name_prefix+"_tcn3")
    y = layers.GlobalAveragePooling1D(name=name_prefix+"_gap")(y)
    return y  # (B, F)

# ---- Top-2 gate as a Keras Layer ----
class Top2Gate(layers.Layer):
    def __init__(self, num_experts=3, gate_hidden=128, tau_init=2.0,
                 lb_coef=0.02, ent_coef=0.01, name="gate", **kwargs):
        super().__init__(name=name, **kwargs)
        self.num_experts = num_experts
        self.gate_hidden = gate_hidden
        self.tau = tf.Variable(tau_init, dtype=tf.float32, trainable=False, name=f"{name}_tau")
        self.lb_coef = lb_coef
        self.ent_coef = ent_coef
        self.d1 = layers.Dense(gate_hidden, activation="relu", name=f"{name}_dense1")
        self.d2 = layers.Dense(num_experts, activation=None, name=f"{name}_logits")

    def call(self, gate_in, training=None):
        logits = self.d2(self.d1(gate_in))           # (B, E)
        w_soft = tf.nn.softmax(logits / self.tau, -1)  # (B, E)

        top2 = tf.math.top_k(w_soft, k=2)
        idx = top2.indices            # (B, 2)
        vals = tf.nn.softmax(top2.values, -1)  # re-norm top-2

        B = tf.shape(w_soft)[0]; E = tf.shape(w_soft)[1]
        row_ids = tf.repeat(tf.range(B), repeats=2)
        col_ids = tf.reshape(idx, (-1,))
        scatter_idx = tf.stack([row_ids, col_ids], axis=1)   # (B*2, 2)
        scatter_vals = tf.reshape(vals, (-1,))
        w_sparse = tf.scatter_nd(scatter_idx, scatter_vals, (B, self.num_experts))  # (B,E)

        # Load-balance loss (uniform usage)
        mean_usage = tf.reduce_mean(w_sparse, axis=0)
        uniform = tf.fill([self.num_experts], 1.0 / tf.cast(self.num_experts, tf.float32))
        lb_loss = tf.reduce_sum(tf.square(mean_usage - uniform))
        self.add_loss(self.lb_coef * lb_loss)

        # Entropy reg on soft weights (encourage exploration early)
        ent = -tf.reduce_mean(tf.reduce_sum(w_soft * tf.math.log(tf.clip_by_value(w_soft, 1e-8, 1.0)), axis=-1))
        self.add_loss(self.ent_coef * (-ent))  # minimize negative entropy

        return w_sparse, w_soft  # sparse mixture, full mixture (for diagnostics)

# ---- Temperature scheduler ----
class GateTemperatureScheduler(callbacks.Callback):
    def __init__(self, gate_layer, start_tau=2.0, end_tau=0.7, anneal_epochs=20):
        super().__init__()
        self.gate = gate_layer
        self.start_tau = float(start_tau)
        self.end_tau = float(end_tau)
        self.anneal_epochs = max(1, int(anneal_epochs))
    def on_epoch_begin(self, epoch, logs=None):
        p = min(1.0, epoch / self.anneal_epochs)
        tau = self.start_tau + p * (self.end_tau - self.start_tau)
        self.gate.tau.assign(tau)
        K.set_value(self.gate.tau, tau)
        if epoch % 5 == 0:
            print(f"[GateTemperatureScheduler] epoch={epoch} tau={tau:.3f}")

# ---- Single-input MoE builder (no domain hint) ----
def build_moe_single_input(
    input_shape, num_classes,
    tau_init=2.0, tau_final=0.7,
    gate_hidden=128, classifier_hidden=256,
    lb_coef=0.02, ent_coef=0.01,
    name="MoE_SingleInput"
):
    seq_in = layers.Input(shape=input_shape, name="seq")

    # Experts
    exp_nat = build_expert_tcn(seq_in, "exp_nat")
    exp_act = build_expert_tcn(seq_in, "exp_act")
    exp_shared = build_expert_tcn(seq_in, "exp_shared")

    # Stack experts into (B, E=3, F) using Lambda + Concatenate (avoid tf.stack on KerasTensors)
    e1 = layers.Lambda(lambda z: tf.expand_dims(z, axis=1), name="exp_nat_expand")(exp_nat)       # (B,1,F)
    e2 = layers.Lambda(lambda z: tf.expand_dims(z, axis=1), name="exp_act_expand")(exp_act)       # (B,1,F)
    e3 = layers.Lambda(lambda z: tf.expand_dims(z, axis=1), name="exp_share_expand")(exp_shared)  # (B,1,F)
    E_stack = layers.Concatenate(axis=1, name="experts_stack")([e1, e2, e3])  # (B,3,F)

    # Gate input: pooled sequence only (single input mode)
    gate_in = layers.GlobalAveragePooling1D(name="gate_pool")(seq_in)

    gate_layer = Top2Gate(num_experts=3, gate_hidden=gate_hidden,
                          tau_init=tau_init, lb_coef=lb_coef, ent_coef=ent_coef, name="gate")
    w_sparse, w_soft = gate_layer(gate_in)   # (B,3), (B,3)

    # Weighted sum: (B,3,1) * (B,3,F) -> sum over experts -> (B,F)
    w_expanded = layers.Lambda(lambda z: tf.expand_dims(z, -1), name="w_expand")(w_sparse)  # (B,3,1)
    fused = layers.Lambda(lambda zs: tf.reduce_sum(zs[0] * zs[1], axis=1), name="fuse")([w_expanded, E_stack])  # (B,F)

    x = layers.Dense(classifier_hidden, activation="relu", name="clf_dense")(fused)
    x = layers.Dropout(0.3, name="clf_do")(x)
    y_out = layers.Dense(num_classes, activation="softmax", name="y")(x)

    model = models.Model(seq_in, y_out, name=name)
    model.compile(optimizer=tf.keras.optimizers.SGD(1e-3),
                  loss="sparse_categorical_crossentropy",
                  metrics=["accuracy"])

    # Provide a temperature scheduler that targets this gate instance
    anneal_cb = GateTemperatureScheduler(gate_layer, start_tau=tau_init, end_tau=tau_final, anneal_epochs=20)
    return model, anneal_cb


# ------------------------------------------------------------
# Rebuild the architecture and load the saved best weights
# ------------------------------------------------------------

export_model, _ = build_moe_single_input(
    input_shape=(SEQ_LENGTH, input_size),
    num_classes=num_classes,
    tau_init=2.0,
    tau_final=0.7,
    gate_hidden=128,
    classifier_hidden=256,
    lb_coef=0.02,
    ent_coef=0.01,
)
export_model.load_weights(str(EXPORT_WEIGHTS))
print("Loaded weights:", EXPORT_WEIGHTS)

# The gate temperature tau is a plain tf.Variable, so it is NOT stored in the
# .keras weights file: after loading it is back at 2.0. The deployed
# sparseVR_teacher_model_acted_only.onnx was exported with tau = 0.7 (the end
# of the annealing schedule; its graph multiplies the gate logits by 1/0.7).
GATE_TAU = 0.7
export_model.get_layer("gate").tau.assign(GATE_TAU)
print("Gate temperature set to:", float(export_model.get_layer("gate").tau.numpy()))


# ------------------------------------------------------------
# Serving function (dynamic batch) and conversion
# ------------------------------------------------------------

input_signature = [
    tf.TensorSpec(shape=(None, SEQ_LENGTH, input_size), dtype=tf.float32, name="seq")
]


@tf.function(input_signature=input_signature)
def export_serving_function(seq):
    return {"probabilities": export_model(seq, training=False)}


trace = export_serving_function.get_concrete_function()(
    tf.zeros((1, SEQ_LENGTH, input_size), dtype=tf.float32)
)
assert trace["probabilities"].shape == (1, num_classes)

tf2onnx.convert.from_function(
    export_serving_function,
    input_signature=input_signature,
    opset=OPSET_VERSION,
    output_path=str(ONNX_EXPORT_PATH),
)
onnx.checker.check_model(onnx.load(str(ONNX_EXPORT_PATH)))
print("ONNX exported and checked:", ONNX_EXPORT_PATH)

session = ort.InferenceSession(str(ONNX_EXPORT_PATH), providers=["CPUExecutionProvider"])
input_meta = session.get_inputs()[0]
output_meta = session.get_outputs()[0]
print("Input: ", input_meta.name, input_meta.shape, input_meta.type)
print("Output:", output_meta.name, output_meta.shape, output_meta.type)
assert input_meta.name == "seq" and output_meta.name == "probabilities"
assert list(input_meta.shape[1:]) == [SEQ_LENGTH, input_size]


# ------------------------------------------------------------
# Keras <-> ONNX parity on synthetic AU windows, one window at a time
# (sparse values in [0, 1], from quiet to strong faces)
# ------------------------------------------------------------

rng = np.random.default_rng(0)
X_parity = rng.uniform(0.0, 1.0, size=(300, SEQ_LENGTH, input_size)).astype(np.float32)
X_parity *= (rng.uniform(size=X_parity.shape) < 0.3)           # most signals near 0
X_parity *= rng.uniform(0.05, 1.0, size=(300, 1, 1)).astype(np.float32)

keras_probabilities = export_model.predict(X_parity, batch_size=64, verbose=0)
onnx_probabilities = np.concatenate([
    session.run(None, {"seq": np.ascontiguousarray(X_parity[i:i + 1])})[0]
    for i in range(len(X_parity))
])

max_absolute_difference = float(np.max(np.abs(keras_probabilities - onnx_probabilities)))
prediction_agreement = float(np.mean(
    keras_probabilities.argmax(axis=1) == onnx_probabilities.argmax(axis=1)
))
print(f"\nParity on {len(X_parity)} synthetic windows:")
print(" prediction agreement:", prediction_agreement)
print(" max |Keras - ONNX| probability:", max_absolute_difference)
assert np.allclose(onnx_probabilities.sum(axis=1), 1.0, atol=1e-4)
assert prediction_agreement == 1.0, "Keras and ONNX produced different class decisions."
assert max_absolute_difference <= 1e-4, "Keras and ONNX probability difference exceeded 1e-4."


# ------------------------------------------------------------
# SHA-256 and metadata
# ------------------------------------------------------------

onnx_sha256 = hashlib.sha256(ONNX_EXPORT_PATH.read_bytes()).hexdigest()

metadata = {
    "schema_version": "acted-only-teacher-onnx-v1",
    "teacher_tag": EXPORT_TAG,
    "onnx_file": ONNX_EXPORT_PATH.name,
    "onnx_sha256": onnx_sha256,
    "opset": OPSET_VERSION,
    "gate_tau": GATE_TAU,
    "training_sources": ["old_vr_acted", "new_15_participants_acted"],
    "emojihero_in_training": False,
    "volume_augmentation_copies_per_window": int(EXPORT_TAG[-1]),
    "report": "Docs/VOLUME_AUGMENTATION_REPORT.md",
    "input": {"name": input_meta.name, "shape": input_meta.shape, "au_order": list(AU_NAMES)},
    "output": {"name": output_meta.name, "shape": output_meta.shape,
               "class_order": CLASS_ORDER, "activation": "softmax"},
    "keras_onnx_parity": {
        "dataset": "synthetic AU windows",
        "sample_count": int(len(X_parity)),
        "prediction_agreement": prediction_agreement,
        "max_absolute_probability_difference": max_absolute_difference,
    },
    "test_sets_evaluated_during_export": False,
}
with open(ONNX_EXPORT_METADATA_PATH, "w", encoding="utf-8") as metadata_file:
    json.dump(metadata, metadata_file, indent=2)

print("\nONNX SHA-256:", onnx_sha256)
print("Saved to Drive:", ONNX_EXPORT_PATH)
print("Metadata:      ", ONNX_EXPORT_METADATA_PATH)
````

````python
import os
import random
import numpy as np
import tensorflow as tf

from sklearn.metrics import f1_score
from tensorflow.keras import callbacks
from tensorflow.keras.callbacks import CSVLogger


# ============================================================
# Configuration
# ============================================================

RANDOM_SEED = 42

NEW_TEACHER_TAG = (
    "teacher-70AU-acted-only-oldacted-emojihero-new15acted-"
    "source-balanced-domain-balanced-validation"
)

NEW_MODEL_PATH = os.path.join(
    models_path,
    NEW_TEACHER_TAG + ".keras",
)

NEW_CHECKPOINT_PATH = os.path.join(
    models_path,
    NEW_TEACHER_TAG + "_best.keras",
)

NEW_TRAINING_LOG_PATH = os.path.join(
    models_path,
    NEW_TEACHER_TAG + "_training_log.csv",
)

os.makedirs(models_path, exist_ok=True)


# ============================================================
# Reproducibility
# ============================================================

os.environ["PYTHONHASHSEED"] = str(RANDOM_SEED)

random.seed(RANDOM_SEED)
np.random.seed(RANDOM_SEED)
tf.keras.utils.set_random_seed(RANDOM_SEED)

try:
    tf.config.experimental.enable_op_determinism()
    print("TensorFlow deterministic operations enabled.")
except Exception as exc:
    print("Could not enable full TensorFlow determinism:", exc)


# ============================================================
# Tensor conversion helpers
# ============================================================

def ensure_sequence_tensor(values):
    """Convert sequence collections to float32 tensors."""

    if isinstance(values, np.ndarray):
        tensor = values
    else:
        tensor = stack_sequences(values)

    tensor = np.asarray(tensor, dtype=np.float32)

    if tensor.ndim != 3:
        raise ValueError(
            f"Expected a rank-3 sequence tensor, received {tensor.shape}"
        )

    return tensor


def ensure_label_vector(values):
    """Convert labels to a flat int64 vector."""

    if isinstance(values, np.ndarray):
        labels = values
    else:
        labels = to_numpy_labels(values)

    return np.asarray(labels, dtype=np.int64).reshape(-1)


# ============================================================
# Build three independent validation domains
# ============================================================

# Previous VR acted validation data
X_val_old_acted_np = ensure_sequence_tensor(X_val_act)
Y_val_old_acted_np = ensure_label_vector(Y_val_act)

# New 15-participant acted validation data
X_val_new_acted_np = ensure_sequence_tensor(X_val_new)
Y_val_new_acted_np = ensure_label_vector(Y_val_new)

# EmojiHero validation data
X_val_emojihero_np = ensure_sequence_tensor(X_val_em_hero)
Y_val_emojihero_np = ensure_label_vector(Y_val_em_hero)


validation_domains = [
    (
        "old_acted",
        X_val_old_acted_np,
        Y_val_old_acted_np,
    ),
    (
        "new_acted",
        X_val_new_acted_np,
        Y_val_new_acted_np,
    ),
    (
        "emojihero",
        X_val_emojihero_np,
        Y_val_emojihero_np,
    ),
]


for domain_name, X_domain, y_domain in validation_domains:
    assert X_domain.shape[1:] == (
        SEQ_LENGTH,
        len(AU_NAMES),
    ), (
        domain_name,
        X_domain.shape,
    )

    assert len(X_domain) == len(y_domain), (
        domain_name,
        len(X_domain),
        len(y_domain),
    )

    assert set(np.unique(y_domain)).issubset(
        set(range(num_classes))
    )

    print(
        f"{domain_name:12s} | "
        f"X={X_domain.shape} | "
        f"y={y_domain.shape} | "
        f"classes={np.unique(y_domain).tolist()}"
    )


# ============================================================
# Create the pooled validation set
#
# This set is used only for standard Keras validation loss and
# accuracy logging. It does not select the best checkpoint.
# ============================================================

X_val_all_acted_np = np.concatenate(
    [
        X_val_old_acted_np,
        X_val_new_acted_np,
        X_val_emojihero_np,
    ],
    axis=0,
)

Y_val_all_acted_np = np.concatenate(
    [
        Y_val_old_acted_np,
        Y_val_new_acted_np,
        Y_val_emojihero_np,
    ],
    axis=0,
)

print("\nPooled validation:")
print("X:", X_val_all_acted_np.shape)
print("y:", Y_val_all_acted_np.shape)


# ============================================================
# Domain-balanced macro-F1 callback
#
# Each validation domain contributes exactly one third of the
# checkpoint-selection score, regardless of its window count.
# ============================================================

class DomainBalancedMacroF1(callbacks.Callback):
    def __init__(
        self,
        validation_domains,
        num_classes,
        prediction_batch_size=256,
    ):
        super().__init__()

        self.validation_domains = validation_domains
        self.num_classes = num_classes
        self.prediction_batch_size = prediction_batch_size

    def on_epoch_end(self, epoch, logs=None):
        logs = logs if logs is not None else {}

        domain_scores = {}

        for domain_name, X_domain, y_domain in self.validation_domains:
            probabilities = self.model.predict(
                X_domain,
                batch_size=self.prediction_batch_size,
                verbose=0,
            )

            predictions = np.argmax(probabilities, axis=1)

            macro_f1 = f1_score(
                y_domain,
                predictions,
                labels=np.arange(self.num_classes),
                average="macro",
                zero_division=0,
            )

            macro_f1 = float(macro_f1)

            domain_scores[domain_name] = macro_f1
            logs[f"val_{domain_name}_macro_f1"] = macro_f1

        domain_balanced_macro_f1 = float(
            np.mean(list(domain_scores.values()))
        )

        logs[
            "val_domain_balanced_macro_f1"
        ] = domain_balanced_macro_f1

        print(
            "\nValidation macro-F1 | "
            f"old acted: {domain_scores['old_acted']:.4f} | "
            f"new acted: {domain_scores['new_acted']:.4f} | "
            f"EmojiHero: {domain_scores['emojihero']:.4f} | "
            f"domain balanced: {domain_balanced_macro_f1:.4f}"
        )


# ============================================================
# Locate the previously prepared source-and-class-balanced
# training sample weights
# ============================================================

sample_weight_candidates = [
    "train_sample_weights",
    "sample_weight_train",
    "sample_weights_train",
    "training_sample_weights",
    "sample_weights",
]

selected_sample_weights = None
selected_sample_weight_name = None

for candidate_name in sample_weight_candidates:
    if candidate_name in globals():
        candidate_value = np.asarray(
            globals()[candidate_name],
            dtype=np.float32,
        ).reshape(-1)

        if len(candidate_value) == len(X_train_np):
            selected_sample_weights = candidate_value
            selected_sample_weight_name = candidate_name
            break

if selected_sample_weights is None:
    raise NameError(
        "The source-and-class-balanced training sample weights "
        "were not found. Run the cell that prints "
        "'Sample weights are ready' before this training cell."
    )


assert len(X_train_np) == len(Y_train_np)
assert len(selected_sample_weights) == len(Y_train_np)
assert np.all(np.isfinite(selected_sample_weights))
assert np.all(selected_sample_weights > 0)

print("\nTraining data:")
print("X:", X_train_np.shape)
print("y:", Y_train_np.shape)
print("Sample-weight variable:", selected_sample_weight_name)
print("Sample-weight minimum:", selected_sample_weights.min())
print("Sample-weight maximum:", selected_sample_weights.max())
print("Sample-weight mean:", selected_sample_weights.mean())


# ============================================================
# Build a new model from scratch
#
# The previously trained model is not reused or fine-tuned.
# ============================================================

tf.keras.backend.clear_session()

random.seed(RANDOM_SEED)
np.random.seed(RANDOM_SEED)
tf.keras.utils.set_random_seed(RANDOM_SEED)

model, anneal_cb = build_moe_single_input(
    input_shape=(SEQ_LENGTH, input_size),
    num_classes=num_classes,
    tau_init=2.0,
    tau_final=0.7,
    gate_hidden=128,
    classifier_hidden=256,
    lb_coef=0.02,
    ent_coef=0.01,
)


# ============================================================
# Configure model-selection callbacks
#
# DomainBalancedMacroF1 must run before EarlyStopping,
# ModelCheckpoint, and CSVLogger so the custom metric is
# available to those callbacks at the end of every epoch.
# ============================================================

domain_f1_cb = DomainBalancedMacroF1(
    validation_domains=validation_domains,
    num_classes=num_classes,
    prediction_batch_size=256,
)

early_stopping_cb = callbacks.EarlyStopping(
    monitor="val_domain_balanced_macro_f1",
    mode="max",
    patience=8,
    restore_best_weights=True,
    verbose=1,
)

checkpoint_cb = callbacks.ModelCheckpoint(
    filepath=NEW_CHECKPOINT_PATH,
    monitor="val_domain_balanced_macro_f1",
    mode="max",
    save_best_only=True,
    verbose=1,
)

csv_logger_cb = CSVLogger(
    NEW_TRAINING_LOG_PATH,
    append=False,
)


# ============================================================
# Train
#
# Do not also pass class_weight here. The selected sample
# weights already contain both emotion-class balancing and
# source balancing.
# ============================================================

history = model.fit(
    X_train_np,
    Y_train_np,
    sample_weight=selected_sample_weights,
    validation_data=(
        X_val_all_acted_np,
        Y_val_all_acted_np,
    ),
    epochs=epochs,
    batch_size=batch_size,
    callbacks=[
        domain_f1_cb,
        early_stopping_cb,
        checkpoint_cb,
        csv_logger_cb,
        anneal_cb,
    ],
    verbose=1,
)


# ============================================================
# Save the model containing the restored best weights
# ============================================================

trained_model = model
trained_model.save(NEW_MODEL_PATH)

best_epoch_index = int(
    np.argmax(
        history.history["val_domain_balanced_macro_f1"]
    )
)

best_epoch_number = best_epoch_index + 1
best_validation_score = history.history[
    "val_domain_balanced_macro_f1"
][best_epoch_index]

print("\nTraining complete.")
print("Best epoch:", best_epoch_number)
print(
    "Best domain-balanced validation macro-F1:",
    f"{best_validation_score:.6f}",
)
print("Final model:", NEW_MODEL_PATH)
print("Best checkpoint:", NEW_CHECKPOINT_PATH)
print("Training log:", NEW_TRAINING_LOG_PATH)
print("\nNo test dataset was evaluated in this cell.")
````

````python
# ============================================================
# EXPORT RESTORED-BEST KERAS TEACHER TO ONNX
#
# - Fixed batch size 1 for Unity/Sentis runtime
# - Input shape:  (1, 5, 70)
# - Output shape: (1, 7)
# - Validation parity only; test sets are not evaluated
# ============================================================

import hashlib
import json
from pathlib import Path

import numpy as np
import onnx
import onnxruntime as ort
import tensorflow as tf
import tf2onnx


# ------------------------------------------------------------
# Preconditions
# ------------------------------------------------------------

assert "trained_model" in globals(), (
    "Run the training cell first."
)

assert "TEACHER_TAG" in globals()
assert "TEACHER_RUN_DIR" in globals()

assert trained_model.input_shape[1:] == (
    SEQ_LENGTH,
    len(AU_NAMES),
)

assert trained_model.output_shape[-1] == len(Emotion)


# ------------------------------------------------------------
# Output paths
# ------------------------------------------------------------

ONNX_OUTPUT_PATH = (
    Path(TEACHER_RUN_DIR)
    / f"{TEACHER_TAG}.onnx"
)

ONNX_METADATA_PATH = (
    Path(TEACHER_RUN_DIR)
    / f"{TEACHER_TAG}.onnx.metadata.json"
)

ONNX_OUTPUT_PATH.parent.mkdir(
    parents=True,
    exist_ok=True,
)


# ------------------------------------------------------------
# Fixed batch-1 inference function
#
# Unity runs one 5-frame window at a time, so exporting batch=1
# avoids relying on dynamic-shape behavior in Sentis.
# ------------------------------------------------------------

input_signature = [
    tf.TensorSpec(
        shape=(
            1,
            SEQ_LENGTH,
            len(AU_NAMES),
        ),
        dtype=tf.float32,
        name="seq",
    )
]


@tf.function(
    input_signature=input_signature
)
def onnx_serving_function(seq):
    probabilities = trained_model(
        seq,
        training=False,
    )

    # Stable logical output name.
    return {
        "probabilities": probabilities
    }


# Trace before conversion so shape errors occur here.
concrete_function = (
    onnx_serving_function
    .get_concrete_function()
)

trace_output = concrete_function(
    tf.zeros(
        (
            1,
            SEQ_LENGTH,
            len(AU_NAMES),
        ),
        dtype=tf.float32,
    )
)

assert trace_output["probabilities"].shape == (
    1,
    len(Emotion),
)


# ------------------------------------------------------------
# Convert to ONNX
# ------------------------------------------------------------

OPSET_VERSION = 17

onnx_model, _ = tf2onnx.convert.from_function(
    onnx_serving_function,
    input_signature=input_signature,
    opset=OPSET_VERSION,
    output_path=str(ONNX_OUTPUT_PATH),
)

assert ONNX_OUTPUT_PATH.exists(), (
    f"ONNX export failed: {ONNX_OUTPUT_PATH}"
)

print("ONNX exported:")
print(ONNX_OUTPUT_PATH)


# ------------------------------------------------------------
# Structural ONNX validation
# ------------------------------------------------------------

loaded_onnx = onnx.load(
    str(ONNX_OUTPUT_PATH)
)

onnx.checker.check_model(
    loaded_onnx
)

print("onnx.checker: OK")


# ------------------------------------------------------------
# Open with ONNX Runtime
# ------------------------------------------------------------

session = ort.InferenceSession(
    str(ONNX_OUTPUT_PATH),
    providers=["CPUExecutionProvider"],
)

inputs = session.get_inputs()
outputs = session.get_outputs()

assert len(inputs) == 1, (
    f"Expected one ONNX input, found {len(inputs)}"
)

assert len(outputs) == 1, (
    f"Expected one ONNX output, found {len(outputs)}"
)

input_meta = inputs[0]
output_meta = outputs[0]

input_name = input_meta.name
output_name = output_meta.name

print("\nONNX input:")
print(" name:", input_name)
print(" shape:", input_meta.shape)
print(" type:", input_meta.type)

print("\nONNX output:")
print(" name:", output_name)
print(" shape:", output_meta.shape)
print(" type:", output_meta.type)


# ------------------------------------------------------------
# Keras ↔ ONNX parity on VR validation data
#
# Do not use X_test_vr_np or X_test_emojihero_np here.
# ------------------------------------------------------------

PARITY_SAMPLE_COUNT = min(
    128,
    len(X_val_np),
)

X_parity = np.asarray(
    X_val_np[:PARITY_SAMPLE_COUNT],
    dtype=np.float32,
)

keras_probabilities = trained_model.predict(
    X_parity,
    batch_size=64,
    verbose=0,
)

onnx_probability_rows = []

for sample_index in range(
    PARITY_SAMPLE_COUNT
):
    sample = np.ascontiguousarray(
        X_parity[
            sample_index:sample_index + 1
        ],
        dtype=np.float32,
    )

    result = session.run(
        [output_name],
        {
            input_name: sample
        },
    )[0]

    result = np.asarray(
        result,
        dtype=np.float32,
    ).reshape(
        1,
        len(Emotion),
    )

    onnx_probability_rows.append(
        result
    )

onnx_probabilities = np.concatenate(
    onnx_probability_rows,
    axis=0,
)


# ------------------------------------------------------------
# Output integrity
# ------------------------------------------------------------

assert keras_probabilities.shape == (
    PARITY_SAMPLE_COUNT,
    len(Emotion),
)

assert onnx_probabilities.shape == (
    PARITY_SAMPLE_COUNT,
    len(Emotion),
)

assert np.isfinite(
    keras_probabilities
).all()

assert np.isfinite(
    onnx_probabilities
).all()

assert np.all(
    onnx_probabilities >= -1e-6
)

assert np.all(
    onnx_probabilities <= 1.0 + 1e-6
)

assert np.allclose(
    onnx_probabilities.sum(
        axis=1
    ),
    1.0,
    atol=1e-4,
    rtol=1e-4,
)


# ------------------------------------------------------------
# Numerical and decision parity
# ------------------------------------------------------------

max_absolute_difference = float(
    np.max(
        np.abs(
            keras_probabilities
            - onnx_probabilities
        )
    )
)

mean_absolute_difference = float(
    np.mean(
        np.abs(
            keras_probabilities
            - onnx_probabilities
        )
    )
)

keras_predictions = np.argmax(
    keras_probabilities,
    axis=1,
)

onnx_predictions = np.argmax(
    onnx_probabilities,
    axis=1,
)

prediction_agreement = float(
    np.mean(
        keras_predictions
        == onnx_predictions
    )
)

mismatch_indices = np.flatnonzero(
    keras_predictions
    != onnx_predictions
)

print("\nKeras ↔ ONNX parity:")
print(
    " Samples:",
    PARITY_SAMPLE_COUNT,
)
print(
    " Prediction agreement:",
    prediction_agreement,
)
print(
    " Mismatches:",
    len(mismatch_indices),
)
print(
    " Maximum absolute probability difference:",
    max_absolute_difference,
)
print(
    " Mean absolute probability difference:",
    mean_absolute_difference,
)

if len(mismatch_indices):
    print(
        "First mismatch indices:",
        mismatch_indices[:10].tolist(),
    )

assert prediction_agreement == 1.0, (
    "Keras and ONNX produced different class decisions."
)

assert max_absolute_difference <= 1e-4, (
    "Keras and ONNX probability difference exceeded 1e-4."
)


# ------------------------------------------------------------
# SHA-256
# ------------------------------------------------------------

onnx_bytes = ONNX_OUTPUT_PATH.read_bytes()

onnx_sha256 = hashlib.sha256(
    onnx_bytes
).hexdigest()

print("\nONNX SHA-256:")
print(onnx_sha256)


# ------------------------------------------------------------
# Save provenance and parity metadata
# ------------------------------------------------------------

class_order = [
    emotion.value
    for emotion in Emotion
]

metadata = {
    "schema_version": "acted-only-teacher-onnx-v1",
    "teacher_tag": TEACHER_TAG,
    "onnx_file": ONNX_OUTPUT_PATH.name,
    "onnx_sha256": onnx_sha256,
    "opset": OPSET_VERSION,

    "training_sources": [
        "old_vr_acted",
        "emojihero_training",
        "new_15_participants_acted",
    ],

    "source_weighting": {
        "method": (
            "equal total loss mass per source, "
            "combined with balanced class weights"
        ),
        "old_acted_window_count": int(
            len(X_train_act)
        ),
        "emojihero_window_count": int(
            len(X_train_em_hero)
        ),
        "new_acted_window_count": int(
            len(X_train_new)
        ),
    },

    "model_selection": {
        "primary_metric": (
            "VR acted validation macro-F1"
        ),
        "best_epoch": int(
            best_epoch_index + 1
        ),
        "best_vr_validation_macro_f1": float(
            vr_history[best_epoch_index]
        ),
        "emojihero_validation_macro_f1_at_selected_epoch": float(
            emojihero_history[
                best_epoch_index
            ]
        ),
    },

    "input": {
        "name": input_name,
        "shape": input_meta.shape,
        "dtype": input_meta.type,
        "sequence_length": int(
            SEQ_LENGTH
        ),
        "feature_count": int(
            len(AU_NAMES)
        ),
        "au_order": list(AU_NAMES),
    },

    "output": {
        "name": output_name,
        "shape": output_meta.shape,
        "dtype": output_meta.type,
        "class_order": class_order,
        "activation": "softmax",
    },

    "keras_onnx_parity": {
        "dataset": "VR acted validation",
        "sample_count": int(
            PARITY_SAMPLE_COUNT
        ),
        "prediction_agreement": (
            prediction_agreement
        ),
        "mismatch_count": int(
            len(mismatch_indices)
        ),
        "max_absolute_probability_difference": (
            max_absolute_difference
        ),
        "mean_absolute_probability_difference": (
            mean_absolute_difference
        ),
    },

    "test_sets_evaluated_during_export": False,
}

with open(
    ONNX_METADATA_PATH,
    "w",
    encoding="utf-8",
) as metadata_file:
    json.dump(
        metadata,
        metadata_file,
        indent=2,
        ensure_ascii=False,
    )

print("\nMetadata saved:")
print(ONNX_METADATA_PATH)

print("\nONNX export and validation complete.")
print("Test sets were not evaluated.")
````

````python
import os
import sys
import json
import hashlib
import tempfile
import subprocess

import numpy as np
import tensorflow as tf


# ============================================================
# Configuration
# ============================================================

ONNX_OPSET = 15
PARITY_SAMPLE_COUNT = 128

ONNX_OUTPUT_DIR = os.path.join(
    predictions_path,
    NEW_TEACHER_TAG,
)

ONNX_OUTPUT_PATH = os.path.join(
    ONNX_OUTPUT_DIR,
    NEW_TEACHER_TAG + ".onnx",
)

ONNX_METADATA_PATH = ONNX_OUTPUT_PATH + ".metadata.json"

os.makedirs(ONNX_OUTPUT_DIR, exist_ok=True)

temporary_root = tempfile.mkdtemp(
    prefix="teacher_onnx_export_",
    dir="/content",
)

saved_model_path = os.path.join(
    temporary_root,
    "saved_model",
)

parity_input_path = os.path.join(
    temporary_root,
    "parity_input.npy",
)

keras_output_path = os.path.join(
    temporary_root,
    "keras_output.npy",
)

onnx_output_path = os.path.join(
    temporary_root,
    "onnx_output.npy",
)


# ============================================================
# Verify that the trained model is still available in memory
# ============================================================

assert "trained_model" in globals(), (
    "trained_model is not available in memory."
)

assert trained_model.input_shape[1:] == (
    SEQ_LENGTH,
    len(AU_NAMES),
)

assert trained_model.output_shape[-1] == num_classes

print("Keras model input shape:", trained_model.input_shape)
print("Keras model output shape:", trained_model.output_shape)


# ============================================================
# Export a TensorFlow SavedModel with explicit interface names
#
# This step runs before reinstalling ONNX packages. It uses the
# trained Keras model that is already available in memory.
# ============================================================

class TeacherServingModule(tf.Module):
    def __init__(self, model):
        super().__init__()
        self.model = model

    @tf.function(
        input_signature=[
            tf.TensorSpec(
                shape=[1, SEQ_LENGTH, len(AU_NAMES)],
                dtype=tf.float32,
                name="seq",
            )
        ]
    )
    def serve(self, seq):
        probabilities = self.model(
            seq,
            training=False,
        )

        return {
            "probabilities": probabilities,
        }


serving_module = TeacherServingModule(trained_model)

tf.saved_model.save(
    serving_module,
    saved_model_path,
    signatures={
        "serving_default": serving_module.serve,
    },
)

print("Temporary SavedModel:", saved_model_path)


# ============================================================
# Prepare Keras predictions for parity validation
# ============================================================

if "X_val_all_acted_np" in globals():
    parity_source = X_val_all_acted_np
elif "X_val_old_acted_np" in globals():
    parity_source = X_val_old_acted_np
elif "X_train_np" in globals():
    parity_source = X_train_np
else:
    raise NameError(
        "No input tensor is available for parity validation."
    )

parity_count = min(
    PARITY_SAMPLE_COUNT,
    len(parity_source),
)

rng = np.random.default_rng(42)

parity_indices = rng.choice(
    len(parity_source),
    size=parity_count,
    replace=False,
)

parity_input = np.asarray(
    parity_source[parity_indices],
    dtype=np.float32,
)

keras_probabilities = trained_model.predict(
    parity_input,
    batch_size=64,
    verbose=0,
)

keras_probabilities = np.asarray(
    keras_probabilities,
    dtype=np.float32,
)

np.save(parity_input_path, parity_input)
np.save(keras_output_path, keras_probabilities)

print("Parity samples prepared:", parity_count)


# ============================================================
# Repair the ONNX installation on disk
#
# The current notebook process will not import these packages.
# A new subprocess will see the clean installation.
# ============================================================

install_command = [
    sys.executable,
    "-m",
    "pip",
    "install",
    "--quiet",
    "--no-cache-dir",
    "--force-reinstall",
    "--no-deps",
    "protobuf==5.29.6",
    "onnx==1.18.0",
    "tf2onnx==1.17.0",
]

subprocess.run(
    install_command,
    check=True,
)

print("ONNX export packages repaired on disk.")


# ============================================================
# Convert SavedModel to ONNX in a fresh Python subprocess
#
# This avoids importing the broken/stale ONNX module in the
# current notebook process.
# ============================================================

conversion_command = [
    sys.executable,
    "-m",
    "tf2onnx.convert",
    "--saved-model",
    saved_model_path,
    "--signature_def",
    "serving_default",
    "--opset",
    str(ONNX_OPSET),
    "--output",
    ONNX_OUTPUT_PATH,
]

conversion_result = subprocess.run(
    conversion_command,
    check=True,
    text=True,
    capture_output=True,
)

print(conversion_result.stdout)

if conversion_result.stderr:
    print(conversion_result.stderr)

assert os.path.exists(ONNX_OUTPUT_PATH), (
    f"ONNX file was not created: {ONNX_OUTPUT_PATH}"
)


# ============================================================
# Validate ONNX and run inference in another clean subprocess
# ============================================================

validation_script = r"""
import sys
import json
import numpy as np
import onnx
import onnxruntime as ort

onnx_path = sys.argv[1]
input_path = sys.argv[2]
output_path = sys.argv[3]

model = onnx.load(onnx_path)
onnx.checker.check_model(model)

session = ort.InferenceSession(
    onnx_path,
    providers=["CPUExecutionProvider"],
)

inputs = session.get_inputs()
outputs = session.get_outputs()

if len(inputs) != 1:
    raise RuntimeError(
        f"Expected one ONNX input, received {len(inputs)}"
    )

if len(outputs) != 1:
    raise RuntimeError(
        f"Expected one ONNX output, received {len(outputs)}"
    )

input_metadata = {
    "name": inputs[0].name,
    "shape": inputs[0].shape,
    "type": inputs[0].type,
}

output_metadata = {
    "name": outputs[0].name,
    "shape": outputs[0].shape,
    "type": outputs[0].type,
}

X = np.load(input_path).astype(np.float32)

predictions = []

for sample in X:
    sample = sample[np.newaxis, ...]

    result = session.run(
        None,
        {
            inputs[0].name: sample,
        },
    )[0]

    predictions.append(result[0])

predictions = np.asarray(
    predictions,
    dtype=np.float32,
)

np.save(output_path, predictions)

print(
    json.dumps(
        {
            "onnx_version": onnx.__version__,
            "onnxruntime_version": ort.__version__,
            "input": input_metadata,
            "output": output_metadata,
            "checker": "OK",
        }
    )
)
"""

validation_result = subprocess.run(
    [
        sys.executable,
        "-c",
        validation_script,
        ONNX_OUTPUT_PATH,
        parity_input_path,
        onnx_output_path,
    ],
    check=True,
    text=True,
    capture_output=True,
)

validation_lines = [
    line
    for line in validation_result.stdout.splitlines()
    if line.strip()
]

validation_metadata = json.loads(
    validation_lines[-1]
)

onnx_probabilities = np.load(
    onnx_output_path,
).astype(np.float32)


# ============================================================
# Keras-to-ONNX parity
# ============================================================

assert keras_probabilities.shape == onnx_probabilities.shape

keras_predictions = np.argmax(
    keras_probabilities,
    axis=1,
)

onnx_predictions = np.argmax(
    onnx_probabilities,
    axis=1,
)

prediction_agreement = float(
    np.mean(keras_predictions == onnx_predictions)
)

mismatch_count = int(
    np.sum(keras_predictions != onnx_predictions)
)

absolute_difference = np.abs(
    keras_probabilities - onnx_probabilities
)

maximum_absolute_difference = float(
    absolute_difference.max()
)

mean_absolute_difference = float(
    absolute_difference.mean()
)


# ============================================================
# Calculate the ONNX SHA-256 hash
# ============================================================

sha256 = hashlib.sha256()

with open(ONNX_OUTPUT_PATH, "rb") as model_file:
    for block in iter(
        lambda: model_file.read(1024 * 1024),
        b"",
    ):
        sha256.update(block)

onnx_sha256 = sha256.hexdigest()


# ============================================================
# Save export metadata
# ============================================================

export_metadata = {
    "teacher_tag": NEW_TEACHER_TAG,
    "onnx_path": ONNX_OUTPUT_PATH,
    "onnx_sha256": onnx_sha256,
    "opset": ONNX_OPSET,
    "checker": validation_metadata["checker"],
    "onnx_version": validation_metadata["onnx_version"],
    "onnxruntime_version": validation_metadata[
        "onnxruntime_version"
    ],
    "input": validation_metadata["input"],
    "output": validation_metadata["output"],
    "parity": {
        "samples": int(parity_count),
        "prediction_agreement": prediction_agreement,
        "mismatches": mismatch_count,
        "maximum_absolute_probability_difference": (
            maximum_absolute_difference
        ),
        "mean_absolute_probability_difference": (
            mean_absolute_difference
        ),
    },
    "test_sets_evaluated": False,
}

with open(
    ONNX_METADATA_PATH,
    "w",
    encoding="utf-8",
) as metadata_file:
    json.dump(
        export_metadata,
        metadata_file,
        indent=2,
    )


# ============================================================
# Final report
# ============================================================

print("\nONNX exported:", ONNX_OUTPUT_PATH)
print("onnx.checker:", validation_metadata["checker"])

print("\nONNX input:")
print(" name:", validation_metadata["input"]["name"])
print(" shape:", validation_metadata["input"]["shape"])
print(" type:", validation_metadata["input"]["type"])

print("\nONNX output:")
print(" name:", validation_metadata["output"]["name"])
print(" shape:", validation_metadata["output"]["shape"])
print(" type:", validation_metadata["output"]["type"])

print("\nKeras ↔ ONNX parity:")
print(" Samples:", parity_count)
print(" Prediction agreement:", prediction_agreement)
print(" Mismatches:", mismatch_count)
print(
    " Maximum absolute probability difference:",
    maximum_absolute_difference,
)
print(
    " Mean absolute probability difference:",
    mean_absolute_difference,
)

print("\nONNX SHA-256:", onnx_sha256)
print("Metadata saved:", ONNX_METADATA_PATH)

assert prediction_agreement == 1.0
assert mismatch_count == 0

print("\nONNX export and validation complete.")
print("The notebook runtime was not restarted.")
print("Test sets were not evaluated.")
````

````python
import os
import sys
import json
import hashlib
import tempfile
import subprocess

from pathlib import Path

import numpy as np


# ============================================================
# Evaluation configuration
# ============================================================

EVALUATION_MODEL_TAG = (
    "teacher-70AU-acted-only-oldacted-emojihero-new15acted-"
    "source-balanced-domain-balanced-validation"
)

EVALUATION_MODEL_DIR = (
    Path(predictions_path)
    / EVALUATION_MODEL_TAG
)

EVALUATION_ONNX_PATH = (
    EVALUATION_MODEL_DIR
    / f"{EVALUATION_MODEL_TAG}.onnx"
)

EVALUATION_OUTPUT_DIR = (
    EVALUATION_MODEL_DIR
    / "final_test_evaluation"
)

SUMMARY_CSV_PATH = (
    EVALUATION_OUTPUT_DIR
    / "final_test_summary.csv"
)

SUMMARY_JSON_PATH = (
    EVALUATION_OUTPUT_DIR
    / "final_test_summary.json"
)

PER_EMOTION_CSV_PATH = (
    EVALUATION_OUTPUT_DIR
    / "final_test_per_emotion.csv"
)

PER_EMOTION_JSON_PATH = (
    EVALUATION_OUTPUT_DIR
    / "final_test_per_emotion.json"
)

PREDICTIONS_DIR = (
    EVALUATION_OUTPUT_DIR
    / "predictions"
)


# ============================================================
# No-overwrite protection
# ============================================================

assert EVALUATION_ONNX_PATH.exists(), (
    f"ONNX model not found: {EVALUATION_ONNX_PATH}"
)

existing_result_files = [
    path
    for path in [
        SUMMARY_CSV_PATH,
        SUMMARY_JSON_PATH,
        PER_EMOTION_CSV_PATH,
        PER_EMOTION_JSON_PATH,
    ]
    if path.exists()
]

if existing_result_files:
    raise FileExistsError(
        "Evaluation results already exist. Refusing to overwrite:\n"
        + "\n".join(str(path) for path in existing_result_files)
    )

print("Evaluation model:", EVALUATION_MODEL_TAG)
print("ONNX:", EVALUATION_ONNX_PATH)
print("Output:", EVALUATION_OUTPUT_DIR)


# ============================================================
# Conversion helpers
# ============================================================

def evaluation_sequence_tensor(values):
    """Convert a sequence collection to a float32 tensor."""

    if isinstance(values, np.ndarray):
        tensor = values
    else:
        tensor = stack_sequences(values)

    tensor = np.asarray(
        tensor,
        dtype=np.float32,
    )

    if tensor.ndim != 3:
        raise ValueError(
            f"Expected a rank-3 tensor, received {tensor.shape}"
        )

    if tensor.shape[1:] != (
        SEQ_LENGTH,
        len(AU_NAMES),
    ):
        raise ValueError(
            f"Unexpected sequence shape: {tensor.shape}"
        )

    return tensor


def evaluation_label_vector(values):
    """Convert labels to a flat int64 vector."""

    if isinstance(values, np.ndarray):
        labels = values
    else:
        labels = to_numpy_labels(values)

    return np.asarray(
        labels,
        dtype=np.int64,
    ).reshape(-1)


# ============================================================
# Prepare the three independent test domains
# ============================================================

X_test_old_acted_np = evaluation_sequence_tensor(
    X_test_act
)

Y_test_old_acted_np = evaluation_label_vector(
    Y_test_act
)

X_test_new_acted_np = evaluation_sequence_tensor(
    X_test_new
)

Y_test_new_acted_np = evaluation_label_vector(
    Y_test_new
)

X_test_emojihero_np = evaluation_sequence_tensor(
    X_test_em_hero
)

Y_test_emojihero_np = evaluation_label_vector(
    Y_test_em_hero
)


source_test_sets = {
    "old_vr_acted": {
        "X": X_test_old_acted_np,
        "y": Y_test_old_acted_np,
    },
    "new_vr_acted": {
        "X": X_test_new_acted_np,
        "y": Y_test_new_acted_np,
    },
    "emojihero": {
        "X": X_test_emojihero_np,
        "y": Y_test_emojihero_np,
    },
}


# ============================================================
# Corpus assertions
# ============================================================

expected_window_counts = {
    "old_vr_acted": 303,
    "new_vr_acted": 304,
    "emojihero": 2268,
}

for dataset_name, dataset in source_test_sets.items():
    X_dataset = dataset["X"]
    y_dataset = dataset["y"]

    assert len(X_dataset) == len(y_dataset)
    assert len(X_dataset) == expected_window_counts[dataset_name]
    assert set(np.unique(y_dataset)).issubset(
        set(range(num_classes))
    )

    print(
        f"{dataset_name:16s} | "
        f"X={X_dataset.shape} | "
        f"y={y_dataset.shape} | "
        f"classes={np.unique(y_dataset).tolist()}"
    )


# ============================================================
# Compute model hash
# ============================================================

sha256 = hashlib.sha256()

with open(EVALUATION_ONNX_PATH, "rb") as model_file:
    for block in iter(
        lambda: model_file.read(1024 * 1024),
        b"",
    ):
        sha256.update(block)

evaluation_model_sha256 = sha256.hexdigest()

expected_model_sha256 = (
    "7f77838a52bbc4e0ebb1ffe122e81f5b196d1fbbc2a376b5b9463bea55ef4de5"
)

assert evaluation_model_sha256 == expected_model_sha256, (
    "The ONNX model hash does not match the newly exported model.\n"
    f"Expected: {expected_model_sha256}\n"
    f"Received: {evaluation_model_sha256}"
)

print("\nONNX SHA-256:", evaluation_model_sha256)


# ============================================================
# ONNX Runtime inference script
#
# Inference runs in a fresh subprocess because the current
# notebook process contains a stale ONNX import state.
# ============================================================

inference_script = r"""
import sys
import json
import numpy as np
import onnxruntime as ort

model_path = sys.argv[1]
input_path = sys.argv[2]
output_path = sys.argv[3]

X = np.load(input_path).astype(np.float32)

session = ort.InferenceSession(
    model_path,
    providers=["CPUExecutionProvider"],
)

model_input = session.get_inputs()[0]
model_output = session.get_outputs()[0]

probabilities = []

for sample in X:
    sample_batch = sample[np.newaxis, ...]

    batch_probabilities = session.run(
        [model_output.name],
        {
            model_input.name: sample_batch,
        },
    )[0]

    probabilities.append(batch_probabilities[0])

probabilities = np.asarray(
    probabilities,
    dtype=np.float32,
)

np.save(output_path, probabilities)

print(
    json.dumps(
        {
            "input_name": model_input.name,
            "input_shape": model_input.shape,
            "input_type": model_input.type,
            "output_name": model_output.name,
            "output_shape": model_output.shape,
            "output_type": model_output.type,
            "n_samples": len(X),
        }
    )
)
"""


# ============================================================
# Run inference once for each independent source domain
# ============================================================

temporary_evaluation_dir = Path(
    tempfile.mkdtemp(
        prefix="teacher_test_evaluation_",
        dir="/content",
    )
)

for dataset_name, dataset in source_test_sets.items():
    temporary_input_path = (
        temporary_evaluation_dir
        / f"{dataset_name}_input.npy"
    )

    temporary_output_path = (
        temporary_evaluation_dir
        / f"{dataset_name}_probabilities.npy"
    )

    np.save(
        temporary_input_path,
        dataset["X"],
    )

    inference_result = subprocess.run(
        [
            sys.executable,
            "-c",
            inference_script,
            str(EVALUATION_ONNX_PATH),
            str(temporary_input_path),
            str(temporary_output_path),
        ],
        check=True,
        text=True,
        capture_output=True,
    )

    output_lines = [
        line
        for line in inference_result.stdout.splitlines()
        if line.strip()
    ]

    runtime_metadata = json.loads(
        output_lines[-1]
    )

    probabilities = np.load(
        temporary_output_path
    ).astype(np.float32)

    predictions = np.argmax(
        probabilities,
        axis=1,
    ).astype(np.int64)

    assert probabilities.shape == (
        len(dataset["y"]),
        num_classes,
    )

    assert predictions.shape == dataset["y"].shape
    assert np.all(np.isfinite(probabilities))

    probability_sums = probabilities.sum(axis=1)

    assert np.allclose(
        probability_sums,
        1.0,
        atol=1e-5,
    )

    dataset["probabilities"] = probabilities
    dataset["predictions"] = predictions
    dataset["runtime_metadata"] = runtime_metadata

    print(
        f"Inference complete: {dataset_name:16s} | "
        f"n={len(predictions)}"
    )


# ============================================================
# Build combined evaluation sets without rerunning inference
# ============================================================

evaluation_sets = dict(source_test_sets)

evaluation_sets["vr_acted_combined"] = {
    "X": np.concatenate(
        [
            source_test_sets["old_vr_acted"]["X"],
            source_test_sets["new_vr_acted"]["X"],
        ],
        axis=0,
    ),
    "y": np.concatenate(
        [
            source_test_sets["old_vr_acted"]["y"],
            source_test_sets["new_vr_acted"]["y"],
        ],
        axis=0,
    ),
    "probabilities": np.concatenate(
        [
            source_test_sets["old_vr_acted"]["probabilities"],
            source_test_sets["new_vr_acted"]["probabilities"],
        ],
        axis=0,
    ),
    "predictions": np.concatenate(
        [
            source_test_sets["old_vr_acted"]["predictions"],
            source_test_sets["new_vr_acted"]["predictions"],
        ],
        axis=0,
    ),
}

evaluation_sets["all_acted_pooled"] = {
    "X": np.concatenate(
        [
            source_test_sets["old_vr_acted"]["X"],
            source_test_sets["new_vr_acted"]["X"],
            source_test_sets["emojihero"]["X"],
        ],
        axis=0,
    ),
    "y": np.concatenate(
        [
            source_test_sets["old_vr_acted"]["y"],
            source_test_sets["new_vr_acted"]["y"],
            source_test_sets["emojihero"]["y"],
        ],
        axis=0,
    ),
    "probabilities": np.concatenate(
        [
            source_test_sets["old_vr_acted"]["probabilities"],
            source_test_sets["new_vr_acted"]["probabilities"],
            source_test_sets["emojihero"]["probabilities"],
        ],
        axis=0,
    ),
    "predictions": np.concatenate(
        [
            source_test_sets["old_vr_acted"]["predictions"],
            source_test_sets["new_vr_acted"]["predictions"],
            source_test_sets["emojihero"]["predictions"],
        ],
        axis=0,
    ),
}


assert len(evaluation_sets["vr_acted_combined"]["y"]) == 607
assert len(evaluation_sets["all_acted_pooled"]["y"]) == 2875

print("\nCombined test sets:")
print(
    "VR acted combined:",
    len(evaluation_sets["vr_acted_combined"]["y"]),
)

print(
    "All acted pooled:",
    len(evaluation_sets["all_acted_pooled"]["y"]),
)

print("\nInference preparation complete.")
````

````python
import json

import numpy as np
import pandas as pd

from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    confusion_matrix,
    f1_score,
    log_loss,
    precision_recall_fscore_support,
    precision_score,
    recall_score,
)


# ============================================================
# Metric configuration
# ============================================================

class_indices = np.arange(num_classes)

emotion_names = [
    emotion.value
    for emotion in Emotion
]

assert len(emotion_names) == num_classes


# ============================================================
# Metric helpers
# ============================================================

def calculate_overall_metrics(
    dataset_name,
    y_true,
    y_pred,
    probabilities,
):
    """Calculate overall metrics for one evaluation set."""

    return {
        "dataset": dataset_name,
        "n_windows": int(len(y_true)),
        "accuracy": float(
            accuracy_score(y_true, y_pred)
        ),
        "balanced_accuracy": float(
            balanced_accuracy_score(y_true, y_pred)
        ),
        "precision_macro": float(
            precision_score(
                y_true,
                y_pred,
                labels=class_indices,
                average="macro",
                zero_division=0,
            )
        ),
        "recall_macro": float(
            recall_score(
                y_true,
                y_pred,
                labels=class_indices,
                average="macro",
                zero_division=0,
            )
        ),
        "f1_macro": float(
            f1_score(
                y_true,
                y_pred,
                labels=class_indices,
                average="macro",
                zero_division=0,
            )
        ),
        "f1_weighted": float(
            f1_score(
                y_true,
                y_pred,
                labels=class_indices,
                average="weighted",
                zero_division=0,
            )
        ),
        "log_loss": float(
            log_loss(
                y_true,
                probabilities,
                labels=class_indices,
            )
        ),
    }


def calculate_per_emotion_metrics(
    dataset_name,
    y_true,
    y_pred,
):
    """Calculate precision, recall, and F1 for each emotion."""

    precision, recall, f1, support = (
        precision_recall_fscore_support(
            y_true,
            y_pred,
            labels=class_indices,
            average=None,
            zero_division=0,
        )
    )

    rows = []

    for class_index, emotion_name in enumerate(emotion_names):
        rows.append(
            {
                "dataset": dataset_name,
                "class_index": int(class_index),
                "emotion": emotion_name,
                "precision": float(precision[class_index]),
                "recall": float(recall[class_index]),
                "f1": float(f1[class_index]),
                "support": int(support[class_index]),
            }
        )

    return rows


# ============================================================
# Calculate metrics
# ============================================================

dataset_order = [
    "old_vr_acted",
    "new_vr_acted",
    "emojihero",
    "vr_acted_combined",
    "all_acted_pooled",
]

overall_rows = []
per_emotion_rows = []
confusion_matrices = {}

for dataset_name in dataset_order:
    dataset = evaluation_sets[dataset_name]

    overall_rows.append(
        calculate_overall_metrics(
            dataset_name=dataset_name,
            y_true=dataset["y"],
            y_pred=dataset["predictions"],
            probabilities=dataset["probabilities"],
        )
    )

    per_emotion_rows.extend(
        calculate_per_emotion_metrics(
            dataset_name=dataset_name,
            y_true=dataset["y"],
            y_pred=dataset["predictions"],
        )
    )

    confusion_matrices[dataset_name] = confusion_matrix(
        dataset["y"],
        dataset["predictions"],
        labels=class_indices,
    )


overall_df = pd.DataFrame(overall_rows)
per_emotion_df = pd.DataFrame(per_emotion_rows)


# ============================================================
# Calculate equal-domain overall metrics
#
# Old VR acted, new VR acted, and EmojiHero each contribute
# exactly one third, regardless of their number of windows.
# ============================================================

independent_domain_names = [
    "old_vr_acted",
    "new_vr_acted",
    "emojihero",
]

independent_domain_df = (
    overall_df[
        overall_df["dataset"].isin(
            independent_domain_names
        )
    ]
    .set_index("dataset")
    .loc[independent_domain_names]
)

domain_balanced_row = {
    "dataset": "all_acted_domain_balanced",
    "n_windows": None,
    "accuracy": float(
        independent_domain_df["accuracy"].mean()
    ),
    "balanced_accuracy": float(
        independent_domain_df[
            "balanced_accuracy"
        ].mean()
    ),
    "precision_macro": float(
        independent_domain_df[
            "precision_macro"
        ].mean()
    ),
    "recall_macro": float(
        independent_domain_df[
            "recall_macro"
        ].mean()
    ),
    "f1_macro": float(
        independent_domain_df["f1_macro"].mean()
    ),
    "f1_weighted": float(
        independent_domain_df[
            "f1_weighted"
        ].mean()
    ),
    "log_loss": float(
        independent_domain_df["log_loss"].mean()
    ),
}

overall_df = pd.concat(
    [
        pd.DataFrame([domain_balanced_row]),
        overall_df,
    ],
    ignore_index=True,
)


# ============================================================
# Calculate equal-domain per-emotion metrics
# ============================================================

independent_per_emotion_df = per_emotion_df[
    per_emotion_df["dataset"].isin(
        independent_domain_names
    )
]

domain_balanced_per_emotion_df = (
    independent_per_emotion_df
    .groupby(
        [
            "class_index",
            "emotion",
        ],
        as_index=False,
    )[
        [
            "precision",
            "recall",
            "f1",
        ]
    ]
    .mean()
)

domain_balanced_per_emotion_df.insert(
    0,
    "dataset",
    "all_acted_domain_balanced",
)

domain_balanced_per_emotion_df["support"] = None

per_emotion_df = pd.concat(
    [
        domain_balanced_per_emotion_df,
        per_emotion_df,
    ],
    ignore_index=True,
)


# ============================================================
# Create the output directory only after all calculations pass
# ============================================================

EVALUATION_OUTPUT_DIR.mkdir(
    parents=True,
    exist_ok=False,
)

PREDICTIONS_DIR.mkdir(
    parents=False,
    exist_ok=False,
)


# ============================================================
# Save raw predictions for reproducibility
# ============================================================

for dataset_name in dataset_order:
    dataset = evaluation_sets[dataset_name]

    np.savez_compressed(
        PREDICTIONS_DIR / f"{dataset_name}.npz",
        y_true=dataset["y"],
        y_pred=dataset["predictions"],
        probabilities=dataset["probabilities"],
    )


# ============================================================
# Save confusion matrices
# ============================================================

np.savez_compressed(
    EVALUATION_OUTPUT_DIR / "confusion_matrices.npz",
    **confusion_matrices,
)


# ============================================================
# Save CSV files
# ============================================================

overall_df.to_csv(
    SUMMARY_CSV_PATH,
    index=False,
)

per_emotion_df.to_csv(
    PER_EMOTION_CSV_PATH,
    index=False,
)


# ============================================================
# Save JSON files
# ============================================================

evaluation_metadata = {
    "model_tag": EVALUATION_MODEL_TAG,
    "onnx_path": str(EVALUATION_ONNX_PATH),
    "onnx_sha256": evaluation_model_sha256,
    "class_order": emotion_names,
    "window_shape": [
        int(SEQ_LENGTH),
        int(len(AU_NAMES)),
    ],
    "source_test_window_counts": {
        name: int(len(source_test_sets[name]["y"]))
        for name in independent_domain_names
    },
    "combined_test_window_counts": {
        "vr_acted_combined": int(
            len(evaluation_sets["vr_acted_combined"]["y"])
        ),
        "all_acted_pooled": int(
            len(evaluation_sets["all_acted_pooled"]["y"])
        ),
    },
    "domain_balanced_definition": (
        "Arithmetic mean of old_vr_acted, new_vr_acted, "
        "and emojihero domain-level metrics."
    ),
    "test_status": (
        "Development evaluation. These test sets were observed "
        "during the preceding model-development process."
    ),
}

summary_payload = {
    "metadata": evaluation_metadata,
    "results": overall_df.to_dict(
        orient="records"
    ),
}

per_emotion_payload = {
    "metadata": evaluation_metadata,
    "results": per_emotion_df.to_dict(
        orient="records"
    ),
}

with open(
    SUMMARY_JSON_PATH,
    "w",
    encoding="utf-8",
) as output_file:
    json.dump(
        summary_payload,
        output_file,
        indent=2,
        allow_nan=False,
    )

with open(
    PER_EMOTION_JSON_PATH,
    "w",
    encoding="utf-8",
) as output_file:
    json.dump(
        per_emotion_payload,
        output_file,
        indent=2,
        allow_nan=False,
    )


print("Evaluation results saved:")
print("Summary CSV:", SUMMARY_CSV_PATH)
print("Summary JSON:", SUMMARY_JSON_PATH)
print("Per-emotion CSV:", PER_EMOTION_CSV_PATH)
print("Per-emotion JSON:", PER_EMOTION_JSON_PATH)
print("Predictions:", PREDICTIONS_DIR)
````

````python
import os
import sys
import json
import hashlib
import tempfile
import subprocess

from pathlib import Path

import numpy as np
import tensorflow as tf


# ============================================================
# Configuration
# ============================================================

ONNX_OPSET = 15
PARITY_SAMPLE_COUNT = 128
DYNAMIC_BATCH_TEST_SIZE = 20

DYNAMIC_EXPORT_TAG = (
    "teacher-70AU-acted-only-oldacted-emojihero-new15acted-"
    "source-balanced-domain-balanced-validation-dynamic-batch"
)

DYNAMIC_ONNX_OUTPUT_DIR = (
    Path(predictions_path)
    / DYNAMIC_EXPORT_TAG
)

DYNAMIC_ONNX_OUTPUT_PATH = (
    DYNAMIC_ONNX_OUTPUT_DIR
    / f"{DYNAMIC_EXPORT_TAG}.onnx"
)

DYNAMIC_ONNX_METADATA_PATH = Path(
    str(DYNAMIC_ONNX_OUTPUT_PATH) + ".metadata.json"
)

DYNAMIC_ONNX_OUTPUT_DIR.mkdir(
    parents=True,
    exist_ok=True,
)

if DYNAMIC_ONNX_OUTPUT_PATH.exists():
    raise FileExistsError(
        "The dynamic-batch ONNX file already exists. "
        "Refusing to overwrite it:\n"
        f"{DYNAMIC_ONNX_OUTPUT_PATH}"
    )


# ============================================================
# Verify the in-memory Keras model
# ============================================================

assert "trained_model" in globals(), (
    "trained_model is not available in memory."
)

assert trained_model.input_shape == (
    None,
    SEQ_LENGTH,
    len(AU_NAMES),
), trained_model.input_shape

assert trained_model.output_shape == (
    None,
    num_classes,
), trained_model.output_shape

print("Keras model input shape:", trained_model.input_shape)
print("Keras model output shape:", trained_model.output_shape)


# ============================================================
# Create temporary paths
# ============================================================

temporary_root = Path(
    tempfile.mkdtemp(
        prefix="teacher_dynamic_onnx_export_",
        dir="/content",
    )
)

saved_model_path = (
    temporary_root
    / "saved_model"
)

parity_input_path = (
    temporary_root
    / "parity_input.npy"
)

keras_output_path = (
    temporary_root
    / "keras_output.npy"
)

onnx_output_path = (
    temporary_root
    / "onnx_output.npy"
)


# ============================================================
# Export a SavedModel with a dynamic batch dimension
#
# None means that the model accepts any batch size:
# [batch, 5, 70] -> [batch, 7]
# ============================================================

class DynamicTeacherServingModule(tf.Module):
    def __init__(self, model):
        super().__init__()
        self.model = model

    @tf.function(
        input_signature=[
            tf.TensorSpec(
                shape=[
                    None,
                    SEQ_LENGTH,
                    len(AU_NAMES),
                ],
                dtype=tf.float32,
                name="seq",
            )
        ]
    )
    def serve(self, seq):
        probabilities = self.model(
            seq,
            training=False,
        )

        return {
            "probabilities": probabilities,
        }


serving_module = DynamicTeacherServingModule(
    trained_model
)

tf.saved_model.save(
    serving_module,
    str(saved_model_path),
    signatures={
        "serving_default": serving_module.serve,
    },
)

print(
    "Temporary dynamic SavedModel:",
    saved_model_path,
)


# ============================================================
# Select parity-test inputs
# ============================================================

if "X_val_all_acted_np" in globals():
    parity_source = X_val_all_acted_np
elif "X_val_old_acted_np" in globals():
    parity_source = X_val_old_acted_np
elif "X_train_np" in globals():
    parity_source = X_train_np
else:
    raise NameError(
        "No suitable tensor is available for parity testing."
    )

parity_source = np.asarray(
    parity_source,
    dtype=np.float32,
)

parity_count = min(
    PARITY_SAMPLE_COUNT,
    len(parity_source),
)

assert parity_count >= DYNAMIC_BATCH_TEST_SIZE

rng = np.random.default_rng(42)

parity_indices = rng.choice(
    len(parity_source),
    size=parity_count,
    replace=False,
)

parity_input = np.asarray(
    parity_source[parity_indices],
    dtype=np.float32,
)

assert parity_input.shape == (
    parity_count,
    SEQ_LENGTH,
    len(AU_NAMES),
)


# ============================================================
# Calculate Keras predictions
# ============================================================

keras_probabilities = trained_model.predict(
    parity_input,
    batch_size=64,
    verbose=0,
)

keras_probabilities = np.asarray(
    keras_probabilities,
    dtype=np.float32,
)

assert keras_probabilities.shape == (
    parity_count,
    num_classes,
)

np.save(
    parity_input_path,
    parity_input,
)

np.save(
    keras_output_path,
    keras_probabilities,
)

print("Parity samples prepared:", parity_count)


# ============================================================
# Repair the ONNX packages on disk
#
# The current notebook process does not import ONNX.
# Conversion and validation run in fresh subprocesses.
# ============================================================

install_command = [
    sys.executable,
    "-m",
    "pip",
    "install",
    "--quiet",
    "--no-cache-dir",
    "--force-reinstall",
    "--no-deps",
    "protobuf==5.29.6",
    "onnx==1.18.0",
    "tf2onnx==1.17.0",
]

subprocess.run(
    install_command,
    check=True,
)

print("ONNX export packages repaired on disk.")


# ============================================================
# Convert the dynamic SavedModel to ONNX
# ============================================================

conversion_command = [
    sys.executable,
    "-m",
    "tf2onnx.convert",
    "--saved-model",
    str(saved_model_path),
    "--signature_def",
    "serving_default",
    "--opset",
    str(ONNX_OPSET),
    "--output",
    str(DYNAMIC_ONNX_OUTPUT_PATH),
]

conversion_result = subprocess.run(
    conversion_command,
    check=True,
    text=True,
    capture_output=True,
)

print(conversion_result.stdout)

if conversion_result.stderr:
    print(conversion_result.stderr)

assert DYNAMIC_ONNX_OUTPUT_PATH.exists(), (
    "The dynamic-batch ONNX file was not created."
)


# ============================================================
# Validate ONNX and run one batched inference operation
#
# All parity samples are passed in one call. This explicitly
# proves that the ONNX model accepts batch sizes greater than 1.
# ============================================================

validation_script = r"""
import sys
import json
import numpy as np
import onnx
import onnxruntime as ort

model_path = sys.argv[1]
input_path = sys.argv[2]
output_path = sys.argv[3]
dynamic_test_size = int(sys.argv[4])

model = onnx.load(model_path)
onnx.checker.check_model(model)

session = ort.InferenceSession(
    model_path,
    providers=["CPUExecutionProvider"],
)

inputs = session.get_inputs()
outputs = session.get_outputs()

if len(inputs) != 1:
    raise RuntimeError(
        f"Expected exactly one input, received {len(inputs)}"
    )

if len(outputs) != 1:
    raise RuntimeError(
        f"Expected exactly one output, received {len(outputs)}"
    )

model_input = inputs[0]
model_output = outputs[0]

if model_input.shape[0] == 1:
    raise RuntimeError(
        "The exported ONNX model still has a fixed batch size of 1."
    )

if model_output.shape[0] == 1:
    raise RuntimeError(
        "The exported ONNX output still has a fixed batch size of 1."
    )

X = np.load(input_path).astype(np.float32)

# Run all parity samples in one ONNX Runtime call.
probabilities = session.run(
    [model_output.name],
    {
        model_input.name: X,
    },
)[0]

probabilities = np.asarray(
    probabilities,
    dtype=np.float32,
)

expected_output_shape = (
    len(X),
    7,
)

if probabilities.shape != expected_output_shape:
    raise RuntimeError(
        f"Unexpected ONNX output shape: {probabilities.shape}; "
        f"expected {expected_output_shape}"
    )

# Run an explicit 20-sample dynamic-batch test.
dynamic_test_input = X[:dynamic_test_size]

dynamic_test_output = session.run(
    [model_output.name],
    {
        model_input.name: dynamic_test_input,
    },
)[0]

expected_dynamic_shape = (
    dynamic_test_size,
    7,
)

if dynamic_test_output.shape != expected_dynamic_shape:
    raise RuntimeError(
        f"Dynamic batch test failed: {dynamic_test_output.shape}; "
        f"expected {expected_dynamic_shape}"
    )

np.save(
    output_path,
    probabilities,
)

metadata = {
    "onnx_version": onnx.__version__,
    "onnxruntime_version": ort.__version__,
    "checker": "OK",
    "input": {
        "name": model_input.name,
        "shape": model_input.shape,
        "type": model_input.type,
    },
    "output": {
        "name": model_output.name,
        "shape": model_output.shape,
        "type": model_output.type,
    },
    "parity_batch_size": int(len(X)),
    "dynamic_batch_test_size": dynamic_test_size,
    "dynamic_batch_test_output_shape": list(
        dynamic_test_output.shape
    ),
    "dynamic_batch_test": "OK",
}

print(json.dumps(metadata))
"""

validation_result = subprocess.run(
    [
        sys.executable,
        "-c",
        validation_script,
        str(DYNAMIC_ONNX_OUTPUT_PATH),
        str(parity_input_path),
        str(onnx_output_path),
        str(DYNAMIC_BATCH_TEST_SIZE),
    ],
    check=True,
    text=True,
    capture_output=True,
)

if validation_result.stderr:
    print(validation_result.stderr)

validation_lines = [
    line
    for line in validation_result.stdout.splitlines()
    if line.strip()
]

validation_metadata = json.loads(
    validation_lines[-1]
)

onnx_probabilities = np.load(
    onnx_output_path
).astype(np.float32)


# ============================================================
# Keras-to-ONNX parity
# ============================================================

assert keras_probabilities.shape == onnx_probabilities.shape

keras_predictions = np.argmax(
    keras_probabilities,
    axis=1,
)

onnx_predictions = np.argmax(
    onnx_probabilities,
    axis=1,
)

prediction_agreement = float(
    np.mean(
        keras_predictions
        == onnx_predictions
    )
)

mismatch_count = int(
    np.sum(
        keras_predictions
        != onnx_predictions
    )
)

absolute_probability_difference = np.abs(
    keras_probabilities
    - onnx_probabilities
)

maximum_absolute_probability_difference = float(
    absolute_probability_difference.max()
)

mean_absolute_probability_difference = float(
    absolute_probability_difference.mean()
)


# ============================================================
# Calculate the new ONNX SHA-256 hash
# ============================================================

sha256 = hashlib.sha256()

with open(
    DYNAMIC_ONNX_OUTPUT_PATH,
    "rb",
) as model_file:
    for block in iter(
        lambda: model_file.read(1024 * 1024),
        b"",
    ):
        sha256.update(block)

dynamic_onnx_sha256 = sha256.hexdigest()


# ============================================================
# Save export metadata
# ============================================================

export_metadata = {
    "teacher_tag": DYNAMIC_EXPORT_TAG,
    "source_teacher_tag": NEW_TEACHER_TAG,
    "onnx_path": str(DYNAMIC_ONNX_OUTPUT_PATH),
    "onnx_sha256": dynamic_onnx_sha256,
    "opset": ONNX_OPSET,
    "checker": validation_metadata["checker"],
    "onnx_version": validation_metadata["onnx_version"],
    "onnxruntime_version": validation_metadata[
        "onnxruntime_version"
    ],
    "input": validation_metadata["input"],
    "output": validation_metadata["output"],
    "dynamic_batch": {
        "enabled": True,
        "test_batch_size": validation_metadata[
            "dynamic_batch_test_size"
        ],
        "test_output_shape": validation_metadata[
            "dynamic_batch_test_output_shape"
        ],
        "status": validation_metadata[
            "dynamic_batch_test"
        ],
    },
    "parity": {
        "samples": int(parity_count),
        "single_onnx_runtime_call": True,
        "prediction_agreement": prediction_agreement,
        "mismatches": mismatch_count,
        "maximum_absolute_probability_difference": (
            maximum_absolute_probability_difference
        ),
        "mean_absolute_probability_difference": (
            mean_absolute_probability_difference
        ),
    },
    "weights_retrained": False,
    "test_sets_evaluated": False,
}

with open(
    DYNAMIC_ONNX_METADATA_PATH,
    "w",
    encoding="utf-8",
) as metadata_file:
    json.dump(
        export_metadata,
        metadata_file,
        indent=2,
    )


# ============================================================
# Final validation
# ============================================================

assert validation_metadata["checker"] == "OK"
assert validation_metadata["dynamic_batch_test"] == "OK"
assert prediction_agreement == 1.0
assert mismatch_count == 0

assert validation_metadata["input"]["name"] == "seq"
assert validation_metadata["output"]["name"] == "probabilities"

assert validation_metadata["input"]["shape"][0] != 1
assert validation_metadata["output"]["shape"][0] != 1


# ============================================================
# Final report
# ============================================================

print("\nDynamic-batch ONNX exported:")
print(DYNAMIC_ONNX_OUTPUT_PATH)

print("\nonnx.checker:")
print(validation_metadata["checker"])

print("\nONNX input:")
print(
    " name:",
    validation_metadata["input"]["name"],
)
print(
    " shape:",
    validation_metadata["input"]["shape"],
)
print(
    " type:",
    validation_metadata["input"]["type"],
)

print("\nONNX output:")
print(
    " name:",
    validation_metadata["output"]["name"],
)
print(
    " shape:",
    validation_metadata["output"]["shape"],
)
print(
    " type:",
    validation_metadata["output"]["type"],
)

print("\nDynamic batch test:")
print(
    " batch size:",
    validation_metadata["dynamic_batch_test_size"],
)
print(
    " output shape:",
    validation_metadata[
        "dynamic_batch_test_output_shape"
    ],
)
print(
    " status:",
    validation_metadata["dynamic_batch_test"],
)

print("\nKeras ↔ ONNX parity:")
print(" Samples:", parity_count)
print(
    " Prediction agreement:",
    prediction_agreement,
)
print(" Mismatches:", mismatch_count)
print(
    " Maximum absolute probability difference:",
    maximum_absolute_probability_difference,
)
print(
    " Mean absolute probability difference:",
    mean_absolute_probability_difference,
)

print("\nNew ONNX SHA-256:")
print(dynamic_onnx_sha256)

print("\nMetadata saved:")
print(DYNAMIC_ONNX_METADATA_PATH)

print("\nDynamic-batch ONNX export complete.")
print("The notebook runtime was not restarted.")
print("The model was not retrained.")
print("Test datasets were not evaluated.")
````

````python
import sys
import json
import tempfile
import subprocess

from pathlib import Path

import numpy as np
import pandas as pd

from IPython.display import display


# ============================================================
# Select the dynamic-batch ONNX model
# ============================================================

if "DYNAMIC_ONNX_OUTPUT_PATH" in globals():
    FEAR_ANALYSIS_ONNX_PATH = Path(
        DYNAMIC_ONNX_OUTPUT_PATH
    )
else:
    DYNAMIC_EXPORT_TAG = (
        "teacher-70AU-acted-only-oldacted-emojihero-new15acted-"
        "source-balanced-domain-balanced-validation-dynamic-batch"
    )

    FEAR_ANALYSIS_ONNX_PATH = (
        Path(predictions_path)
        / DYNAMIC_EXPORT_TAG
        / f"{DYNAMIC_EXPORT_TAG}.onnx"
    )

assert FEAR_ANALYSIS_ONNX_PATH.exists(), (
    f"Dynamic ONNX model not found: "
    f"{FEAR_ANALYSIS_ONNX_PATH}"
)

print("ONNX model:", FEAR_ANALYSIS_ONNX_PATH)


# ============================================================
# Conversion helpers
# ============================================================

def fear_analysis_sequence_tensor(values):
    """Convert sequence collections to float32 tensors."""

    if isinstance(values, np.ndarray):
        tensor = values
    else:
        tensor = stack_sequences(values)

    tensor = np.asarray(
        tensor,
        dtype=np.float32,
    )

    assert tensor.ndim == 3
    assert tensor.shape[1:] == (
        SEQ_LENGTH,
        len(AU_NAMES),
    )

    return tensor


def fear_analysis_label_vector(values):
    """Convert labels to a flat int64 vector."""

    if isinstance(values, np.ndarray):
        labels = values
    else:
        labels = to_numpy_labels(values)

    return np.asarray(
        labels,
        dtype=np.int64,
    ).reshape(-1)


# ============================================================
# Determine the Fear class index from the model class order
# ============================================================

emotion_names = [
    emotion.value
    for emotion in Emotion
]

fear_class_index = emotion_names.index("Fear")

print("Class order:", emotion_names)
print("Fear class index:", fear_class_index)


# ============================================================
# Prepare source-specific datasets
# ============================================================

fear_analysis_datasets = {
    "old_acted_train_in_sample": {
        "X": fear_analysis_sequence_tensor(
            X_train_act
        ),
        "y": fear_analysis_label_vector(
            Y_train_act
        ),
        "split_status": "train / in-sample",
    },
    "old_acted_test_held_out": {
        "X": fear_analysis_sequence_tensor(
            X_test_act
        ),
        "y": fear_analysis_label_vector(
            Y_test_act
        ),
        "split_status": "test / held-out",
    },
    "new_acted_test_held_out": {
        "X": fear_analysis_sequence_tensor(
            X_test_new
        ),
        "y": fear_analysis_label_vector(
            Y_test_new
        ),
        "split_status": "test / held-out",
    },
    "emojihero_test_held_out": {
        "X": fear_analysis_sequence_tensor(
            X_test_em_hero
        ),
        "y": fear_analysis_label_vector(
            Y_test_em_hero
        ),
        "split_status": "test / held-out",
    },
}


for dataset_name, dataset in fear_analysis_datasets.items():
    assert len(dataset["X"]) == len(dataset["y"])

    print(
        f"{dataset_name:30s} | "
        f"n={len(dataset['y'])}"
    )


# ============================================================
# ONNX Runtime subprocess
#
# ONNX is not imported in the current notebook process.
# Each complete dataset is evaluated in one dynamic-batch call.
# ============================================================

inference_script = r"""
import sys
import numpy as np
import onnxruntime as ort

model_path = sys.argv[1]
input_path = sys.argv[2]
output_path = sys.argv[3]

X = np.load(input_path).astype(np.float32)

session = ort.InferenceSession(
    model_path,
    providers=["CPUExecutionProvider"],
)

model_input = session.get_inputs()[0]
model_output = session.get_outputs()[0]

probabilities = session.run(
    [model_output.name],
    {
        model_input.name: X,
    },
)[0]

probabilities = np.asarray(
    probabilities,
    dtype=np.float32,
)

np.save(
    output_path,
    probabilities,
)

print("Input:", X.shape)
print("Output:", probabilities.shape)
"""


temporary_directory = Path(
    tempfile.mkdtemp(
        prefix="fear_source_analysis_",
        dir="/content",
    )
)

fear_rows = []

for dataset_name, dataset in fear_analysis_datasets.items():
    input_path = (
        temporary_directory
        / f"{dataset_name}_input.npy"
    )

    output_path = (
        temporary_directory
        / f"{dataset_name}_probabilities.npy"
    )

    np.save(
        input_path,
        dataset["X"],
    )

    result = subprocess.run(
        [
            sys.executable,
            "-c",
            inference_script,
            str(FEAR_ANALYSIS_ONNX_PATH),
            str(input_path),
            str(output_path),
        ],
        check=True,
        text=True,
        capture_output=True,
    )

    probabilities = np.load(
        output_path
    ).astype(np.float32)

    predictions = np.argmax(
        probabilities,
        axis=1,
    ).astype(np.int64)

    y_true = dataset["y"]

    assert probabilities.shape == (
        len(y_true),
        len(emotion_names),
    )

    fear_mask = (
        y_true == fear_class_index
    )

    number_of_fear_windows = int(
        fear_mask.sum()
    )

    if number_of_fear_windows == 0:
        raise ValueError(
            f"No Fear windows found in {dataset_name}"
        )

    correctly_predicted_fear = int(
        np.sum(
            predictions[fear_mask]
            == fear_class_index
        )
    )

    fear_recall = (
        correctly_predicted_fear
        / number_of_fear_windows
    )

    predicted_as_fear_total = int(
        np.sum(
            predictions == fear_class_index
        )
    )

    fear_rows.append(
        {
            "dataset": dataset_name,
            "split_status": dataset["split_status"],
            "n_windows_total": int(len(y_true)),
            "n_fear_windows": number_of_fear_windows,
            "fear_correct": correctly_predicted_fear,
            "fear_recall": float(fear_recall),
            "n_predicted_as_fear": predicted_as_fear_total,
        }
    )


# ============================================================
# Display and save the result
# ============================================================

fear_source_results_df = pd.DataFrame(
    fear_rows
)

display(
    fear_source_results_df.round(4)
)

FEAR_RESULTS_PATH = (
    Path(predictions_path)
    / DYNAMIC_EXPORT_TAG
    / "fear_recall_by_source.csv"
)

fear_source_results_df.to_csv(
    FEAR_RESULTS_PATH,
    index=False,
)

print("Saved:", FEAR_RESULTS_PATH)
````

````python
import sys
import tempfile
import subprocess

from pathlib import Path

import numpy as np
import pandas as pd

from IPython.display import display


# ============================================================
# Preconditions
# ============================================================

required_variables = [
    "fear_analysis_datasets",
    "FEAR_ANALYSIS_ONNX_PATH",
    "emotion_names",
]

missing_variables = [
    name
    for name in required_variables
    if name not in globals()
]

if missing_variables:
    raise NameError(
        "Run the previous Fear source-analysis cell first. "
        f"Missing variables: {missing_variables}"
    )


# ============================================================
# Target classes
# ============================================================

target_emotions = [
    "Anger",
    "Fear",
]

target_class_indices = {
    emotion_name: emotion_names.index(emotion_name)
    for emotion_name in target_emotions
}

print("Target classes:", target_class_indices)


# ============================================================
# ONNX Runtime subprocess
#
# Each complete dataset is evaluated in one dynamic-batch call.
# ============================================================

source_inference_script = r"""
import sys
import numpy as np
import onnxruntime as ort

model_path = sys.argv[1]
input_path = sys.argv[2]
output_path = sys.argv[3]

X = np.load(input_path).astype(np.float32)

session = ort.InferenceSession(
    model_path,
    providers=["CPUExecutionProvider"],
)

model_input = session.get_inputs()[0]
model_output = session.get_outputs()[0]

probabilities = session.run(
    [model_output.name],
    {
        model_input.name: X,
    },
)[0]

probabilities = np.asarray(
    probabilities,
    dtype=np.float32,
)

np.save(
    output_path,
    probabilities,
)
"""


temporary_source_analysis_dir = Path(
    tempfile.mkdtemp(
        prefix="anger_fear_source_analysis_",
        dir="/content",
    )
)

class_metric_rows = []
confusion_destination_rows = []


# ============================================================
# Run source-specific inference
# ============================================================

for dataset_name, dataset in fear_analysis_datasets.items():
    X = np.asarray(
        dataset["X"],
        dtype=np.float32,
    )

    y_true = np.asarray(
        dataset["y"],
        dtype=np.int64,
    )

    input_path = (
        temporary_source_analysis_dir
        / f"{dataset_name}_input.npy"
    )

    output_path = (
        temporary_source_analysis_dir
        / f"{dataset_name}_probabilities.npy"
    )

    np.save(
        input_path,
        X,
    )

    subprocess.run(
        [
            sys.executable,
            "-c",
            source_inference_script,
            str(FEAR_ANALYSIS_ONNX_PATH),
            str(input_path),
            str(output_path),
        ],
        check=True,
        text=True,
        capture_output=True,
    )

    probabilities = np.load(
        output_path
    ).astype(np.float32)

    y_pred = np.argmax(
        probabilities,
        axis=1,
    ).astype(np.int64)

    assert probabilities.shape == (
        len(y_true),
        len(emotion_names),
    )

    # --------------------------------------------------------
    # Calculate Anger and Fear metrics
    # --------------------------------------------------------

    for emotion_name in target_emotions:
        class_index = target_class_indices[
            emotion_name
        ]

        true_mask = (
            y_true == class_index
        )

        predicted_mask = (
            y_pred == class_index
        )

        true_positive = int(
            np.sum(
                true_mask
                & predicted_mask
            )
        )

        n_true = int(
            true_mask.sum()
        )

        n_predicted = int(
            predicted_mask.sum()
        )

        recall = (
            true_positive / n_true
            if n_true > 0
            else np.nan
        )

        precision = (
            true_positive / n_predicted
            if n_predicted > 0
            else 0.0
        )

        f1 = (
            2.0 * precision * recall
            / (precision + recall)
            if precision + recall > 0
            else 0.0
        )

        class_metric_rows.append(
            {
                "dataset": dataset_name,
                "split_status": dataset[
                    "split_status"
                ],
                "emotion": emotion_name,
                "n_windows_total": int(
                    len(y_true)
                ),
                "n_true_windows": n_true,
                "n_correct": true_positive,
                "n_predicted_as_class": n_predicted,
                "precision": float(precision),
                "recall": float(recall),
                "f1": float(f1),
            }
        )

        # ----------------------------------------------------
        # Record where true Anger/Fear windows are predicted
        # ----------------------------------------------------

        true_class_predictions = y_pred[
            true_mask
        ]

        for predicted_class_index, predicted_emotion in enumerate(
            emotion_names
        ):
            count = int(
                np.sum(
                    true_class_predictions
                    == predicted_class_index
                )
            )

            rate = (
                count / n_true
                if n_true > 0
                else np.nan
            )

            confusion_destination_rows.append(
                {
                    "dataset": dataset_name,
                    "split_status": dataset[
                        "split_status"
                    ],
                    "true_emotion": emotion_name,
                    "predicted_emotion": predicted_emotion,
                    "count": count,
                    "rate": float(rate),
                }
            )


# ============================================================
# Create result tables
# ============================================================

anger_fear_metrics_df = pd.DataFrame(
    class_metric_rows
)

anger_fear_destinations_df = pd.DataFrame(
    confusion_destination_rows
)


# ============================================================
# Display the source-specific metric table
# ============================================================

print("ANGER AND FEAR — SOURCE-SPECIFIC METRICS")

display(
    anger_fear_metrics_df[
        [
            "dataset",
            "split_status",
            "emotion",
            "n_windows_total",
            "n_true_windows",
            "n_correct",
            "precision",
            "recall",
            "f1",
        ]
    ].round(4)
)


# ============================================================
# Display only the held-out recall table
# ============================================================

held_out_recall_df = (
    anger_fear_metrics_df[
        anger_fear_metrics_df[
            "split_status"
        ].str.contains(
            "held-out",
            case=False,
        )
    ]
    .pivot(
        index="dataset",
        columns="emotion",
        values="recall",
    )
    .reset_index()
)

held_out_support_df = (
    anger_fear_metrics_df[
        anger_fear_metrics_df[
            "split_status"
        ].str.contains(
            "held-out",
            case=False,
        )
    ]
    .pivot(
        index="dataset",
        columns="emotion",
        values="n_true_windows",
    )
    .reset_index()
)

print("\nHELD-OUT RECALL")
display(
    held_out_recall_df.round(4)
)

print("\nHELD-OUT SUPPORT")
display(
    held_out_support_df
)


# ============================================================
# Display confusion destinations for true Anger and Fear
# ============================================================

held_out_destinations_df = (
    anger_fear_destinations_df[
        anger_fear_destinations_df[
            "split_status"
        ].str.contains(
            "held-out",
            case=False,
        )
    ]
)

for true_emotion in target_emotions:
    print(
        f"\nTRUE {true_emotion.upper()} — "
        "PREDICTION DESTINATIONS"
    )

    destination_table = (
        held_out_destinations_df[
            held_out_destinations_df[
                "true_emotion"
            ] == true_emotion
        ]
        .pivot(
            index="dataset",
            columns="predicted_emotion",
            values="rate",
        )
        .reset_index()
    )

    display(
        destination_table.round(4)
    )


# ============================================================
# Save the analysis
# ============================================================

ANGER_FEAR_RESULTS_DIR = (
    Path(predictions_path)
    / DYNAMIC_EXPORT_TAG
)

ANGER_FEAR_METRICS_PATH = (
    ANGER_FEAR_RESULTS_DIR
    / "anger_fear_metrics_by_source.csv"
)

ANGER_FEAR_DESTINATIONS_PATH = (
    ANGER_FEAR_RESULTS_DIR
    / "anger_fear_prediction_destinations_by_source.csv"
)

anger_fear_metrics_df.to_csv(
    ANGER_FEAR_METRICS_PATH,
    index=False,
)

anger_fear_destinations_df.to_csv(
    ANGER_FEAR_DESTINATIONS_PATH,
    index=False,
)

print("\nSaved:")
print(ANGER_FEAR_METRICS_PATH)
print(ANGER_FEAR_DESTINATIONS_PATH)
````

````python
# ============================================================
# FINAL TEST EVALUATION — OVERALL METRICS
#
# Primary:
#   old VR acted test + new participant acted test
#
# Secondary:
#   EmojiHero test
#
# Diagnostic:
#   old VR acted, new VR acted and pooled acted
#
# Metrics are window-level.
# ============================================================

from pathlib import Path
import hashlib
import json

import numpy as np
import pandas as pd
import onnxruntime as ort

from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    f1_score,
    log_loss,
    precision_score,
    recall_score,
)


# ------------------------------------------------------------
# Preconditions
# ------------------------------------------------------------

assert "ONNX_OUTPUT_PATH" in globals()
assert Path(ONNX_OUTPUT_PATH).exists()
assert "X_test_vr_np" in globals()
assert "Y_test_vr_np" in globals()
assert "X_test_emojihero_np" in globals()
assert "Y_test_emojihero_np" in globals()


# ------------------------------------------------------------
# Result directory — refuse accidental overwrite
# ------------------------------------------------------------
MODEL_TAG = EVALUATION_MODEL_TAG
ONNX_MODEL_PATH = EVALUATION_ONNX_PATH
FINAL_EVAL_DIR = EVALUATION_OUTPUT_DIR

SUMMARY_CSV_PATH = (
    FINAL_EVAL_DIR / "final_test_summary.csv"
)

SUMMARY_JSON_PATH = (
    FINAL_EVAL_DIR / "final_test_summary.json"
)

FINAL_EVAL_DIR = (
    Path(TEACHER_RUN_DIR)
    / "final_test_evaluation"
)

FINAL_EVAL_DIR.mkdir(
    parents=True,
    exist_ok=True,
)

SUMMARY_CSV_PATH = (
    FINAL_EVAL_DIR
    / "overall_metrics.csv"
)

SUMMARY_JSON_PATH = (
    FINAL_EVAL_DIR
    / "overall_metrics.json"
)

if SUMMARY_CSV_PATH.exists() or SUMMARY_JSON_PATH.exists():
    raise FileExistsError(
        "Final test results already exist. "
        "Refusing to overwrite them:\n"
        f"{FINAL_EVAL_DIR}"
    )


# ------------------------------------------------------------
# Build source-specific test tensors
# ------------------------------------------------------------

X_test_old_acted_np = stack_sequences(
    X_test_act
).astype(np.float32)

Y_test_old_acted_np = to_numpy_labels(
    Y_test_act
)

X_test_new_acted_np = stack_sequences(
    X_test_new
).astype(np.float32)

Y_test_new_acted_np = to_numpy_labels(
    Y_test_new
)

X_test_vr_np = np.asarray(
    X_test_vr_np,
    dtype=np.float32,
)

Y_test_vr_np = np.asarray(
    Y_test_vr_np,
    dtype=np.int64,
).reshape(-1)

X_test_emojihero_np = np.asarray(
    X_test_emojihero_np,
    dtype=np.float32,
)

Y_test_emojihero_np = np.asarray(
    Y_test_emojihero_np,
    dtype=np.int64,
).reshape(-1)


# ------------------------------------------------------------
# Load ONNX
# ------------------------------------------------------------

onnx_session = ort.InferenceSession(
    str(ONNX_OUTPUT_PATH),
    providers=["CPUExecutionProvider"],
)

onnx_input = onnx_session.get_inputs()[0]
onnx_output = onnx_session.get_outputs()[0]

onnx_input_name = onnx_input.name
onnx_output_name = onnx_output.name

print("Evaluating ONNX:")
print(ONNX_OUTPUT_PATH)
print("Input:", onnx_input_name, onnx_input.shape)
print("Output:", onnx_output_name, onnx_output.shape)


# ------------------------------------------------------------
# Fixed-batch-1 ONNX inference
# ------------------------------------------------------------

def predict_onnx_batch1(X, label):
    X = np.asarray(
        X,
        dtype=np.float32,
    )

    assert X.shape[1:] == (
        SEQ_LENGTH,
        len(AU_NAMES),
    )

    rows = []

    for index in range(len(X)):
        sample = np.ascontiguousarray(
            X[index:index + 1],
            dtype=np.float32,
        )

        output = onnx_session.run(
            [onnx_output_name],
            {
                onnx_input_name: sample
            },
        )[0]

        output = np.asarray(
            output,
            dtype=np.float32,
        ).reshape(
            1,
            len(Emotion),
        )

        rows.append(output)

        if (index + 1) % 500 == 0:
            print(
                f"{label}: "
                f"{index + 1}/{len(X)}"
            )

    probabilities = np.concatenate(
        rows,
        axis=0,
    )

    assert probabilities.shape == (
        len(X),
        len(Emotion),
    )

    assert np.isfinite(
        probabilities
    ).all()

    assert np.allclose(
        probabilities.sum(axis=1),
        1.0,
        atol=1e-4,
        rtol=1e-4,
    )

    return probabilities


# ------------------------------------------------------------
# Metric helper
# ------------------------------------------------------------

CLASS_NAMES = [
    emotion.value
    for emotion in Emotion
]

CLASS_LABELS = np.arange(
    len(CLASS_NAMES)
)

def overall_metrics(
    dataset_name,
    y_true,
    probabilities,
):
    y_true = np.asarray(
        y_true,
        dtype=np.int64,
    ).reshape(-1)

    probabilities = np.asarray(
        probabilities,
        dtype=np.float32,
    )

    predictions = np.argmax(
        probabilities,
        axis=1,
    )

    clipped_probabilities = np.clip(
        probabilities,
        1e-12,
        1.0,
    )

    clipped_probabilities /= (
        clipped_probabilities.sum(
            axis=1,
            keepdims=True,
        )
    )

    return {
        "dataset": dataset_name,
        "n_windows": int(len(y_true)),
        "accuracy": float(
            accuracy_score(
                y_true,
                predictions,
            )
        ),
        "balanced_accuracy": float(
            balanced_accuracy_score(
                y_true,
                predictions,
            )
        ),
        "precision_macro": float(
            precision_score(
                y_true,
                predictions,
                labels=CLASS_LABELS,
                average="macro",
                zero_division=0,
            )
        ),
        "recall_macro": float(
            recall_score(
                y_true,
                predictions,
                labels=CLASS_LABELS,
                average="macro",
                zero_division=0,
            )
        ),
        "f1_macro": float(
            f1_score(
                y_true,
                predictions,
                labels=CLASS_LABELS,
                average="macro",
                zero_division=0,
            )
        ),
        "f1_weighted": float(
            f1_score(
                y_true,
                predictions,
                labels=CLASS_LABELS,
                average="weighted",
                zero_division=0,
            )
        ),
        "log_loss": float(
            log_loss(
                y_true,
                clipped_probabilities,
                labels=CLASS_LABELS,
            )
        ),
    }


# ------------------------------------------------------------
# Run each independent source once
# ------------------------------------------------------------

prob_old_acted = predict_onnx_batch1(
    X_test_old_acted_np,
    "old VR acted",
)

prob_new_acted = predict_onnx_batch1(
    X_test_new_acted_np,
    "new VR acted",
)

prob_emojihero = predict_onnx_batch1(
    X_test_emojihero_np,
    "EmojiHero",
)


# Primary VR pool
Y_test_vr_primary = np.concatenate(
    [
        Y_test_old_acted_np,
        Y_test_new_acted_np,
    ]
)

prob_vr_primary = np.concatenate(
    [
        prob_old_acted,
        prob_new_acted,
    ],
    axis=0,
)


# Pooled acted diagnostic
Y_test_all_acted_final = np.concatenate(
    [
        Y_test_vr_primary,
        Y_test_emojihero_np,
    ]
)

prob_all_acted = np.concatenate(
    [
        prob_vr_primary,
        prob_emojihero,
    ],
    axis=0,
)


# ------------------------------------------------------------
# Results
# ------------------------------------------------------------

evaluation_arrays = {
    "vr_acted_primary": {
        "y_true": Y_test_vr_primary,
        "probabilities": prob_vr_primary,
    },
    "old_vr_acted_diagnostic": {
        "y_true": Y_test_old_acted_np,
        "probabilities": prob_old_acted,
    },
    "new_vr_acted_diagnostic": {
        "y_true": Y_test_new_acted_np,
        "probabilities": prob_new_acted,
    },
    "emojihero_secondary": {
        "y_true": Y_test_emojihero_np,
        "probabilities": prob_emojihero,
    },
    "all_acted_pooled_diagnostic": {
        "y_true": Y_test_all_acted_final,
        "probabilities": prob_all_acted,
    },
}

overall_rows = []

for dataset_name, result in evaluation_arrays.items():
    result["predictions"] = np.argmax(
        result["probabilities"],
        axis=1,
    )

    metrics = overall_metrics(
        dataset_name,
        result["y_true"],
        result["probabilities"],
    )

    overall_rows.append(metrics)

    np.savez_compressed(
        FINAL_EVAL_DIR
        / f"{dataset_name}_predictions.npz",
        y_true=result["y_true"],
        y_pred=result["predictions"],
        y_probability=result["probabilities"],
        class_names=np.asarray(CLASS_NAMES),
    )


overall_df = pd.DataFrame(
    overall_rows
)

display(
    overall_df.style.format({
        "accuracy": "{:.4f}",
        "balanced_accuracy": "{:.4f}",
        "precision_macro": "{:.4f}",
        "recall_macro": "{:.4f}",
        "f1_macro": "{:.4f}",
        "f1_weighted": "{:.4f}",
        "log_loss": "{:.4f}",
    })
)


# ------------------------------------------------------------
# Save results
# ------------------------------------------------------------

overall_df.to_csv(
    SUMMARY_CSV_PATH,
    index=False,
)

onnx_sha256 = hashlib.sha256(
    Path(ONNX_OUTPUT_PATH).read_bytes()
).hexdigest()

summary_payload = {
    "schema_version": "acted-only-teacher-final-evaluation-v1",
    "teacher_tag": TEACHER_TAG,
    "onnx_file": Path(ONNX_OUTPUT_PATH).name,
    "onnx_sha256": onnx_sha256,
    "class_order": CLASS_NAMES,
    "metric_unit": "window",
    "primary_dataset": "vr_acted_primary",
    "secondary_dataset": "emojihero_secondary",
    "metrics": overall_rows,
}

with open(
    SUMMARY_JSON_PATH,
    "w",
    encoding="utf-8",
) as summary_file:
    json.dump(
        summary_payload,
        summary_file,
        indent=2,
        ensure_ascii=False,
    )

print("\nFinal evaluation saved:")
print(FINAL_EVAL_DIR)
````

````python
# ============================================================
# PER-EMOTION METRICS + CONFUSION MATRICES
# ============================================================

import json

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns

from sklearn.metrics import (
    classification_report,
    confusion_matrix,
)


# ------------------------------------------------------------
# Per-emotion metrics
# ------------------------------------------------------------

per_emotion_rows = []
confusion_matrices = {}

for dataset_name, result in evaluation_arrays.items():
    y_true = np.asarray(
        result["y_true"],
        dtype=np.int64,
    ).reshape(-1)

    predictions = np.asarray(
        result["predictions"],
        dtype=np.int64,
    ).reshape(-1)

    report = classification_report(
        y_true,
        predictions,
        labels=CLASS_LABELS,
        target_names=CLASS_NAMES,
        output_dict=True,
        zero_division=0,
    )

    matrix = confusion_matrix(
        y_true,
        predictions,
        labels=CLASS_LABELS,
    )

    confusion_matrices[dataset_name] = matrix

    for class_index, class_name in enumerate(
        CLASS_NAMES
    ):
        class_result = report[class_name]

        per_emotion_rows.append({
            "dataset": dataset_name,
            "class_index": class_index,
            "emotion": class_name,
            "precision": float(
                class_result["precision"]
            ),
            "recall": float(
                class_result["recall"]
            ),
            "f1": float(
                class_result["f1-score"]
            ),
            "support": int(
                class_result["support"]
            ),
        })


per_emotion_df = pd.DataFrame(
    per_emotion_rows
)

display(
    per_emotion_df.style.format({
        "precision": "{:.4f}",
        "recall": "{:.4f}",
        "f1": "{:.4f}",
    })
)


# ------------------------------------------------------------
# Easier comparison tables
# ------------------------------------------------------------

f1_table = (
    per_emotion_df
    .pivot(
        index="emotion",
        columns="dataset",
        values="f1",
    )
    .reindex(CLASS_NAMES)
)

recall_table = (
    per_emotion_df
    .pivot(
        index="emotion",
        columns="dataset",
        values="recall",
    )
    .reindex(CLASS_NAMES)
)

precision_table = (
    per_emotion_df
    .pivot(
        index="emotion",
        columns="dataset",
        values="precision",
    )
    .reindex(CLASS_NAMES)
)

print("\nPER-EMOTION F1")
display(
    f1_table.style.format("{:.4f}")
)

print("\nPER-EMOTION RECALL")
display(
    recall_table.style.format("{:.4f}")
)

print("\nPER-EMOTION PRECISION")
display(
    precision_table.style.format("{:.4f}")
)


# ------------------------------------------------------------
# Save metric tables
# ------------------------------------------------------------

per_emotion_df.to_csv(
    FINAL_EVAL_DIR
    / "per_emotion_metrics.csv",
    index=False,
)

f1_table.to_csv(
    FINAL_EVAL_DIR
    / "per_emotion_f1_comparison.csv"
)

recall_table.to_csv(
    FINAL_EVAL_DIR
    / "per_emotion_recall_comparison.csv"
)

precision_table.to_csv(
    FINAL_EVAL_DIR
    / "per_emotion_precision_comparison.csv"
)


# ------------------------------------------------------------
# Save raw and normalized confusion matrices
# ------------------------------------------------------------

for dataset_name, matrix in confusion_matrices.items():
    matrix_df = pd.DataFrame(
        matrix,
        index=[
            f"true_{name}"
            for name in CLASS_NAMES
        ],
        columns=[
            f"pred_{name}"
            for name in CLASS_NAMES
        ],
    )

    matrix_df.to_csv(
        FINAL_EVAL_DIR
        / f"{dataset_name}_confusion_counts.csv"
    )

    row_totals = matrix.sum(
        axis=1,
        keepdims=True,
    )

    normalized_matrix = (
        matrix.astype(np.float64)
        / np.maximum(row_totals, 1)
    )

    normalized_df = pd.DataFrame(
        normalized_matrix,
        index=[
            f"true_{name}"
            for name in CLASS_NAMES
        ],
        columns=[
            f"pred_{name}"
            for name in CLASS_NAMES
        ],
    )

    normalized_df.to_csv(
        FINAL_EVAL_DIR
        / f"{dataset_name}_confusion_normalized.csv"
    )


# ------------------------------------------------------------
# Plot confusion matrices
# ------------------------------------------------------------

plot_datasets = [
    "vr_acted_primary",
    "emojihero_secondary",
    "old_vr_acted_diagnostic",
    "new_vr_acted_diagnostic",
]

fig, axes = plt.subplots(
    2,
    2,
    figsize=(18, 15),
)

for axis, dataset_name in zip(
    axes.flat,
    plot_datasets,
):
    matrix = confusion_matrices[
        dataset_name
    ]

    normalized_matrix = (
        matrix.astype(np.float64)
        / np.maximum(
            matrix.sum(
                axis=1,
                keepdims=True,
            ),
            1,
        )
    )

    sns.heatmap(
        normalized_matrix,
        annot=True,
        fmt=".2f",
        cmap="Blues",
        vmin=0,
        vmax=1,
        xticklabels=CLASS_NAMES,
        yticklabels=CLASS_NAMES,
        ax=axis,
    )

    axis.set_title(dataset_name)
    axis.set_xlabel("Predicted")
    axis.set_ylabel("True")

plt.tight_layout()

confusion_figure_path = (
    FINAL_EVAL_DIR
    / "confusion_matrices.png"
)

plt.savefig(
    confusion_figure_path,
    dpi=200,
    bbox_inches="tight",
)

plt.show()


# ------------------------------------------------------------
# Per-emotion F1 chart
# ------------------------------------------------------------

selected_f1_table = f1_table[
    [
        "vr_acted_primary",
        "emojihero_secondary",
    ]
]

axis = selected_f1_table.plot(
    kind="bar",
    figsize=(13, 6),
    ylim=(0, 1),
)

axis.set_title(
    "Per-emotion F1: VR acted vs EmojiHero"
)
axis.set_xlabel("Emotion")
axis.set_ylabel("F1")
axis.grid(
    axis="y",
    alpha=0.3,
)

plt.xticks(
    rotation=30,
    ha="right",
)

plt.tight_layout()

f1_figure_path = (
    FINAL_EVAL_DIR
    / "per_emotion_f1.png"
)

plt.savefig(
    f1_figure_path,
    dpi=200,
    bbox_inches="tight",
)

plt.show()


print("\nPer-emotion evaluation saved:")
print(FINAL_EVAL_DIR)
````

````python
import json
import numpy as np
import pandas as pd
import onnxruntime as ort

from pathlib import Path
from sklearn.metrics import (
    f1_score,
    balanced_accuracy_score,
    classification_report,
    confusion_matrix,
)

ONNX_PATH = Path(
    "/content/sparseVR_teacher_model.onnx"
)

assert ONNX_PATH.exists(), f"ONNX bulunamadı: {ONNX_PATH}"

session = ort.InferenceSession(
    str(ONNX_PATH),
    providers=["CPUExecutionProvider"],
)

input_name = session.get_inputs()[0].name
output_name = session.get_outputs()[0].name

print("Input:", input_name, session.get_inputs()[0].shape)
print("Output:", output_name, session.get_outputs()[0].shape)
````

````python
def predict_onnx(session, X):
    X = np.asarray(X, dtype=np.float32)
    outputs = []

    # Even if the batch size is fixed, we pass the samples one by one for it to work.
    for i in range(len(X)):
        result = session.run(
            [output_name],
            {input_name: X[i:i+1]},
        )[0]

        outputs.append(np.asarray(result).reshape(-1, 7)[0])

    probabilities = np.asarray(outputs, dtype=np.float32)

    assert probabilities.shape == (len(X), 7)
    assert np.all(np.isfinite(probabilities))

    probability_sums = probabilities.sum(axis=1)

    print(
        "Probability sums:",
        probability_sums.min(),
        probability_sums.max(),
    )

    assert np.allclose(
        probability_sums,
        1.0,
        atol=1e-4,
    ), "The ONNX outputs are not probability distributions."

    return probabilities
````

````python
import json
import numpy as np
import pandas as pd

from pathlib import Path
from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    f1_score,
    balanced_accuracy_score,
)

CLASS_NAMES = list(le.classes_)

EXPECTED_CLASS_ORDER = [
    "Anger",
    "Disgust",
    "Fear",
    "Happiness",
    "Neutral",
    "Sadness",
    "Surprise",
]

assert CLASS_NAMES == EXPECTED_CLASS_ORDER, (
    f"Unexpected class order: {CLASS_NAMES}"
)

ANGER_INDEX = CLASS_NAMES.index("Anger")


def to_arrays(X, y):
    X_np = stack_sequences(list(X))
    y_np = np.asarray(list(y), dtype=np.int64)

    assert len(X_np) == len(y_np)
    assert X_np.shape[1:] == (5, 70)

    return X_np, y_np


def select_domain(X, y, domains, wanted_domain):
    selected_X = []
    selected_y = []

    for window, label, domain in zip(X, y, domains):
        if domain == wanted_domain:
            selected_X.append(window)
            selected_y.append(label)

    return selected_X, selected_y


def print_class_counts(name, y):
    counts = np.bincount(
        np.asarray(y, dtype=np.int64),
        minlength=len(CLASS_NAMES),
    )

    print(f"\n{name}: total={len(y)}")

    for class_name, count in zip(CLASS_NAMES, counts):
        print(f"  {class_name:<10}: {count}")


# Eski own-acted
X_val_old_acted_np, Y_val_old_acted_np = to_arrays(
    X_val_act,
    Y_val_act,
)

X_test_old_acted_np, Y_test_old_acted_np = to_arrays(
    X_test_act,
    Y_test_act,
)

# Yeni katılımcılardan yalnızca acted.
# Notebook'ta domain 1 = acted.
X_val_new_acted, Y_val_new_acted = select_domain(
    X_val_new,
    Y_val_new,
    Y_val_domain_new,
    wanted_domain=1,
)

X_test_new_acted, Y_test_new_acted = select_domain(
    X_test_new,
    Y_test_new,
    Y_test_domain_new,
    wanted_domain=1,
)

X_val_new_acted_np, Y_val_new_acted_np = to_arrays(
    X_val_new_acted,
    Y_val_new_acted,
)

X_test_new_acted_np, Y_test_new_acted_np = to_arrays(
    X_test_new_acted,
    Y_test_new_acted,
)

# Birleşik own-acted
X_val_own_acted_np = np.concatenate(
    [
        X_val_old_acted_np,
        X_val_new_acted_np,
    ],
    axis=0,
)

Y_val_own_acted_np = np.concatenate(
    [
        Y_val_old_acted_np,
        Y_val_new_acted_np,
    ],
    axis=0,
)

X_test_own_acted_np = np.concatenate(
    [
        X_test_old_acted_np,
        X_test_new_acted_np,
    ],
    axis=0,
)

Y_test_own_acted_np = np.concatenate(
    [
        Y_test_old_acted_np,
        Y_test_new_acted_np,
    ],
    axis=0,
)

print_class_counts(
    "Old acted validation",
    Y_val_old_acted_np,
)

print_class_counts(
    "New acted validation",
    Y_val_new_acted_np,
)

print_class_counts(
    "Combined own-acted validation",
    Y_val_own_acted_np,
)

print_class_counts(
    "Combined own-acted test",
    Y_test_own_acted_np,
)
````

````python
def reproduce_old_acted_subject_split():
    all_files = all_files_list(DataType.OUR_ACTED)

    files, emotions = zip(*all_files)

    labels = np.asarray([
        le.transform([emotion.value])[0]
        for emotion in emotions
    ])

    groups = np.asarray([
        subject_from_path(path)
        for path in files
    ])

    splitter = StratifiedGroupKFold(
        n_splits=10,
        shuffle=True,
        random_state=42,
    )

    dummy_X = np.zeros((len(labels), 1))

    folds = [
        heldout_indices
        for _, heldout_indices in splitter.split(
            dummy_X,
            labels,
            groups=groups,
        )
    ]

    val_indices = folds[7]
    test_indices = np.concatenate(folds[8:])

    old_val_subjects = sorted(
        set(groups[val_indices])
    )

    old_test_subjects = sorted(
        set(groups[test_indices])
    )

    return old_val_subjects, old_test_subjects


old_val_subjects, old_test_subjects = (
    reproduce_old_acted_subject_split()
)

new_val_acted_records = new_val_records[
    new_val_records["recording_type"] == "ActingRecordings"
]

new_test_acted_records = new_test_records[
    new_test_records["recording_type"] == "ActingRecordings"
]

new_val_subjects = sorted(
    new_val_acted_records["participant"].unique()
)

new_test_subjects = sorted(
    new_test_acted_records["participant"].unique()
)

print("Old acted validation subjects:", old_val_subjects)
print("New acted validation subjects:", new_val_subjects)

print("Old acted test subjects:", old_test_subjects)
print("New acted test subjects:", new_test_subjects)

all_val_subjects = sorted(
    set(old_val_subjects) |
    set(new_val_subjects)
)

all_test_subjects = sorted(
    set(old_test_subjects) |
    set(new_test_subjects)
)

print("\nTotal own-acted validation participants:",
      len(all_val_subjects))

print("Total own-acted test participants:",
      len(all_test_subjects))

print("Validation participants:", all_val_subjects)
print("Test participants:", all_test_subjects)

assert set(all_val_subjects).isdisjoint(all_test_subjects)
````

````python
validation_probabilities = {
    "old_acted": predict_onnx(
        session,
        X_val_old_acted_np,
    ),

    "new_acted": predict_onnx(
        session,
        X_val_new_acted_np,
    ),

    "own_acted_combined": predict_onnx(
        session,
        X_val_own_acted_np,
    ),

    "emojihero": predict_onnx(
        session,
        stack_sequences(X_val_em_hero),
    ),
}

validation_labels = {
    "old_acted": Y_val_old_acted_np,
    "new_acted": Y_val_new_acted_np,
    "own_acted_combined": Y_val_own_acted_np,
    "emojihero": np.asarray(
        Y_val_em_hero,
        dtype=np.int64,
    ),
}

test_probabilities = {
    "old_acted": predict_onnx(
        session,
        X_test_old_acted_np,
    ),

    "new_acted": predict_onnx(
        session,
        X_test_new_acted_np,
    ),

    "own_acted_combined": predict_onnx(
        session,
        X_test_own_acted_np,
    ),

    "emojihero": predict_onnx(
        session,
        stack_sequences(X_test_em_hero),
    ),
}

test_labels = {
    "old_acted": Y_test_old_acted_np,
    "new_acted": Y_test_new_acted_np,
    "own_acted_combined": Y_test_own_acted_np,
    "emojihero": np.asarray(
        Y_test_em_hero,
        dtype=np.int64,
    ),
}

# Own acted + EmojiHero, yalnızca ikincil raporlama için
test_probabilities["all_acted"] = np.concatenate(
    [
        test_probabilities["own_acted_combined"],
        test_probabilities["emojihero"],
    ],
    axis=0,
)

test_labels["all_acted"] = np.concatenate(
    [
        test_labels["own_acted_combined"],
        test_labels["emojihero"],
    ],
    axis=0,
)

print("Probability generation complete.")
````

````python
def apply_anger_weight(probabilities, weight):
    raw_scores = np.asarray(
        probabilities,
        dtype=np.float64,
    )

    decision_scores = raw_scores.copy()

    decision_scores[:, ANGER_INDEX] *= weight

    predictions = decision_scores.argmax(axis=1)

    adjusted_probabilities = (
        decision_scores /
        decision_scores.sum(axis=1, keepdims=True)
    )

    return predictions, adjusted_probabilities


def evaluate_weight(y_true, probabilities, weight):
    predictions, _ = apply_anger_weight(
        probabilities,
        weight,
    )

    report = classification_report(
        y_true,
        predictions,
        labels=range(len(CLASS_NAMES)),
        target_names=CLASS_NAMES,
        output_dict=True,
        zero_division=0,
    )

    result = {
        "weight": float(weight),

        "macro_f1": f1_score(
            y_true,
            predictions,
            labels=range(len(CLASS_NAMES)),
            average="macro",
            zero_division=0,
        ),

        "balanced_accuracy": balanced_accuracy_score(
            y_true,
            predictions,
        ),

        "anger_precision": report["Anger"]["precision"],
        "anger_recall": report["Anger"]["recall"],
        "anger_f1": report["Anger"]["f1-score"],
    }

    for class_name in CLASS_NAMES:
        result[f"{class_name.lower()}_recall"] = (
            report[class_name]["recall"]
        )

    return result


def scan_weights(y_true, probabilities):
    weights = np.round(
        np.arange(1.0, 4.001, 0.01),
        2,
    )

    rows = [
        evaluate_weight(
            y_true,
            probabilities,
            weight,
        )
        for weight in weights
    ]

    results = pd.DataFrame(rows)

    best_macro_f1 = results["macro_f1"].max()

    candidates = results[
        np.isclose(
            results["macro_f1"],
            best_macro_f1,
            atol=1e-12,
        )
    ]
    best = candidates.sort_values("weight").iloc[0]

    return results, best
````

````python
validation_scans = {}
validation_best_rows = {}

for source_name in [
    "old_acted",
    "new_acted",
    "own_acted_combined",
]:
    scan, best = scan_weights(
        validation_labels[source_name],
        validation_probabilities[source_name],
    )

    validation_scans[source_name] = scan
    validation_best_rows[source_name] = best

best_summary = pd.DataFrame([
    {
        "source": source_name,
        **best.to_dict(),
    }
    for source_name, best
    in validation_best_rows.items()
])

display(
    best_summary[
        [
            "source",
            "weight",
            "macro_f1",
            "balanced_accuracy",
            "anger_precision",
            "anger_recall",
            "anger_f1",
        ]
    ].round(4)
)

selected_weight = float(
    validation_best_rows[
        "own_acted_combined"
    ]["weight"]
)

print("Candidate weight:", selected_weight)
````

````python
validation_comparison_rows = []

for source_name, probabilities in (
    validation_probabilities.items()
):
    y_true = validation_labels[source_name]

    baseline = evaluate_weight(
        y_true,
        probabilities,
        1.0,
    )

    adjusted = evaluate_weight(
        y_true,
        probabilities,
        selected_weight,
    )

    validation_comparison_rows.append({
        "source": source_name,
        "version": "baseline",
        **baseline,
    })

    validation_comparison_rows.append({
        "source": source_name,
        "version": "adjusted",
        **adjusted,
    })

validation_comparison = pd.DataFrame(
    validation_comparison_rows
)

display(
    validation_comparison[
        [
            "source",
            "version",
            "weight",
            "macro_f1",
            "balanced_accuracy",
            "anger_precision",
            "anger_recall",
            "anger_f1",
            "neutral_recall",
            "disgust_recall",
        ]
    ].round(4)
)
````

````python
import matplotlib.pyplot as plt
import seaborn as sns


def plot_confusion(
    source_name,
    weight,
    title,
):
    y_true = validation_labels[source_name]
    probabilities = validation_probabilities[source_name]

    predictions, _ = apply_anger_weight(
        probabilities,
        weight,
    )

    matrix = confusion_matrix(
        y_true,
        predictions,
        labels=range(len(CLASS_NAMES)),
    )

    plt.figure(figsize=(9, 7))

    sns.heatmap(
        matrix,
        annot=True,
        fmt="d",
        cmap="Blues",
        xticklabels=CLASS_NAMES,
        yticklabels=CLASS_NAMES,
    )

    plt.xlabel("Predicted")
    plt.ylabel("True")
    plt.title(title)
    plt.tight_layout()
    plt.show()


plot_confusion(
    "own_acted_combined",
    1.0,
    "Own-acted validation — baseline",
)

plot_confusion(
    "own_acted_combined",
    selected_weight,
    f"Own-acted validation — Anger weight={selected_weight}",
)

plot_confusion(
    "emojihero",
    selected_weight,
    f"EmojiHero validation — Anger weight={selected_weight}",
)
````

````python
import numpy as np


def roughness(W, min_std=1e-3):
    """Temporal roughness of a window set.

    W : array (n_windows, n_frames, n_channels)

    For each channel, computes std(frame-to-frame difference) / std(signal)
    within each window, then averages over windows. Smoothed (time-averaged)
    input gives a low ratio; independently sampled input gives a high one.
    White noise tends to sqrt(2) ~ 1.41.
    """
    W = np.asarray(W, dtype=np.float64)

    ratios = []
    for channel in range(W.shape[2]):
        x = W[:, :, channel]

        signal_std = x.std(axis=1)
        diff_std = np.diff(x, axis=1).std(axis=1)

        active = signal_std > min_std
        if active.sum() >= 5:
            ratios.append((diff_std[active] / signal_std[active]).mean())

    return np.array(ratios)


sources = {
    "old acted    (unknown)":       X_val_old_acted_np,
    "new acted    (mean of 6)":     X_val_new_acted_np,
    "EmojiHeroVR  (mean of 3)":     stack_sequences(X_val_em_hero),
    "old natural  (unknown)":       stack_sequences(list(X_val_nat)),
}

print(f"{'source':<28} {'windows':>8} {'channels':>9} {'mean r':>8} {'median r':>10}")
print("-" * 67)

for name, windows in sources.items():
    r = roughness(windows)
    print(f"{name:<28} {len(windows):>8} {len(r):>9} "
          f"{r.mean():>8.3f} {np.median(r):>10.3f}")

print()
print("Live headset reference (measured on logged sessions): mean r = 1.18 - 1.25")
print("White noise ~ 1.41   |   heavily smoothed << 1")
````

````python
# ============================================================
# COMPARE 3 ONNX MODELS ON ACTED-ONLY HELD-OUT TEST DATA
#
# Acted domain includes:
#   - old acted test data
#   - EmojiHeroVR test data
#   - held-out new-participant ActingRecordings
#
# Natural / VideoRecordings are excluded.
# ============================================================

from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    classification_report,
    confusion_matrix,
    f1_score,
    log_loss,
)

try:
    import onnxruntime as ort
except ImportError:
    raise ImportError(
        "onnxruntime is not installed. Run this once in a separate cell:\n"
        "!pip install -q onnxruntime"
    )


# ------------------------------------------------------------
# 1. PUT YOUR THREE ONNX MODEL PATHS HERE
# ------------------------------------------------------------

MODEL_PATHS = {
    "model_1": Path(
        "/content/act_trial_92.onnx"
    ),
    "model_2": Path(
        "/content/natural_trial_96.onnx"
    ),
    "model_3": Path(
        "/content/sparseVR_teacher_model.onnx"
    ),
}


# ------------------------------------------------------------
# 2. CLASS ORDER
#
# This must match the Keras/ONNX/Unity output order.
# ------------------------------------------------------------

CLASS_NAMES = [
    "Anger",
    "Disgust",
    "Fear",
    "Happiness",
    "Neutral",
    "Sadness",
    "Surprise",
]

CLASS_LABELS = np.arange(len(CLASS_NAMES))


# ------------------------------------------------------------
# 3. BUILD THE ACTED-ONLY TEST SET
# ------------------------------------------------------------

required_variables = [
    "X_test_comb_np",
    "Y_test_comb_np",
    "Y_test_domain_comb_np",
]

missing_variables = [
    name for name in required_variables
    if name not in globals()
]

if missing_variables:
    raise RuntimeError(
        "Run the dataset-construction cells first. "
        f"Missing variables: {missing_variables}"
    )


X_test_all = np.asarray(
    X_test_comb_np,
    dtype=np.float32,
)

y_test_all = np.asarray(
    Y_test_comb_np,
    dtype=np.int64,
).reshape(-1)

domain_test_all = np.asarray(
    Y_test_domain_comb_np,
    dtype=np.int64,
).reshape(-1)


assert len(X_test_all) == len(y_test_all)
assert len(y_test_all) == len(domain_test_all)
assert X_test_all.shape[1:] == (5, 70)
assert set(np.unique(domain_test_all)).issubset({0, 1})


acted_mask = domain_test_all == 1

X_acted = np.ascontiguousarray(
    X_test_all[acted_mask],
    dtype=np.float32,
)

y_acted = y_test_all[acted_mask]


assert len(X_acted) > 0
assert set(np.unique(y_acted)) == set(CLASS_LABELS)

print("Acted-only test set")
print("--------------------")
print("X shape:", X_acted.shape)
print("y shape:", y_acted.shape)

print("\nClass distribution:")
for class_index, class_name in enumerate(CLASS_NAMES):
    count = int(np.sum(y_acted == class_index))
    print(f"  {class_name:12s}: {count}")


# ------------------------------------------------------------
# 4. HELPER FUNCTIONS
# ------------------------------------------------------------

def softmax(logits):
    """Numerically stable softmax."""
    logits = np.asarray(logits, dtype=np.float64)
    shifted = logits - np.max(logits, axis=1, keepdims=True)
    exp_values = np.exp(shifted)

    return (
        exp_values
        / np.sum(exp_values, axis=1, keepdims=True)
    ).astype(np.float32)


def convert_to_probabilities(raw_output):
    """
    Keep valid probability outputs unchanged.

    If the model returns logits, apply softmax for probability-based
    metrics. Argmax is unchanged by this conversion.
    """
    raw_output = np.asarray(raw_output, dtype=np.float32)

    if raw_output.ndim > 2:
        raw_output = np.squeeze(raw_output)

    if raw_output.ndim == 1:
        raw_output = raw_output.reshape(1, -1)

    if raw_output.shape[1] != len(CLASS_NAMES):
        raise ValueError(
            "Expected 7 model outputs, "
            f"but received shape {raw_output.shape}"
        )

    row_sums = raw_output.sum(axis=1)

    already_probabilities = (
        np.all(np.isfinite(raw_output))
        and np.all(raw_output >= -1e-6)
        and np.all(raw_output <= 1.0 + 1e-6)
        and np.allclose(
            row_sums,
            1.0,
            atol=1e-4,
            rtol=1e-4,
        )
    )

    if already_probabilities:
        return raw_output, "probabilities"

    return softmax(raw_output), "logits → softmax"


def run_onnx_model(model_path, X):
    """
    Run an ONNX model.

    First attempts batched inference. If the model has a fixed batch
    size of 1, it automatically falls back to one window at a time.
    """
    if not model_path.exists():
        raise FileNotFoundError(
            f"ONNX model not found:\n{model_path}"
        )

    session = ort.InferenceSession(
        str(model_path),
        providers=["CPUExecutionProvider"],
    )

    inputs = session.get_inputs()
    outputs = session.get_outputs()

    if len(inputs) != 1:
        raise ValueError(
            f"{model_path.name} has {len(inputs)} inputs; expected 1."
        )

    input_meta = inputs[0]
    output_meta = outputs[0]

    input_name = input_meta.name
    output_name = output_meta.name

    print(f"\nLoading: {model_path.name}")
    print("  input name: ", input_name)
    print("  input shape:", input_meta.shape)
    print("  output name:", output_name)
    print("  output shape:", output_meta.shape)

    try:
        raw_output = session.run(
            [output_name],
            {input_name: X},
        )[0]

        inference_mode = "full batch"

    except Exception as batch_error:
        print(
            "  Full-batch inference was unavailable; "
            "falling back to batch size 1."
        )

        predictions = []

        for index in range(len(X)):
            sample = X[index:index + 1]

            sample_output = session.run(
                [output_name],
                {input_name: sample},
            )[0]

            predictions.append(
                np.asarray(sample_output).reshape(1, -1)
            )

        raw_output = np.concatenate(
            predictions,
            axis=0,
        )

        inference_mode = "batch size 1"

    raw_output = np.asarray(raw_output)

    if raw_output.ndim > 2:
        raw_output = np.squeeze(raw_output)

    if raw_output.ndim == 1:
        raw_output = raw_output.reshape(1, -1)

    probabilities, output_type = convert_to_probabilities(
        raw_output
    )

    return {
        "session": session,
        "raw_output": raw_output,
        "probabilities": probabilities,
        "output_type": output_type,
        "inference_mode": inference_mode,
        "input_name": input_name,
        "output_name": output_name,
    }


# ------------------------------------------------------------
# 5. RUN AND EVALUATE ALL THREE MODELS
# ------------------------------------------------------------

summary_rows = []
per_class_rows = []
all_results = {}

for model_name, model_path in MODEL_PATHS.items():

    inference_result = run_onnx_model(
        model_path,
        X_acted,
    )

    probabilities = inference_result["probabilities"]

    if len(probabilities) != len(y_acted):
        raise ValueError(
            f"{model_name}: expected {len(y_acted)} predictions, "
            f"received {len(probabilities)}."
        )

    predictions = np.argmax(
        probabilities,
        axis=1,
    )

    report = classification_report(
        y_acted,
        predictions,
        labels=CLASS_LABELS,
        target_names=CLASS_NAMES,
        output_dict=True,
        zero_division=0,
    )

    cm = confusion_matrix(
        y_acted,
        predictions,
        labels=CLASS_LABELS,
    )

    accuracy = accuracy_score(
        y_acted,
        predictions,
    )

    balanced_accuracy = balanced_accuracy_score(
        y_acted,
        predictions,
    )

    macro_f1 = f1_score(
        y_acted,
        predictions,
        labels=CLASS_LABELS,
        average="macro",
        zero_division=0,
    )

    weighted_f1 = f1_score(
        y_acted,
        predictions,
        labels=CLASS_LABELS,
        average="weighted",
        zero_division=0,
    )

    model_log_loss = log_loss(
        y_acted,
        probabilities,
        labels=CLASS_LABELS,
    )

    summary_rows.append({
        "model": model_name,
        "file": model_path.name,
        "n_samples": len(y_acted),
        "accuracy": accuracy,
        "balanced_accuracy": balanced_accuracy,
        "macro_f1": macro_f1,
        "weighted_f1": weighted_f1,
        "log_loss": model_log_loss,
        "output_type": inference_result["output_type"],
        "inference_mode": inference_result["inference_mode"],
    })

    for class_name in CLASS_NAMES:
        class_result = report[class_name]

        per_class_rows.append({
            "model": model_name,
            "class": class_name,
            "precision": class_result["precision"],
            "recall": class_result["recall"],
            "f1": class_result["f1-score"],
            "support": int(class_result["support"]),
        })

    all_results[model_name] = {
        "model_path": model_path,
        "probabilities": probabilities,
        "predictions": predictions,
        "classification_report": report,
        "confusion_matrix": cm,
    }

    print("\n" + "=" * 72)
    print(f"{model_name}: {model_path.name}")
    print("=" * 72)

    print(
        classification_report(
            y_acted,
            predictions,
            labels=CLASS_LABELS,
            target_names=CLASS_NAMES,
            zero_division=0,
        )
    )


# ------------------------------------------------------------
# 6. OVERALL COMPARISON
# ------------------------------------------------------------

summary_df = pd.DataFrame(summary_rows).sort_values(
    "macro_f1",
    ascending=False,
).reset_index(drop=True)

print("\n\nOVERALL COMPARISON")
display(
    summary_df.style.format({
        "accuracy": "{:.4f}",
        "balanced_accuracy": "{:.4f}",
        "macro_f1": "{:.4f}",
        "weighted_f1": "{:.4f}",
        "log_loss": "{:.4f}",
    })
)


# ------------------------------------------------------------
# 7. PER-CLASS F1 COMPARISON
# ------------------------------------------------------------

per_class_df = pd.DataFrame(per_class_rows)

f1_comparison = (
    per_class_df
    .pivot(
        index="class",
        columns="model",
        values="f1",
    )
    .reindex(CLASS_NAMES)
)

recall_comparison = (
    per_class_df
    .pivot(
        index="class",
        columns="model",
        values="recall",
    )
    .reindex(CLASS_NAMES)
)

precision_comparison = (
    per_class_df
    .pivot(
        index="class",
        columns="model",
        values="precision",
    )
    .reindex(CLASS_NAMES)
)

print("\nPER-CLASS F1")
display(f1_comparison.style.format("{:.4f}"))

print("\nPER-CLASS RECALL")
display(recall_comparison.style.format("{:.4f}"))

print("\nPER-CLASS PRECISION")
display(precision_comparison.style.format("{:.4f}"))


# ------------------------------------------------------------
# 8. CONFUSION MATRICES
# ------------------------------------------------------------

number_of_models = len(MODEL_PATHS)

fig, axes = plt.subplots(
    1,
    number_of_models,
    figsize=(7 * number_of_models, 6),
)

if number_of_models == 1:
    axes = [axes]

for axis, model_name in zip(axes, MODEL_PATHS.keys()):
    cm = all_results[model_name]["confusion_matrix"]

    cm_normalized = (
        cm.astype(np.float64)
        / np.maximum(
            cm.sum(axis=1, keepdims=True),
            1,
        )
    )

    sns.heatmap(
        cm_normalized,
        annot=True,
        fmt=".2f",
        cmap="Blues",
        vmin=0,
        vmax=1,
        xticklabels=CLASS_NAMES,
        yticklabels=CLASS_NAMES,
        ax=axis,
    )

    axis.set_title(
        f"{model_name}\nActed-only test"
    )
    axis.set_xlabel("Predicted")
    axis.set_ylabel("True")

plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 9. F1 BAR CHART
# ------------------------------------------------------------

ax = f1_comparison.plot(
    kind="bar",
    figsize=(13, 6),
    ylim=(0, 1),
)

ax.set_title(
    "ONNX models — per-class F1 on acted-only held-out test data"
)
ax.set_xlabel("Emotion")
ax.set_ylabel("F1 score")
ax.grid(axis="y", alpha=0.3)
plt.xticks(rotation=30, ha="right")
plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 10. OPTIONAL: PAIRWISE PREDICTION AGREEMENT
# ------------------------------------------------------------

model_names = list(MODEL_PATHS.keys())

agreement_matrix = pd.DataFrame(
    index=model_names,
    columns=model_names,
    dtype=float,
)

for model_a in model_names:
    for model_b in model_names:
        prediction_a = all_results[model_a]["predictions"]
        prediction_b = all_results[model_b]["predictions"]

        agreement_matrix.loc[model_a, model_b] = np.mean(
            prediction_a == prediction_b
        )

print("\nPAIRWISE ARGMAX AGREEMENT")
display(agreement_matrix.style.format("{:.2%}"))
````

````python
# ============================================================================
# LOAD THE SAVED TEACHER AND TEST ON ACTED TEST DATA ONLY
#
# Domain convention:
#   0 = natural
#   1 = acted
#
# EmojiHeroVR is classified as acted in this notebook.
# This cell performs inference only—no training or tuning.
# ============================================================================

from pathlib import Path

import numpy as np
import pandas as pd

from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    classification_report,
    confusion_matrix,
    f1_score,
    precision_score,
    recall_score,
)


# ---------------------------------------------------------------------------
# 1. Select only acted windows from the complete held-out test corpus
# ---------------------------------------------------------------------------

X_test_all = np.asarray(
    X_test_comb_np,
    dtype=np.float32,
)

y_test_all = np.asarray(
    Y_test_comb_np,
    dtype=np.int64,
).reshape(-1)

domain_test_all = np.asarray(
    Y_test_domain_comb_np,
    dtype=np.int64,
).reshape(-1)


assert len(X_test_all) == len(y_test_all) == len(domain_test_all)

assert set(np.unique(domain_test_all)).issubset({0, 1}), (
    f"Unexpected domain labels: {np.unique(domain_test_all)}"
)


acted_mask = domain_test_all == 1

X_test_acted_only = X_test_all[acted_mask]
Y_test_acted_only = y_test_all[acted_mask]


assert len(Y_test_acted_only) > 0, (
    "No acted test windows were found."
)

assert X_test_acted_only.shape[1:] == (
    SEQ_LENGTH,
    len(AU_NAMES),
), X_test_acted_only.shape


# ---------------------------------------------------------------------------
# 2. Load the saved teacher model
# ---------------------------------------------------------------------------

teacher_model_path = Path(
    "/content/drive/MyDrive/"
    "outputs_15_participants_70AU-NOT_normalize_teacher_data-"
    "self_filtered_NATURAL_ONLY-EmojiHero_Fixed/"
    "models/"
    "teacher-70AU-MoE_3experts_2gate-with_15_participants-"
    "NOT-normalized_every_teacher_dataset-"
    "self_filtered_ONLY_NATURAL-EmojiHero_fixed.keras"
)


if not teacher_model_path.exists():
    raise FileNotFoundError(
        f"Teacher model was not found:\n{teacher_model_path}"
    )


# Rebuild the exact teacher architecture.
acted_teacher_model, _ = build_moe_single_input(
    input_shape=(
        SEQ_LENGTH,
        input_size,
    ),
    num_classes=num_classes,
    tau_init=2.0,
    tau_final=0.7,
    gate_hidden=128,
    classifier_hidden=256,
    lb_coef=0.02,
    ent_coef=0.01,
)


# Restore the saved teacher weights.
acted_teacher_model.load_weights(
    teacher_model_path
)

print("Loaded teacher model:")
print(teacher_model_path)


# ---------------------------------------------------------------------------
# 3. Evaluate the teacher on acted test data only
# ---------------------------------------------------------------------------

acted_test_loss, _ = acted_teacher_model.evaluate(
    X_test_acted_only,
    Y_test_acted_only,
    batch_size=64,
    verbose=0,
)


acted_probabilities = acted_teacher_model.predict(
    X_test_acted_only,
    batch_size=64,
    verbose=0,
)

acted_predictions = np.argmax(
    acted_probabilities,
    axis=1,
)


# ---------------------------------------------------------------------------
# 4. Calculate evaluation metrics
# ---------------------------------------------------------------------------

class_names = [
    emotion.value
    for emotion in Emotion
]

class_labels = np.arange(
    len(class_names)
)


acted_summary = pd.DataFrame([
    {
        "model": teacher_model_path.name,
        "test_set": "acted_only_including_EmojiHero",
        "n_samples": int(
            len(Y_test_acted_only)
        ),
        "loss": float(
            acted_test_loss
        ),
        "accuracy": accuracy_score(
            Y_test_acted_only,
            acted_predictions,
        ),
        "balanced_accuracy": balanced_accuracy_score(
            Y_test_acted_only,
            acted_predictions,
        ),
        "precision_macro": precision_score(
            Y_test_acted_only,
            acted_predictions,
            labels=class_labels,
            average="macro",
            zero_division=0,
        ),
        "recall_macro": recall_score(
            Y_test_acted_only,
            acted_predictions,
            labels=class_labels,
            average="macro",
            zero_division=0,
        ),
        "f1_macro": f1_score(
            Y_test_acted_only,
            acted_predictions,
            labels=class_labels,
            average="macro",
            zero_division=0,
        ),
        "f1_weighted": f1_score(
            Y_test_acted_only,
            acted_predictions,
            labels=class_labels,
            average="weighted",
            zero_division=0,
        ),
    }
])


print(
    f"\nActed-only test windows: "
    f"{len(Y_test_acted_only):,}"
)

display(acted_summary)


# ---------------------------------------------------------------------------
# 5. Per-class results
# ---------------------------------------------------------------------------

print("\nClassification report:")

print(
    classification_report(
        Y_test_acted_only,
        acted_predictions,
        labels=class_labels,
        target_names=class_names,
        zero_division=0,
    )
)


# ---------------------------------------------------------------------------
# 6. Confusion matrix
# ---------------------------------------------------------------------------

acted_confusion_matrix = pd.DataFrame(
    confusion_matrix(
        Y_test_acted_only,
        acted_predictions,
        labels=class_labels,
    ),
    index=[
        f"true_{name}"
        for name in class_names
    ],
    columns=[
        f"pred_{name}"
        for name in class_names
    ],
)

display(acted_confusion_matrix)
````

````python
import json
import numpy as np
import pandas as pd
from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    classification_report,
    cohen_kappa_score,
    confusion_matrix,
    f1_score,
    log_loss,
    matthews_corrcoef,
    precision_score,
    recall_score,
    roc_auc_score,
    average_precision_score,
    top_k_accuracy_score,
)
from sklearn.preprocessing import label_binarize


def compute_full_metrics(model, X, y_true, class_labels, class_names, model_name="model", test_name="test"):
    """
    Run a model on (X, y_true) and compute the full metric suite.
    Returns a dict with scalar metrics, per-class metrics, confusion matrices,
    and the raw predictions/probabilities (so plots can reuse them without re-predicting).
    """
    y_true = np.reshape(np.asarray(y_true), -1)
    y_proba = model.predict(X, batch_size=64, verbose=0)
    y_proba = np.asarray(y_proba)
    y_pred = np.argmax(y_proba, axis=1)

    n_classes = len(class_labels)

    # --- Basic ---
    acc = accuracy_score(y_true, y_pred)
    bal_acc = balanced_accuracy_score(y_true, y_pred)

    # --- Precision/Recall/F1 (aggregate) ---
    prec_macro = precision_score(y_true, y_pred, labels=class_labels, average="macro", zero_division=0)
    rec_macro = recall_score(y_true, y_pred, labels=class_labels, average="macro", zero_division=0)
    f1_macro = f1_score(y_true, y_pred, labels=class_labels, average="macro", zero_division=0)

    prec_weighted = precision_score(y_true, y_pred, labels=class_labels, average="weighted", zero_division=0)
    rec_weighted = recall_score(y_true, y_pred, labels=class_labels, average="weighted", zero_division=0)
    f1_weighted = f1_score(y_true, y_pred, labels=class_labels, average="weighted", zero_division=0)

    prec_micro = precision_score(y_true, y_pred, labels=class_labels, average="micro", zero_division=0)
    rec_micro = recall_score(y_true, y_pred, labels=class_labels, average="micro", zero_division=0)
    f1_micro = f1_score(y_true, y_pred, labels=class_labels, average="micro", zero_division=0)

    # --- Agreement metrics ---
    kappa = cohen_kappa_score(y_true, y_pred, labels=class_labels)
    mcc = matthews_corrcoef(y_true, y_pred)

    # --- Top-2 accuracy ---
    present_labels = np.unique(y_true)
    try:
        top2_acc = top_k_accuracy_score(y_true, y_proba, k=2, labels=class_labels)
    except Exception as e:
        top2_acc = np.nan
        print(f"[{model_name} | {test_name}] top-2 accuracy skipped: {e}")

    # --- Log loss ---
    try:
        logloss = log_loss(y_true, y_proba, labels=class_labels)
    except Exception as e:
        logloss = np.nan
        print(f"[{model_name} | {test_name}] log loss skipped: {e}")

    # --- ROC-AUC and PR-AUC (One-vs-Rest); only well-defined if >=2 classes present in y_true ---
    roc_auc_macro = roc_auc_score_safe = np.nan
    roc_auc_weighted = np.nan
    pr_auc_macro = np.nan
    pr_auc_weighted = np.nan
    if len(present_labels) >= 2:
        y_true_bin = label_binarize(y_true, classes=class_labels)
        try:
            roc_auc_macro = roc_auc_score(y_true_bin, y_proba, average="macro", multi_class="ovr")
            roc_auc_weighted = roc_auc_score(y_true_bin, y_proba, average="weighted", multi_class="ovr")
        except Exception as e:
            print(f"[{model_name} | {test_name}] ROC-AUC skipped: {e}")
        try:
            pr_auc_macro = average_precision_score(y_true_bin, y_proba, average="macro")
            pr_auc_weighted = average_precision_score(y_true_bin, y_proba, average="weighted")
        except Exception as e:
            print(f"[{model_name} | {test_name}] PR-AUC skipped: {e}")
    else:
        print(f"[{model_name} | {test_name}] Only one class present in y_true; ROC-AUC/PR-AUC undefined, skipped.")

    # --- Per-class report ---
    report_dict = classification_report(
        y_true, y_pred, labels=class_labels, target_names=class_names,
        output_dict=True, zero_division=0,
    )
    per_class_df = pd.DataFrame(report_dict).T.loc[class_names]
    per_class_df.index.name = "class"

    # --- Confusion matrices ---
    cm = confusion_matrix(y_true, y_pred, labels=class_labels)
    with np.errstate(all="ignore"):
        cm_norm = cm.astype(np.float64) / cm.sum(axis=1, keepdims=True)
    cm_norm = np.nan_to_num(cm_norm)

    scalar_metrics = {
        "model": model_name,
        "test_set": test_name,
        "n_samples": int(len(y_true)),
        "accuracy": acc,
        "balanced_accuracy": bal_acc,
        "precision_macro": prec_macro,
        "recall_macro": rec_macro,
        "f1_macro": f1_macro,
        "precision_weighted": prec_weighted,
        "recall_weighted": rec_weighted,
        "f1_weighted": f1_weighted,
        "precision_micro": prec_micro,
        "recall_micro": rec_micro,
        "f1_micro": f1_micro,
        "cohen_kappa": kappa,
        "matthews_corrcoef": mcc,
        "top2_accuracy": top2_acc,
        "log_loss": logloss,
        "roc_auc_macro_ovr": roc_auc_macro,
        "roc_auc_weighted_ovr": roc_auc_weighted,
        "pr_auc_macro_ovr": pr_auc_macro,
        "pr_auc_weighted_ovr": pr_auc_weighted,
    }

    return {
        "scalar_metrics": scalar_metrics,
        "per_class_df": per_class_df,
        "confusion_matrix": cm,
        "confusion_matrix_normalized": cm_norm,
        "y_true": y_true,
        "y_pred": y_pred,
        "y_proba": y_proba,
    }


print("compute_full_metrics() defined.")
````

### 7.4.1 Compact per-test-set results

`eval_and_pack` runs the trained model on one test set and packages accuracy, macro-F1, the confusion
matrix, and the full classification report. Raw predictions are written as `.npz` and the whole set
is summarised in `summary.json`.

````python
# ---- Save teacher_plus_10participants (70-AU) results: natural / acted / combined test sets ----
teacher10_dir = TEACHER_RUN_DIR / "results_eval_and_pack"
teacher10_dir.mkdir(parents=True, exist_ok=True)

def eval_and_pack(model, X_test, Y_test, name):
    Y_pred = model.predict(X_test)
    Y_pred = np.argmax(Y_pred, axis=1)
    Y_test_flat = np.reshape(Y_test, -1)

    acc = accuracy_score(Y_test_flat, Y_pred)
    f1_macro = f1_score(Y_test_flat, Y_pred, average="macro")
    cm = confusion_matrix(Y_test_flat, Y_pred)
    report = classification_report(
        Y_test_flat, Y_pred,
        labels=range(len(Emotion)),
        target_names=[emo.value for emo in Emotion],
        output_dict=True,
    )

    np.savez(
        teacher10_dir / f"raw_predictions_{name}.npz",
        y_true=Y_test_flat,
        y_pred=Y_pred,
        confusion_matrix=cm,
    )

    return {
        "test_set": name,
        "n_samples": int(len(Y_test_flat)),
        "accuracy": float(acc),
        "f1_macro": float(f1_macro),
        "confusion_matrix": cm.tolist(),
        "classification_report": report,
    }

results_by_testset = {
    "natural": eval_and_pack(trained_model, X_test_nat_np, Y_test_nat_np, "natural"),
    "acted_plus_emojihero": eval_and_pack(trained_model, X_test_act_comb_np, Y_test_act_comb_np, "acted_plus_emojihero"),
    "combined": eval_and_pack(trained_model, X_test_comb_np, Y_test_comb_np, "combined"),
}

summary = {
    "model_name": TEACHER_MODEL_NAME,
    "input_shape": [SEQ_LENGTH, len(AU_NAMES)],
    "results_by_testset": results_by_testset,
}

with open(teacher10_dir / "summary.json", "w") as f:
    json.dump(summary, f, indent=2, default=float)

print("Saved teacher_plus_new_participants (70-AU) results to:", teacher10_dir)
for name, r in results_by_testset.items():
    print(f"  {name}: accuracy={r['accuracy']:.4f}  f1_macro={r['f1_macro']:.4f}  n={r['n_samples']}")
````

## 8.2 Load the Teacher by Rebuilding + Restoring Weights

Rebuilds the exact architecture from `build_moe_single_input()` and then loads only the weights.

This is the required workaround for the `Lambda` layer serialisation problem: calling
`keras.models.load_model()` on the saved file fails because the expert-stacking and fusion `Lambda`
layers cannot be reconstructed from their serialised form.

````python
def load_teacher_model(model_path, input_shape=(SEQ_LENGTH, input_size), num_classes=num_classes):
    # Rebuild the exact same architecture used at training time
    model, _ = build_moe_single_input(
        input_shape=input_shape, num_classes=num_classes,
        tau_init=2.0, tau_final=0.7,
        gate_hidden=128, classifier_hidden=256,
        lb_coef=0.02, ent_coef=0.01,
    )
    # Load only the weights, skip reconstructing the saved Functional graph
    model.load_weights(model_path)
    return model
````

## 8.4 Fine-Tune

Recompiles the loaded teacher with a deliberately small learning rate (SGD, 1e-4) so that the
pre-trained representation is nudged rather than overwritten, and trains for at most 15 epochs with
early stopping on validation loss (patience 3).

**Note:** the model is fine-tuned on the new participants only, but validated on the *combined*
validation set. Early stopping therefore selects the checkpoint that best preserves overall
performance rather than the one that best fits the new participants.

````python
from sklearn.utils.class_weight import compute_class_weight
import numpy as np
import tensorflow as tf
from tensorflow.keras import callbacks

# ---------------------------------------------------------
# 1. Reload the original teacher from disk.
# This guarantees that fine-tuning starts from the original
# teacher rather than an accidentally modified model.
# ---------------------------------------------------------
finetune_model = load_teacher_model(
    ORIGINAL_TEACHER_MODEL_PATH,
    input_shape=(SEQ_LENGTH, 70),
    num_classes=num_classes,
)

# ---------------------------------------------------------
# 2. Prepare the new-participant tensors.
# ---------------------------------------------------------
X_train_new_np = stack_sequences(X_train_new)
Y_train_new_np = to_numpy_labels(Y_train_new)

X_val_new_np = stack_sequences(X_val_new)
Y_val_new_np = to_numpy_labels(Y_val_new)

print("Fine-tune train shape:", X_train_new_np.shape)
print("Fine-tune validation shape:", X_val_new_np.shape)
print("Classes in new train:", np.unique(Y_train_new_np))
print("Classes in new validation:", np.unique(Y_val_new_np))

assert X_train_new_np.shape[1:] == (SEQ_LENGTH, 70)
assert X_val_new_np.shape[1:] == (SEQ_LENGTH, 70)
assert len(X_train_new_np) == len(Y_train_new_np)
assert len(X_val_new_np) == len(Y_val_new_np)

# ---------------------------------------------------------
# 3. Calculate class weights from fine-tuning training data.
# ---------------------------------------------------------
present_classes = np.unique(Y_train_new_np)

ft_class_weights = compute_class_weight(
    class_weight="balanced",
    classes=present_classes,
    y=Y_train_new_np,
)

ft_class_weight_dict = {
    int(class_id): float(weight)
    for class_id, weight in zip(present_classes, ft_class_weights)
}

print("Fine-tune class weights:", ft_class_weight_dict)

# ---------------------------------------------------------
# 4. Compile with a small learning rate.
# ---------------------------------------------------------
finetune_model.compile(
    optimizer=tf.keras.optimizers.SGD(
        learning_rate=1e-4,
        momentum=0.9,
    ),
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"],
)

# ---------------------------------------------------------
# 5. Stop according to validation macro F1 if available.
# Otherwise val_loss is usually safer than val_accuracy for
# imbalanced emotion classes.
# ---------------------------------------------------------
es_cb_finetune = callbacks.EarlyStopping(
    monitor="val_loss",
    mode="min",
    patience=5,
    restore_best_weights=True,
)

#automatically makes the learning rate smaller when validation performance stops improving.
reduce_lr_cb = callbacks.ReduceLROnPlateau(
    monitor="val_loss",
    mode="min",
    factor=0.5,
    patience=2,
    min_lr=1e-6,
    verbose=1,
)

# ---------------------------------------------------------
# 6. Fine-tune on new-participant training data.
# ---------------------------------------------------------
history_finetune = finetune_model.fit(
    X_train_new_np,
    Y_train_new_np,
    validation_data=(X_val_new_np, Y_val_new_np),
    epochs=30,
    batch_size=batch_size,
    class_weight=ft_class_weight_dict,
    callbacks=[
        es_cb_finetune,
        reduce_lr_cb,
    ],
    verbose=True,
)
````

### 9.1.1 Assemble the comparison table

Loads the saved `summary.json` files for each teacher variant. The pipelines used different test-set
names over time (`natural` vs `old_natural`, `combined` vs `all_combined`), so `TEST_SET_ALIASES`
maps them onto one canonical set of names before anything is compared.

````python
import json
import pandas as pd

# Canonical test_set name mapping (different pipelines used different naming)
TEST_SET_ALIASES = {
    "natural": "natural",
    "old_natural": "natural",
    "acted_plus_emojihero": "acted_plus_emojihero",
    "old_acted_plus_EmojiHero": "acted_plus_emojihero",
    "combined": "combined",
    "all_combined": "combined",
}

rows = []

def add_rows_from_results_by_testset(model_name, summary):
    for test_name, metrics in summary["results_by_testset"].items():
        canon = TEST_SET_ALIASES.get(test_name, test_name)
        row = {"model": model_name, "test_set": canon}
        for k, v in metrics.items():
            if isinstance(v, (int, float)):
                row[k] = v
        rows.append(row)

def add_rows_from_scalar_metrics_list(summary):
    for entry in summary["scalar_metrics"]:
        canon = TEST_SET_ALIASES.get(entry["test_set"], entry["test_set"])
        row = {"model": entry["model"], "test_set": canon}
        for k, v in entry.items():
            if k in ("model", "test_set"):
                continue
            if isinstance(v, (int, float)):
                row[k] = v
        rows.append(row)

# original_teacher and teacher_plus_10participants both come from the rich,
# up-to-date comparison file (correct test-set sizes, full metric suite).
rich_path = Path(predictions_path) / "teacher_model_comparison" / "summary.json"
if rich_path.exists():
    with open(rich_path) as f:
        rich_summary = json.load(f)
    add_rows_from_scalar_metrics_list(rich_summary)
else:
    print(f"WARNING: {rich_path} not found.")

# Fine-tuned model: separate summary.json using results_by_testset schema
finetuned_path = Path(predictions_path) / "teacher_70AU_finetuned_on_10participants" / "summary.json"
if finetuned_path.exists():
    with open(finetuned_path) as f:
        finetuned_summary = json.load(f)
    add_rows_from_results_by_testset("teacher_finetuned_on_10p", finetuned_summary)
else:
    print(f"WARNING: {finetuned_path} not found.")

comparison_df = pd.DataFrame(rows)
print("Models found:", sorted(comparison_df["model"].unique()))
print("Test sets found:", sorted(comparison_df["test_set"].unique()))
print("Available metric columns:", [c for c in comparison_df.columns if c not in ("model", "test_set")])
print()

comparison_out_dir = Path(predictions_path) / "three_way_teacher_comparison"
comparison_out_dir.mkdir(parents=True, exist_ok=True)
comparison_df.to_csv(comparison_out_dir / "comparison_long_all_metrics.csv", index=False)

metric_cols = [c for c in comparison_df.columns if c not in ("model", "test_set", "n_samples")]
for metric in metric_cols:
    pivot = comparison_df.pivot(index="test_set", columns="model", values=metric)
    print(f"=== {metric} ===")
    print(pivot.round(4))
    print()
    pivot.to_csv(comparison_out_dir / f"{metric}_pivot.csv")

print("Saved all comparison files to:", comparison_out_dir)
````

### 9.1.2 Metric-by-metric comparison with win counts

Pivots every metric across models and test sets, marks the best model per row, and tallies wins.
`log_loss` is the only metric where lower is better.

````python
def print_three_way_comparison(comparison_df, models_order=None, lower_is_better=("log_loss",)):
    """
    Pretty-print a model x test_set x metric comparison with a winner annotation per row.
    """
    if models_order is None:
        models_order = sorted(comparison_df["model"].unique())

    metric_cols = [c for c in comparison_df.columns if c not in ("model", "test_set", "n_samples")]
    test_sets = sorted(comparison_df["test_set"].unique())

    win_counts = {m: 0 for m in models_order}

    for metric in metric_cols:
        print("=" * 100)
        print(f"METRIC: {metric}")
        print("=" * 100)
        pivot = comparison_df.pivot(index="test_set", columns="model", values=metric)[models_order]

        for test_set in test_sets:
            row = pivot.loc[test_set]
            higher_better = metric not in lower_is_better
            best_model = row.idxmax() if higher_better else row.idxmin()
            win_counts[best_model] += 1

            vals_str = "  ".join(f"{m}={row[m]:.4f}" for m in models_order)
            print(f"  [{test_set:22s}] {vals_str}   -> best: {best_model}")
        print()

    print("=" * 100)
    print("OVERALL WIN COUNT (across all metric x test_set combinations)")
    print("=" * 100)
    total = sum(win_counts.values())
    for m in sorted(win_counts, key=win_counts.get, reverse=True):
        pct = 100 * win_counts[m] / total
        print(f"  {m:35s}: {win_counts[m]:3d} wins  ({pct:.1f}%)")
    print()

    overall_winner = max(win_counts, key=win_counts.get)
    print(f"VERDICT: '{overall_winner}' wins the most metric/test_set combinations overall.")
    return win_counts


win_counts = print_three_way_comparison(
    comparison_df,
    models_order=["original_teacher", "teacher_finetuned_on_10p", "teacher_plus_10participants"],
)
````

### 9.1.3 Per-class F1 comparison

Recomputes per-class metrics from the saved raw predictions rather than re-running the models, so
the numbers are exactly the ones that were reported.

````python
import numpy as np
from sklearn.metrics import classification_report

class_names = [emo.value for emo in Emotion]
class_labels = list(range(len(Emotion)))

NPZ_PATHS = {
    "original_teacher": Path(predictions_path) / "teacher_model_comparison" / "raw_predictions",
    "teacher_plus_10participants": Path(predictions_path) / "teacher_model_comparison" / "raw_predictions",
    "teacher_finetuned_on_10p": Path(predictions_path) / "teacher_70AU_finetuned_on_10participants" / "raw_predictions",
}

TEST_NAME_FOR_MODEL = {
    "original_teacher": {
        "natural": "original_teacher__old_natural.npz",
        "acted_plus_emojihero": "original_teacher__old_acted_plus_EmojiHero.npz",
        "combined": "original_teacher__all_combined.npz",
    },
    "teacher_plus_10participants": {
        "natural": "teacher_plus_10participants__old_natural.npz",
        "acted_plus_emojihero": "teacher_plus_10participants__old_acted_plus_EmojiHero.npz",
        "combined": "teacher_plus_10participants__all_combined.npz",
    },
    "teacher_finetuned_on_10p": {
        "natural": "teacher_finetuned_on_10p__old_natural.npz",
        "acted_plus_emojihero": "teacher_finetuned_on_10p__old_acted_plus_EmojiHero.npz",
        "combined": "teacher_finetuned_on_10p__all_combined.npz",
    },
}

per_class_rows = []

for model_name, file_map in TEST_NAME_FOR_MODEL.items():
    base_dir = NPZ_PATHS[model_name]
    for test_set, fname in file_map.items():
        fpath = base_dir / fname
        if not fpath.exists():
            print(f"WARNING: {fpath} not found, skipping.")
            continue
        data = np.load(fpath)
        y_true = data["y_true"]
        y_pred = data["y_pred"]

        report = classification_report(
            y_true, y_pred,
            labels=class_labels, target_names=class_names,
            output_dict=True, zero_division=0,
        )
        for cls in class_names:
            per_class_rows.append({
                "model": model_name,
                "test_set": test_set,
                "class": cls,
                "precision": report[cls]["precision"],
                "recall": report[cls]["recall"],
                "f1": report[cls]["f1-score"],
                "support": report[cls]["support"],
            })

per_class_df = pd.DataFrame(per_class_rows)

per_class_out_dir = Path(predictions_path) / "three_way_teacher_comparison"
per_class_out_dir.mkdir(parents=True, exist_ok=True)
per_class_df.to_csv(per_class_out_dir / "per_class_f1_comparison.csv", index=False)

# Per-class F1 pivot, one table per test_set
for test_set in per_class_df["test_set"].unique():
    print("=" * 90)
    print(f"PER-CLASS F1 — test_set: {test_set}")
    print("=" * 90)
    sub = per_class_df[per_class_df["test_set"] == test_set]
    pivot = sub.pivot(index="class", columns="model", values="f1")[
        ["original_teacher", "teacher_finetuned_on_10p", "teacher_plus_10participants"]
    ]
    pivot = pivot.loc[class_names]  # keep emotion order consistent
    print(pivot.round(4))
    print()

print("Saved:", per_class_out_dir / "per_class_f1_comparison.csv")
````

### 9.1.4 Focused comparison: macro-F1, accuracy, weighted F1

Restricts the comparison to the three headline metrics and prints the original -> fine-tuned delta.

Macro-F1 is treated as the primary metric throughout: accuracy and weighted F1 are sample-weighted
and are inflated by the majority classes, which is misleading under this level of class imbalance.

````python
import json
import numpy as np
import pandas as pd
from pathlib import Path

# Load comparison data
with open(Path(predictions_path) / "teacher_model_comparison" / "summary.json") as f:
    rich_summary = json.load(f)

with open(Path(predictions_path) / "teacher_70AU_finetuned_on_10participants-7train_1val" / "summary.json") as f:
    ft_summary = json.load(f)

TEST_SET_ALIASES = {
    "old_natural": "natural",
    "old_acted_plus_EmojiHero": "acted_plus_emojihero",
    "all_combined": "combined",
    "natural": "natural",
    "acted_plus_emojihero": "acted_plus_emojihero",
    "combined": "combined",
}

# Build unified dict: metrics[metric][test_set][model] = value
rows = []

for entry in rich_summary["scalar_metrics"]:
    ts = TEST_SET_ALIASES.get(entry["test_set"], entry["test_set"])
    model = entry["model"]  # "original_teacher" or "teacher_plus_10participants"
    for k, v in entry.items():
        if isinstance(v, (int, float)) and k not in ("n_samples",):
            rows.append({"model": model, "test_set": ts, "metric": k, "value": v})

for ts_raw, metrics in ft_summary["results_by_testset"].items():
    ts = TEST_SET_ALIASES.get(ts_raw, ts_raw)
    for k, v in metrics.items():
        if isinstance(v, (int, float)) and k not in ("n_samples",):
            rows.append({"model": "teacher_finetuned_on_10p", "test_set": ts, "metric": k, "value": v})

df = pd.DataFrame(rows)

# Focus metrics and model order
focus_metrics = ["f1_macro", "accuracy", "f1_weighted"]
model_order   = ["original_teacher", "teacher_finetuned_on_10p", "teacher_plus_10participants"]
model_short   = {"original_teacher": "OT", "teacher_finetuned_on_10p": "FT", "teacher_plus_10participants": "P10"}
test_sets     = ["natural", "acted_plus_emojihero", "combined"]

win_counts = {m: 0 for m in model_order}
total_comparisons = 0

for metric in focus_metrics:
    print("=" * 78)
    print(f"METRIC: {metric}")
    print("=" * 78)
    print(f"  {'test_set':<25} {'OT':>7} {'FT':>7} {'P10':>7}  {'best':>5}  {'OT→FT':>8}")
    print("-" * 78)

    for ts in test_sets:
        vals = {}
        for m in model_order:
            row = df[(df["model"] == m) & (df["test_set"] == ts) & (df["metric"] == metric)]
            vals[m] = row["value"].values[0] if len(row) > 0 else float("nan")

        best_model = max(model_order, key=lambda m: vals[m])
        win_counts[best_model] += 1
        total_comparisons += 1

        ot_ft_diff = vals["teacher_finetuned_on_10p"] - vals["original_teacher"]
        best_short = model_short[best_model]

        print(f"  {ts:<25} "
              f"{vals['original_teacher']:>7.4f} "
              f"{vals['teacher_finetuned_on_10p']:>7.4f} "
              f"{vals['teacher_plus_10participants']:>7.4f}  "
              f"{best_short:>5}  "
              f"{ot_ft_diff:>+8.4f}")
    print()

print("=" * 78)
print(f"WIN COUNT ({total_comparisons} comparisons: {len(focus_metrics)} metrics x {len(test_sets)} test sets)")
print("=" * 78)
for m in sorted(model_order, key=lambda x: win_counts[x], reverse=True):
    pct = 100 * win_counts[m] / total_comparisons
    print(f"  {model_short[m]} ({m:<35}): {win_counts[m]:2d} wins ({pct:.0f}%)")

print()
print("=" * 78)
print("VERDICT")
print("=" * 78)
overall_best = max(model_order, key=lambda m: win_counts[m])
print(f"  Overall best: {overall_best} ({model_short[overall_best]})")
print()
print("  Note: OT=original_teacher, FT=teacher_finetuned_on_10p, P10=teacher_plus_10participants")
print("  f1_macro   = treats all 7 emotions equally (recommended primary metric)")
print("  accuracy   = sample-weighted, favors majority classes (Happiness/Disgust)")
print("  f1_weighted= sample-weighted F1, also biased toward majority classes")
print("  -> f1_macro is the most honest metric for imbalanced emotion recognition.")
````

# Part 9: Teacher Variant Comparisons

## 9.1 Five-Way Comparison

Extends the comparison to every teacher variant that was trained:

| Key | Variant |
|---|---|
| OT | Original teacher (no new data) |
| P10 | Original + 10 participants, naive joint training |
| FT | Original + 10 participants, fine-tuned |
| RE | Original + 10 participants, self-report filtered |
| RE70T10V20T | Original + 10 participants, self-report filtered, 70/10/20 split |

### 9.1. Configuration and loading

````python
import json
import numpy as np
import pandas as pd
from pathlib import Path
from sklearn.metrics import classification_report

MODEL_ORDER = [
    "original_teacher",
    "plus_10participants",
    "finetuned_on_10p",
    "removed_emotions",
    "removed_emotions_70train_10val_20test_ratio",
]

MODEL_SHORT = {
    "original_teacher":   "OT",
    "plus_10participants": "P10",
    "finetuned_on_10p":   "FT",
    "removed_emotions":   "RE",
    "removed_emotions_70train_10val_20test_ratio": "RE70T10V20T",
}

MODEL_LABELS = {
    "original_teacher":   "Original Teacher (no new data)",
    "plus_10participants": "Original + 10p naive joint training",
    "finetuned_on_10p":   "Original + 10p fine-tuned",
    "removed_emotions":   "Original + 10p self-report filtered",
    "removed_emotions_70train_10val_20test_ratio": "Original + 10p self-report filtered (70:10:20)",
}

TEST_SET_ALIASES = {
    "old_natural":              "natural",
    "old_acted_plus_EmojiHero": "acted_plus_emojihero",
    "all_combined":             "combined",
    "natural":                  "natural",
    "acted_plus_emojihero":     "acted_plus_emojihero",
    "combined":                 "combined",
}

LOWER_IS_BETTER = {"log_loss"}
TEST_SETS     = ["natural", "acted_plus_emojihero", "combined"]
EMOTIONS      = ['Anger','Disgust','Fear','Happiness','Neutral','Sadness','Surprise']
FOCUS_METRICS = [
    "f1_macro", "accuracy", "f1_weighted", "balanced_accuracy",
    "roc_auc_macro_ovr", "log_loss", "matthews_corrcoef", "top2_accuracy",
]

PRED = Path(predictions_path)

rows = []

def add_from_scalar_metrics_list(data):
    for entry in data["scalar_metrics"]:
        ts = TEST_SET_ALIASES.get(entry["test_set"], entry["test_set"])
        m  = "plus_10participants" if entry["model"] == "teacher_plus_10participants" else entry["model"]
        row = {"model": m, "test_set": ts}
        for k, v in entry.items():
            if isinstance(v, (int, float)) and k != "n_samples":
                row[k] = v
        rows.append(row)

def add_from_results_by_testset(data, model_key):
    for ts_raw, metrics in data["results_by_testset"].items():
        ts = TEST_SET_ALIASES.get(ts_raw, ts_raw)
        row = {"model": model_key, "test_set": ts}
        for k, v in metrics.items():
            if isinstance(v, (int, float)) and k != "n_samples":
                row[k] = v
        rows.append(row)

# original_teacher + plus_10participants
p = Path("/content/drive/MyDrive/outputs_10participants_70AU/predictions/teacher_model_comparison/summary.json")
if p.exists():
    with open(p) as f:
        add_from_scalar_metrics_list(json.load(f))
else:
    print(f"WARNING: not found: {p}")

# finetuned_on_10p
p = Path("/content/drive/MyDrive/outputs_10participants_70AU/predictions/teacher_70AU_finetuned_on_10participants/summary.json")
if p.exists():
    with open(p) as f:
        add_from_results_by_testset(json.load(f), "finetuned_on_10p")
else:
    print(f"WARNING: not found: {p}")

p = Path("/content/drive/MyDrive/outputs_10participants_70AU-removed_emotions/predictions/teacher_with_10participants_70AU-removed_emotions-7train_1val/results_all_combined/summary.json")
if p.exists():
    with open(p) as f:
        add_from_results_by_testset(json.load(f), "removed_emotions")
else:
    print(f"WARNING: not found: {p}")

p = Path("/content/drive/MyDrive/outputs_10participants_70AU-removed_emotions/predictions/teacher_with_10participants_70AU-removed_emotions-5train_1val_2test/results_all_combined/summary.json")
if p.exists():
    with open(p) as f:
        add_from_results_by_testset(json.load(f), "removed_emotions_70train_10val_20test_ratio")
else:
    print(f"WARNING: not found: {p}")


compare_df = pd.DataFrame(rows)
print("Models loaded:", sorted(compare_df["model"].unique()))
print("Test sets:    ", sorted(compare_df["test_set"].unique()))
print("Metrics:      ", [c for c in compare_df.columns if c not in ("model", "test_set")])
````

## 9.2 Scalar metric comparison and win counts

````python
win_counts = {m: 0 for m in MODEL_ORDER}
total_comparisons = 0

print("=" * 115)
print("TEACHER MODELS COMPARISON — SCALAR METRICS")
print("=" * 115)
for m in MODEL_ORDER:
    print(f"  {MODEL_SHORT[m]:<6} = {MODEL_LABELS[m]}")
print("=" * 115)

for metric in FOCUS_METRICS:
    if metric not in compare_df.columns:
        continue
    arrow = " ↓ lower=better" if metric in LOWER_IS_BETTER else ""
    print(f"\n── {metric}{arrow}")
    print(f"  {'test_set':<25} " + "  ".join(f"{MODEL_SHORT[m]:>8}" for m in MODEL_ORDER) + "   best")
    print(f"  {'-'*25} " + "  ".join("--------" for _ in MODEL_ORDER))

    for ts in TEST_SETS:
        vals = {}
        for m in MODEL_ORDER:
            row = compare_df[(compare_df["model"] == m) & (compare_df["test_set"] == ts)]
            vals[m] = float(row[metric].values[0]) if len(row) > 0 and not row[metric].isna().all() else float("nan")

        valid = {m: v for m, v in vals.items() if not np.isnan(v)}
        best  = min(valid, key=valid.get) if metric in LOWER_IS_BETTER else max(valid, key=valid.get)
        win_counts[best] += 1
        total_comparisons += 1

        val_str = "  ".join(f"{vals[m]:>8.4f}" if not np.isnan(vals[m]) else f"{'N/A':>8}" for m in MODEL_ORDER)
        print(f"  {ts:<25} {val_str}   {MODEL_SHORT[best]}")

print()
print("=" * 115)
print(f"WIN COUNT ({total_comparisons} comparisons: {len(FOCUS_METRICS)} metrics × {len(TEST_SETS)} test sets)")
print("=" * 115)
for m in sorted(MODEL_ORDER, key=lambda x: win_counts[x], reverse=True):
    pct = 100 * win_counts[m] / total_comparisons if total_comparisons > 0 else 0
    bar = "█" * win_counts[m]
    print(f"  {MODEL_SHORT[m]:<6} ({MODEL_LABELS[m]:<45}): {win_counts[m]:3d} wins ({pct:.0f}%)  {bar}")

best_overall = max(MODEL_ORDER, key=lambda m: win_counts[m])
print(f"\n  VERDICT: {MODEL_SHORT[best_overall]} — {MODEL_LABELS[best_overall]}")
````

````python
RAW_NPZ = {
    "original_teacher":   Path("/content/drive/MyDrive/outputs_10participants_70AU/predictions") / "teacher_model_comparison" / "raw_predictions",
    "plus_10participants": Path("/content/drive/MyDrive/outputs_10participants_70AU/predictions") / "teacher_model_comparison" / "raw_predictions",
    "finetuned_on_10p":   Path("/content/drive/MyDrive/outputs_10participants_70AU/predictions") / "teacher_70AU_finetuned_on_10participants" / "raw_predictions",
    "removed_emotions":   Path("/content/drive/MyDrive/outputs_10participants_70AU-removed_emotions/predictions/teacher_with_10participants_70AU-removed_emotions-7train_1val/results_all_combined/raw_predictions"),
    "removed_emotions_70train_10val_20test_ratio": Path("/content/drive/MyDrive/outputs_10participants_70AU-removed_emotions/predictions/teacher_with_10participants_70AU-removed_emotions-5train_1val_2test/results_all_combined/raw_predictions")
}

NPZ_FILES = {
    "original_teacher":   {
        "natural":              "original_teacher__old_natural.npz",
        "acted_plus_emojihero": "original_teacher__old_acted_plus_EmojiHero.npz",
        "combined":             "original_teacher__all_combined.npz",
    },
    "plus_10participants": {
        "natural":              "teacher_plus_10participants__old_natural.npz",
        "acted_plus_emojihero": "teacher_plus_10participants__old_acted_plus_EmojiHero.npz",
        "combined":             "teacher_plus_10participants__all_combined.npz",
    },
    "finetuned_on_10p": {
        "natural":              "teacher_finetuned_on_10p__old_natural.npz",
        "acted_plus_emojihero": "teacher_finetuned_on_10p__old_acted_plus_EmojiHero.npz",
        "combined":             "teacher_finetuned_on_10p__all_combined.npz",
    },
    "removed_emotions": {
        "natural":              "removed_emotions__old_natural.npz",
        "acted_plus_emojihero": "removed_emotions__old_acted_plus_EmojiHero.npz",
        "combined":             "removed_emotions__all_combined.npz",
    },
    "removed_emotions_70train_10val_20test_ratio": {
        "natural":              "removed_emotions__old_natural.npz",
        "acted_plus_emojihero": "removed_emotions__old_acted_plus_EmojiHero.npz",
        "combined":             "removed_emotions__all_combined.npz",
    },
}

pc_rows = []
for model_key, file_map in NPZ_FILES.items():
    base = RAW_NPZ[model_key]
    for ts, fname in file_map.items():
        fpath = base / fname
        if not fpath.exists():
            print(f"WARNING: {fpath} not found, skipping.")
            continue
        data = np.load(fpath)
        report = classification_report(
            data["y_true"], data["y_pred"],
            labels=list(range(len(EMOTIONS))),
            target_names=EMOTIONS,
            output_dict=True, zero_division=0,
        )
        for cls in EMOTIONS:
            pc_rows.append({
                "model": model_key, "test_set": ts,
                "class": cls, "f1": report[cls]["f1-score"],
            })

pc_df = pd.DataFrame(pc_rows)

for ts in TEST_SETS:
    print("=" * 95)
    print(f"PER-CLASS F1 — {ts.upper()}")
    print("=" * 95)
    sub = pc_df[pc_df["test_set"] == ts]
    if sub.empty:
        print("  No data.")
        continue
    avail = [m for m in MODEL_ORDER if m in sub["model"].unique()]
    pivot = sub.pivot(index="class", columns="model", values="f1")[avail]
    pivot = pivot.reindex(EMOTIONS)
    pivot.columns = [MODEL_SHORT[m] for m in avail]
    pivot["best"] = pivot.idxmax(axis=1)
    print(pivot.round(4).to_string())
    print()
````

# Part 10: Teacher Soft Labels for Distillation

This is the hand-off point to the student notebook. The trained teacher predicts a full
class-probability vector for every train / validation / test window; those vectors are the soft
targets used by the RQ2 distillation objectives.

Soft labels and hard labels are saved together so that student and teacher stay index-aligned.

## 10.1 Generate Teacher Soft Labels

Runs the trained teacher over the train, validation, and test windows and keeps the full softmax
output — not the argmax. The relative probability mass across the non-predicted classes is precisely
the "dark knowledge" the student is meant to inherit.

**Note:** this requires `trained_model` to still be in memory from Section 7.3.

````python
teacher_soft_train = trained_model.predict(X_train_np, batch_size=64)
teacher_soft_val = trained_model.predict(X_val_np, batch_size=64)
teacher_soft_test = trained_model.predict(X_test_comb_np, batch_size=64)
````

# Part 11: Original vs. +10-Participant Teacher (head-to-head)

A controlled comparison of the two saved teacher checkpoints on the same three test sets, using the
full metric suite. This section also defines `compute_full_metrics()`, which earlier sections reuse.

## 11.1 Checkpoint Paths

The two teacher checkpoints being compared, plus the output directory for this section.

````python
from pathlib import Path

# Edit these two paths if your saved model filenames are different.
ORIGINAL_TEACHER_MODEL_PATH = Path("/content/drive/MyDrive/outputs_10participants_70AU/models/teacher-70AU-acted-natural-MoE_3experts_2gate-without_10participants.keras")
TEN_PARTICIPANT_TEACHER_MODEL_PATH = Path("/content/drive/MyDrive/outputs_10participants_70AU/models/teacher-70AU-acted-natural-MoE_3experts_2gate_with_10participants_data.keras")

# Save comparison outputs here.
output_path = Path("/content/drive/MyDrive/outputs_10participants_70AU")
predictions_path = output_path / "predictions"
comparison_path = predictions_path / "teacher_model_comparison"
comparison_path.mkdir(parents=True, exist_ok=True)

print("Original model:", ORIGINAL_TEACHER_MODEL_PATH, "exists=", ORIGINAL_TEACHER_MODEL_PATH.exists())
print("10-participant model:", TEN_PARTICIPANT_TEACHER_MODEL_PATH, "exists=", TEN_PARTICIPANT_TEACHER_MODEL_PATH.exists())
print("Saving comparison outputs to:", comparison_path)
````

## 11.2 Load Both Teachers

Both are loaded by rebuilding the architecture and restoring weights (see the note in Section 8.2).

````python
def load_teacher_model(model_path, input_shape=(SEQ_LENGTH, input_size), num_classes=num_classes):
    # Rebuild the exact same architecture used at training time
    model, _ = build_moe_single_input(
        input_shape=input_shape, num_classes=num_classes,
        tau_init=2.0, tau_final=0.7,
        gate_hidden=128, classifier_hidden=256,
        lb_coef=0.02, ent_coef=0.01,
    )
    # Load only the weights, skip reconstructing the saved Functional graph
    model.load_weights(model_path)
    return model

original_teacher_model = load_teacher_model(ORIGINAL_TEACHER_MODEL_PATH)
teacher_plus_10_model = load_teacher_model(TEN_PARTICIPANT_TEACHER_MODEL_PATH)

MODELS_TO_COMPARE = {
    "original_teacher": original_teacher_model,
    "teacher_plus_10participants": teacher_plus_10_model,
}
````

## 11.3 Full Metric Suite

`compute_full_metrics()` runs a model on one test set and returns everything needed downstream
without re-predicting: scalar metrics (accuracy, balanced accuracy, macro/weighted/micro precision,
recall and F1, Cohen's kappa, MCC, top-2 accuracy, log loss, one-vs-rest ROC-AUC and PR-AUC),
a per-class table, raw and row-normalised confusion matrices, and the predictions and probabilities
themselves.

Metrics that are undefined for a given test set (e.g. ROC-AUC with a single class present) are
skipped with a warning rather than raising.

**Note:** Sections 7.5 and 8.5 call this function, so it must be defined before them.

````python
import json
import numpy as np
import pandas as pd
from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    classification_report,
    cohen_kappa_score,
    confusion_matrix,
    f1_score,
    log_loss,
    matthews_corrcoef,
    precision_score,
    recall_score,
    roc_auc_score,
    average_precision_score,
    top_k_accuracy_score,
)
from sklearn.preprocessing import label_binarize


def compute_full_metrics(model, X, y_true, class_labels, class_names, model_name="model", test_name="test"):
    """
    Run a model on (X, y_true) and compute the full metric suite.
    Returns a dict with scalar metrics, per-class metrics, confusion matrices,
    and the raw predictions/probabilities (so plots can reuse them without re-predicting).
    """
    y_true = np.reshape(np.asarray(y_true), -1)
    y_proba = model.predict(X, batch_size=64, verbose=0)
    y_proba = np.asarray(y_proba)
    y_pred = np.argmax(y_proba, axis=1)

    n_classes = len(class_labels)

    # --- Basic ---
    acc = accuracy_score(y_true, y_pred)
    bal_acc = balanced_accuracy_score(y_true, y_pred)

    # --- Precision/Recall/F1 (aggregate) ---
    prec_macro = precision_score(y_true, y_pred, labels=class_labels, average="macro", zero_division=0)
    rec_macro = recall_score(y_true, y_pred, labels=class_labels, average="macro", zero_division=0)
    f1_macro = f1_score(y_true, y_pred, labels=class_labels, average="macro", zero_division=0)

    prec_weighted = precision_score(y_true, y_pred, labels=class_labels, average="weighted", zero_division=0)
    rec_weighted = recall_score(y_true, y_pred, labels=class_labels, average="weighted", zero_division=0)
    f1_weighted = f1_score(y_true, y_pred, labels=class_labels, average="weighted", zero_division=0)

    prec_micro = precision_score(y_true, y_pred, labels=class_labels, average="micro", zero_division=0)
    rec_micro = recall_score(y_true, y_pred, labels=class_labels, average="micro", zero_division=0)
    f1_micro = f1_score(y_true, y_pred, labels=class_labels, average="micro", zero_division=0)

    # --- Agreement metrics ---
    kappa = cohen_kappa_score(y_true, y_pred, labels=class_labels)
    mcc = matthews_corrcoef(y_true, y_pred)

    # --- Top-2 accuracy ---
    present_labels = np.unique(y_true)
    try:
        top2_acc = top_k_accuracy_score(y_true, y_proba, k=2, labels=class_labels)
    except Exception as e:
        top2_acc = np.nan
        print(f"[{model_name} | {test_name}] top-2 accuracy skipped: {e}")

    # --- Log loss ---
    try:
        logloss = log_loss(y_true, y_proba, labels=class_labels)
    except Exception as e:
        logloss = np.nan
        print(f"[{model_name} | {test_name}] log loss skipped: {e}")

    # --- ROC-AUC and PR-AUC (One-vs-Rest); only well-defined if >=2 classes present in y_true ---
    roc_auc_macro = roc_auc_score_safe = np.nan
    roc_auc_weighted = np.nan
    pr_auc_macro = np.nan
    pr_auc_weighted = np.nan
    if len(present_labels) >= 2:
        y_true_bin = label_binarize(y_true, classes=class_labels)
        try:
            roc_auc_macro = roc_auc_score(y_true_bin, y_proba, average="macro", multi_class="ovr")
            roc_auc_weighted = roc_auc_score(y_true_bin, y_proba, average="weighted", multi_class="ovr")
        except Exception as e:
            print(f"[{model_name} | {test_name}] ROC-AUC skipped: {e}")
        try:
            pr_auc_macro = average_precision_score(y_true_bin, y_proba, average="macro")
            pr_auc_weighted = average_precision_score(y_true_bin, y_proba, average="weighted")
        except Exception as e:
            print(f"[{model_name} | {test_name}] PR-AUC skipped: {e}")
    else:
        print(f"[{model_name} | {test_name}] Only one class present in y_true; ROC-AUC/PR-AUC undefined, skipped.")

    # --- Per-class report ---
    report_dict = classification_report(
        y_true, y_pred, labels=class_labels, target_names=class_names,
        output_dict=True, zero_division=0,
    )
    per_class_df = pd.DataFrame(report_dict).T.loc[class_names]
    per_class_df.index.name = "class"

    # --- Confusion matrices ---
    cm = confusion_matrix(y_true, y_pred, labels=class_labels)
    with np.errstate(all="ignore"):
        cm_norm = cm.astype(np.float64) / cm.sum(axis=1, keepdims=True)
    cm_norm = np.nan_to_num(cm_norm)

    scalar_metrics = {
        "model": model_name,
        "test_set": test_name,
        "n_samples": int(len(y_true)),
        "accuracy": acc,
        "balanced_accuracy": bal_acc,
        "precision_macro": prec_macro,
        "recall_macro": rec_macro,
        "f1_macro": f1_macro,
        "precision_weighted": prec_weighted,
        "recall_weighted": rec_weighted,
        "f1_weighted": f1_weighted,
        "precision_micro": prec_micro,
        "recall_micro": rec_micro,
        "f1_micro": f1_micro,
        "cohen_kappa": kappa,
        "matthews_corrcoef": mcc,
        "top2_accuracy": top2_acc,
        "log_loss": logloss,
        "roc_auc_macro_ovr": roc_auc_macro,
        "roc_auc_weighted_ovr": roc_auc_weighted,
        "pr_auc_macro_ovr": pr_auc_macro,
        "pr_auc_weighted_ovr": pr_auc_weighted,
    }

    return {
        "scalar_metrics": scalar_metrics,
        "per_class_df": per_class_df,
        "confusion_matrix": cm,
        "confusion_matrix_normalized": cm_norm,
        "y_true": y_true,
        "y_pred": y_pred,
        "y_proba": y_proba,
    }


print("compute_full_metrics() defined.")
````

## 11.4 Define the Test Sets

Three fixed test sets — natural, acted + EmojiHeroVR, combined — plus the new-participant test set if
it exists in the current session.

````python
from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    classification_report,
    confusion_matrix,
    f1_score,
    precision_score,
    recall_score,
)

class_names = [emo.value for emo in Emotion]
class_labels = list(range(len(class_names)))

TEST_SETS = {
    "old_natural": (X_test_nat_np, Y_test_nat_np),
    "old_acted_plus_EmojiHero": (X_test_act_comb_np, Y_test_act_comb_np),
    "all_combined": (X_test_comb_np, Y_test_comb_np),
}

if "X_test_new_np" in globals() and "Y_test_new_np" in globals():
    TEST_SETS["new_10participants"] = (X_test_new_np, Y_test_new_np)
elif "X_test_new" in globals() and "Y_test_new" in globals():
    X_test_new_np = stack_sequences(X_test_new)
    Y_test_new_np = to_numpy_labels(Y_test_new)
    TEST_SETS["new_10participants"] = (X_test_new_np, Y_test_new_np)

for test_name, (X, y) in TEST_SETS.items():
    print(test_name, X.shape, y.shape)
````

## 11.5 Evaluate Every (Model, Test Set) Pair

````python
all_results = {}  # all_results[(model_name, test_name)] = result dict from compute_full_metrics

for model_name, model in MODELS_TO_COMPARE.items():
    for test_name, (X, y) in TEST_SETS.items():
        print(f"Evaluating {model_name} on {test_name} ...")
        result = compute_full_metrics(
            model, X, y,
            class_labels=class_labels,
            class_names=class_names,
            model_name=model_name,
            test_name=test_name,
        )
        all_results[(model_name, test_name)] = result
        print(
            f"  acc={result['scalar_metrics']['accuracy']:.4f}  "
            f"bal_acc={result['scalar_metrics']['balanced_accuracy']:.4f}  "
            f"f1_macro={result['scalar_metrics']['f1_macro']:.4f}"
        )

print("\nDone. Collected", len(all_results), "(model, test_set) evaluations.")
````

## 11.6 Scalar Metric Tables

Writes three views of the same results: long format (one row per model/test set), wide format (one
row per metric), and a delta table giving `teacher_plus_10participants - original_teacher` for every
metric on every test set.

````python
scalar_rows = [res["scalar_metrics"] for res in all_results.values()]
scalar_df = pd.DataFrame(scalar_rows).sort_values(["test_set", "model"]).reset_index(drop=True)

display(scalar_df)

scalar_df.to_csv(comparison_path / "scalar_metrics_all_models_all_testsets.csv", index=False)
print("Saved:", comparison_path / "scalar_metrics_all_models_all_testsets.csv")

# ---- Wide format: one row per metric, one column per (model, test_set) for easy side-by-side reading ----
wide_df = scalar_df.set_index(["test_set", "model"]).drop(columns=["n_samples"]).T
wide_df.to_csv(comparison_path / "scalar_metrics_wide.csv")
display(wide_df)

# ---- Delta table: original vs +10participants, per test set, per metric ----
model_a, model_b = "original_teacher", "teacher_plus_10participants"
metric_cols = [c for c in scalar_df.columns if c not in ("model", "test_set", "n_samples")]

delta_rows = []
for test_name in TEST_SETS.keys():
    row_a = scalar_df[(scalar_df.model == model_a) & (scalar_df.test_set == test_name)]
    row_b = scalar_df[(scalar_df.model == model_b) & (scalar_df.test_set == test_name)]
    if row_a.empty or row_b.empty:
        continue
    row_a = row_a.iloc[0]
    row_b = row_b.iloc[0]
    for metric in metric_cols:
        delta_rows.append({
            "test_set": test_name,
            "metric": metric,
            model_a: row_a[metric],
            model_b: row_b[metric],
            "delta (b - a)": row_b[metric] - row_a[metric],
        })

delta_df = pd.DataFrame(delta_rows)
delta_df.to_csv(comparison_path / "metric_deltas_original_vs_10participants.csv", index=False)
print("\nSaved:", comparison_path / "metric_deltas_original_vs_10participants.csv")
display(delta_df)
````

## 11.7 Per-Class Metric Tables

Per-class precision, recall, F1, and support for both models, plus a side-by-side per-class F1 pivot
with deltas. This is where a change in overall macro-F1 can be traced to the specific emotions that
moved.

````python
per_class_rows = []
for (model_name, test_name), res in all_results.items():
    df = res["per_class_df"].copy()
    df = df.reset_index().rename(columns={"index": "class"})
    df["model"] = model_name
    df["test_set"] = test_name
    per_class_rows.append(df)

per_class_all_df = pd.concat(per_class_rows, ignore_index=True)
per_class_all_df = per_class_all_df[["test_set", "class", "model", "precision", "recall", "f1-score", "support"]]
per_class_all_df = per_class_all_df.sort_values(["test_set", "class", "model"]).reset_index(drop=True)

per_class_all_df.to_csv(comparison_path / "per_class_metrics_all_models_all_testsets.csv", index=False)
print("Saved:", comparison_path / "per_class_metrics_all_models_all_testsets.csv")
display(per_class_all_df)

# Pivot: per-class F1, side by side for the two models, one table per test set
for test_name in TEST_SETS.keys():
    sub = per_class_all_df[per_class_all_df.test_set == test_name]
    pivot = sub.pivot(index="class", columns="model", values="f1-score")
    if model_a in pivot.columns and model_b in pivot.columns:
        pivot["delta (b - a)"] = pivot[model_b] - pivot[model_a]
    print(f"\n--- Per-class F1 | test_set = {test_name} ---")
    display(pivot)
````

## 11.8 Confusion Matrices

Raw and row-normalised confusion matrices for both models on each test set, plotted side by side and
saved as PNG. Normalised matrices are read row-wise: what fraction of each true emotion went where.

````python
import matplotlib.pyplot as plt
import seaborn as sns

def plot_confusion_pair(test_name, normalize=False):
    res_a = all_results[(model_a, test_name)]
    res_b = all_results[(model_b, test_name)]

    cm_a = res_a["confusion_matrix_normalized"] if normalize else res_a["confusion_matrix"]
    cm_b = res_b["confusion_matrix_normalized"] if normalize else res_b["confusion_matrix"]

    fmt = ".2f" if normalize else "d"
    vmax = max(cm_a.max(), cm_b.max())

    fig, axes = plt.subplots(1, 2, figsize=(14, 6))
    sns.heatmap(cm_a, annot=True, fmt=fmt, cmap="Blues", vmin=0, vmax=vmax,
                xticklabels=class_names, yticklabels=class_names, ax=axes[0], cbar=False)
    axes[0].set_title(f"{model_a}\n{test_name}")
    axes[0].set_xlabel("Predicted")
    axes[0].set_ylabel("True")

    sns.heatmap(cm_b, annot=True, fmt=fmt, cmap="Blues", vmin=0, vmax=vmax,
                xticklabels=class_names, yticklabels=class_names, ax=axes[1], cbar=False)
    axes[1].set_title(f"{model_b}\n{test_name}")
    axes[1].set_xlabel("Predicted")
    axes[1].set_ylabel("True")

    plt.tight_layout()
    suffix = "normalized" if normalize else "raw"
    fname = comparison_path / f"confusion_matrix_{suffix}_{test_name}.png"
    plt.savefig(fname, dpi=150, bbox_inches="tight")
    plt.show()
    print("Saved:", fname)


for test_name in TEST_SETS.keys():
    plot_confusion_pair(test_name, normalize=False)
    plot_confusion_pair(test_name, normalize=True)
````

## 11.11 Save Raw Predictions and Final Verdict

Persists every model's predictions and probabilities as `.npz` so that any metric can be recomputed
later without re-running the models, writes the scalar summary as JSON, and prints a per-metric
winner for each test set.

````python
# ---- Save raw predictions/probabilities for every (model, test_set) so results are reproducible later ----
raw_dir = comparison_path / "raw_predictions"
raw_dir.mkdir(exist_ok=True)

for (model_name, test_name), res in all_results.items():
    np.savez(
        raw_dir / f"{model_name}__{test_name}.npz",
        y_true=res["y_true"],
        y_pred=res["y_pred"],
        y_proba=res["y_proba"],
        confusion_matrix=res["confusion_matrix"],
    )

print("Saved raw predictions/probabilities to:", raw_dir)

# ---- JSON summary (scalar metrics only, easy to diff or re-load) ----
summary_json = {
    "class_names": class_names,
    "models_compared": list(MODELS_TO_COMPARE.keys()),
    "test_sets": {name: {"n_samples": int(len(y))} for name, (X, y) in TEST_SETS.items()},
    "scalar_metrics": scalar_df.to_dict(orient="records"),
}
with open(comparison_path / "summary.json", "w") as f:
    json.dump(summary_json, f, indent=2, default=float)
print("Saved:", comparison_path / "summary.json")

# ---- Final printed summary: which model wins on which metric, per test set ----
print("\n" + "=" * 70)
print("FINAL SUMMARY:", model_a, "vs", model_b)
print("=" * 70)
for test_name in TEST_SETS.keys():
    row_a = scalar_df[(scalar_df.model == model_a) & (scalar_df.test_set == test_name)].iloc[0]
    row_b = scalar_df[(scalar_df.model == model_b) & (scalar_df.test_set == test_name)].iloc[0]
    print(f"\n[{test_name}]  (n={int(row_a['n_samples'])})")
    for metric in key_metrics:
        a_val, b_val = row_a[metric], row_b[metric]
        winner = model_b if b_val > a_val else (model_a if a_val > b_val else "tie")
        print(f"  {metric:22s} {model_a}={a_val:.4f}  {model_b}={b_val:.4f}   -> better: {winner}")

print("\nAll comparison artifacts (CSV/JSON/PNG/raw predictions) saved under:")
print(" ", comparison_path)
````

### Prediction distribution check

Counts how many windows each model assigns to each emotion. A model that collapses onto one or two
majority classes will show it immediately here, even when its accuracy looks acceptable.

````python
import numpy as np

ORIGINAL_NO10_PATH = "/content/drive/MyDrive/outputs_10participants_70AU/models/teacher-70AU-acted-natural-MoE_3experts_2gate-without_10participants.keras"
WITH10_PATH = "/content/drive/MyDrive/outputs_10participants_70AU/models/teacher-70AU-acted-natural-MoE_3experts_2gate_with_10participants_data.keras"

model_no10 = load_teacher_model(ORIGINAL_NO10_PATH, input_shape=(SEQ_LENGTH, 70))
model_with10 = load_teacher_model(WITH10_PATH, input_shape=(SEQ_LENGTH, 70))

emo_names = [emo.value for emo in Emotion]

for name, m in [("original_teacher (no 10p)", model_no10), ("teacher_plus_10participants", model_with10)]:
    preds = m.predict(X_test_comb_np)
    pred_labels = np.argmax(preds, axis=1)
    counts = np.bincount(pred_labels, minlength=len(Emotion))
    print(f"\n{name}")
    for emo, c in zip(emo_names, counts):
        print(f"  {emo:12s}: {c}")
    print(f"  total: {counts.sum()}")
````
